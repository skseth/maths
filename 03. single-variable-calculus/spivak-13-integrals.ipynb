# Integral

**partition**

Let a < b. A partition of the interval $[a,b]$ is a finite collection of points in $[a,b]$, one of which is a, and one of which is b.

$a = t_0 < t_1 < t_2 < \dots < t_{n-1} < t_n < b$



## Darboux Sums

Suppose f is bounded on $[a,b]$ and $P = \{t_0, t_1, \dots, t_{n-1}, t_n \}$

$m_i = inf\{ f(x): t_{i-1} \le x \le t_i \}$

$M_i = sup\{ f(x): t_{i-1} \le x \le t_i \}$

**lower sum**

$L(f,p) = \sum\limits_{i=1}^{n} m_i(t_i - t_{i-1})$

**upper sum**

$L(f,p) = \sum\limits_{i=1}^{n} M_i(t_i - t_{i-1})$

NOTE: Not from Spivak: Riemann's original sum was based on using any point within the interval. Darboux made this rigorous using upper / lower sums. 


## Theorem $L(f,P_1) \le U(f,P_2)$ for any two partitions $P_1$ and $P_2$

**Lemma: If P and Q are partitions over the same interval, $P \subset Q$, then**

$$L(f, P) \le L(f, Q)$$

$$U(f, P) \ge U(f, Q)$$

Proof: 

Consider the case where Q contains exactly one more point that P: 

$$P = \{t_0, t_1, \dots, t_{k-1}, t_{k}, \dots, t_n\}, Q = \{t_0, t_1, \dots, t_{k-1}, u, t_{k}, \dots, t_n\} $$

Then, we see that question is just about the interval of $(t_{k-1},t_k)$

Let $m' = inf\{f(x): t_{k-1} \le x \le u\}$, $m'' = inf\{f(x): t_{u} \le x \le t_k\}$, $m_k = \min{m',m''}$

Let $M' = sup\{f(x): t_{k-1} \le x \le u\}$, $M'' = sup\{f(x): t_{u} \le x \le t_k\}$, $M_k = \max{M',M''}$

The : 
$$
\begin{aligned}
L(f,Q) - L(f,P) &= m'(t_u - t_{k-1}) + m''(t_{k-1} - t_u) -  m_k(t_k - t_{k-1}) \\
        &\ge m_k(t_u - t_{k-1}) - m_k(t_{k-1} - t_u) = 0
\end{aligned}
$$


And, similarly:

$$
\begin{aligned}
U(f,P) - U(f,Q) &= M_k(t_k - t_{k-1}) - M'(t_u - t_{k-1}) - M''(t_k - t_u) \\
        &\ge M_k(t_k - t_{k-1}) - M_k(t_u - t_{k-1}) - M_k(t_k - t_u) = 0 \\
\end{aligned}
$$

Now, given any general partition Q, we can generate a sequence of partitions $P = P_1, P_2, \dots, P_{\alpha} = Q$, where $P_j+1$ contains just one more point that $P_j$.

Then we have $L(f,P) \le L(f,P_1) \le L(f,P_2) \dots L(f,Q)$, and $U(f,P) \ge U(f,P_1) \le U(f,P_2) \dots U(f,Q)$

**Proof of theorem**

Let $P_1$ and $P_2$ be any partitions of $[a,b]$. Let partition P be formed of all points in $P_1$ and $P_2$. Then, based on the lemma:

$L(f,P_1) \le L(f,P) \le U(f,P) \le U(f,P_2)$

**Definition (Integral)**

Let f be a function bounded on the interval $[a,b]$.

Let sup{L(f,P): P is a partition on $[a,b]$} = inf{U(f,P): P is a partition on $[a,b]$}. Then f is said to be **integrable** and common number is referred to by :

$$\int\limits_a^b f$$

a,b are referred to as the **lower limit** and **upper limit** of integration.

## Theorem 2 ($\epsilon \text{ form of integration}$)

If f is bounded on $[a,b]$, then f is integrable on $[a,b]$ iff for every $\epsilon > 0$, there is a partition P such that $U(f,P) - L(f,P) \le \epsilon$


Proof:

a) => (f is integrable on $[a,b]$): Then we have sup{L(f,P): P is a partition on $[a,b]$} = inf{U(f,P): P is a partition on $[a,b]$} = I, say. Given any $\epsilon$ we have a $\delta_1$ in the interval $(I - \delta_1, I)$, such that there is a partition $P_1$ such that $I - L(f,P_1) \lt \epsilon/2$, and similarly we can have a $\delta_2$ such that there is a partition $P_2$, such that $U(f,P_1) - I \lt \epsilon/2$. Then we form a partition $P = P_1 \cup P_2$. As we have proved $L(f,P_1) \le L(f,P) \le I \le U(f,P) \le U(f,P_2)$. Given that $U(f,P_2) - L(f,P_1) \lt \epsilon$, it follows that $U(f,P) - L(f,P) \lt \epsilon$.

b) <= (for every $\epsilon > 0$, there is a partition P such that $U(f,P) - L(f,P) \le \epsilon$): It is easy to see that the set of lower sums of all partitions is not empty (take any $\epsilon$ ...) bounded above (the upper bound of any partition is an upper bound). It follows that $\sup{\{L(f,P): \text{P is a partition on } [a,b]\}} = I_l$ exists. Similarly, $\inf{\{U(f,P): P \text{ is a partition on } [a,b]\}} = I_u$. 

Question: Is $I_u = I_l$? Assume they are not. Then $I_u - I_l = c$, say. Taking $\epsilon = c/2$, say, we have some partition P such that  U(f,P) - L(f,P) < c/2. If $U(f,P) \ge I_u$, then this would imply $L(f,P) \gt I_l$ - a contradiction. And we get a contradiction also if $L(f,P) \lt I_l$. So it must be that $I_u = I_l$. It follows then that f is integrable.

## Theorem (Continuity implies integrability) 

If f is continuous on an interval $[a,b]$, it is integrable on $[a,b]$

First: f is bounded on the interval $[a,b]$ since f is continuous and $[a,b]$ is a closed, bounded interval. For the same reason, f is also uniformly continuous i.e. given any $\epsilon > 0$, there is a $\delta > 0$, such that for all x,y in $[a,b]$, if |$x - y| < \delta$, then $|f(x) - f(y)| < \epsilon$.

So suppose we are given some $\epsilon > 0$. 

Use the facts above, we choose a $\delta > 0$ first as follows: If $|x -y| < \delta$, $|f(x) - f(y)| < \frac{\epsilon}{2(b-a)}$

Now, we choose any integer $n \gt \frac{b-a}{\delta}$. Choose a partition $\{t_0, t_1, \dots, t_{n-1}, t_n \}$, so the length of each interval $(t_{i-1}, t_i) < \frac{b-a}{\delta}$. It follows that if $m_i$ is the infimum of f in the interval $(t_{i-1}, t_i)$, and $M_i$ is the supremum of f in the interval $(t_{i-1}, t_i)$. Then by our hypothesis, $M_i - m_i \lt  \frac{\epsilon}{2(b-a)} \le \frac{\epsilon}{n\delta}$.

It follows that : $U(f,P) - L(f,P) = \sum\limits_{i=1}^{n} (M_i - m_i)(t_{i=1} - t_i) \lt \sum\limits_{i=1}^{n} \dfrac{\epsilon}{n\delta}\delta = \epsilon$


## Theorem (Splitting the interval of integration)

Let a < c < b. If f is integrable on $[a,b]$, then f is integrable on $[a,c]$ and on $[c,b]$. Conversely, if f is integrable on $[a,c]$ and on $[c,b]$, then f is integrable on $[a,b]$. Finally, if f is integrable on $[a,b]$, then:

$$\int_{a}^{b} f = \int_{a}^{c} f + \int_{c}^{b} f$$

Proof (outline):

a) => Take any partition P of f, and add c to it. 

b) <= Conversely, just use $\epsilon/2$ for each sum on the right. Then we get a sum lower than $\epsilon$ on the left.

## Theorem (Integral of Sum)

Let a < c < b. If f and are integrable on $[a,b]$, then f+g is integrable on $[a,b]$:

$$\int_{a}^{b} f + g = \int_{a}^{b} f + \int_{a}^{b} g$$

Proof:

Basically: $m_i \ge m_{i_f} + m_{i_g}$, $M_i \le M_{i_f} + M_{i_g}$

It follows that for any partition P: $L(f,P) + L(g,P) \le L(f+g,P) \le U(f+g,P) \le U(f,P) + U(g,P)$

## Theorem (Integral of a scaled functions)

If f is integrable on $[a,b]$, then for any number c, the function $cf$ is integrable on $[a,b]$ and:

$$\int_{a}^{b} cf = c\int_{a}^{b} f$$

Proof:
(for $c \ge 0$):

Given $\epsilon > 0$, choose P such that $U(f,P) - L(f,P) < \epsilon/c$.Then $cU(f,P) - cL(f,P) < \epsilon$, assuming c greater than zero. 

Now, choose g = -f. Then g + f = 0 i.e. cg + cf = 0 i.e. cg = -cf. Taking integrals, we get 

$$\int cg = -\int cf$$ i.e. 

$$\int -cf = -\int cf$$

## Theorem(Bounds of the Integral)

If f is integrable on $[a,b]$, and that :

$$m \le f(x) \le M \quad \text{for all x in } [a,b]$$

Then,

$$m(b-a) \le \int_{a}^{b} f \le M(b-a)$$

Proof:

Fairly obvious given $m(b-a) \le L(f,P)$ for all P, and $M(b-a) \ge U(f,P)$ for all P.

## Theorem (Integral is continuous)

If f is integrable on $[a,b]$ and F is defined on $[a,b]$ by $F(x) = \int_{a}^{x} f$, 
then, F is continuous on $[a,b]$

Proof:

The basic idea is to show, for any c in $[a,b]$ that $|F(c+h) - F(c)| \le M|h|$, when M is a bound of f on $[a,b]$, i.e $|f(x)| \le M$

Then given $\epsilon > 0$, if $|h| \lt \epsilon/M$, then $|F(c+h) - F(c)| \lt \epsilon$. 

We can basically see that $-Mh \le \int_{c}^{c+h} f \le Mh$, for h > 0. i.e. $-Mh \le F(c+h) - F(c) \le Mh$

Similarly, for h < 0, $Mh \le \int_{c+h}^{c} f \le -Mh$ i.e. $-Mh \le F(c) - F(c+h) \le Mh$. Thus we get $|F(c+h) - F(c)| \le M|h|$.
