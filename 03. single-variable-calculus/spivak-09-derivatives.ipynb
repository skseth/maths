## Derivative

The function f is differentiable at a if the following limit exists:

$$\lim_{h \to 0} \frac{f(a+h) - f(a)}{h}$$

The limit is denoted by $f'(a)$ and is called the **derivative of f at a**. We also say f is **differentiable** if f is differentiable at every point of its domain.

We can define the **left-hand derivative** and **right-hand derivative** in the expected way matching limits.

**Tangent Line of f at a:** The line through $(a, f(a))$ with slope $f'(a)$

**Average vs Instantaneous Rate of Change** $\frac{f(a+h) - f(a)}{h}$ is the average rate of change of f between a and (a+h). The limit on the other hand, is the instantaneous rate of change of f at a.

## Leibniz notation $\frac{df(x)}{dx}$ considered harmful, but necessary!

Spivak finds the Leibniz notation ambiguous. In particular, if it often unclear whether $\frac{df(x)}{dx}$ refers to $f'$ or $f'(x)$ i.e. the derivative function, or its value at a particular point. 



## Theorem (Differentiability implies Continuity)

"If a function f is differentiable at a point a, it is also continuous at that point."

Proof: $\lim_{h \to 0} f(a+h) - f(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h}h = f'(a).0 = 0$

**Note: Converse is not true - Continuity does not imply Differentiability**

e.g f(x) = |x|, at x = 0.

## Higher-order Derivatives

$f^{(1)} = f'$

$f^{(2)} = f'' = (f')'$

$f^{(3)} = f''' = (f'')'$

$f^{(4)} = f'''' = (f''')'$

Leibniz notation:

$$\frac{d\left( \frac{df(x)}{dx} \right)}{dx} = \frac{d^2f(x)}{(dx)^2} = \frac{d^2f(x)}{{dx}^2}$$
