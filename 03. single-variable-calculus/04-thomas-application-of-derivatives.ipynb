# Extreme Values of Functions

f(x) at x = a, with domain D:

**absolute maximum**: $f(x) \le f(a)$ for all x in D

**absolute minimum**: $f(x) \ge f(c)$ for all x in D

**local maximum**: $f(x) \le f(c)$ for all x in any open interval containing c

**local minimum**: $f(x) \ge f(c)$ for all x in any open interval containing c

These are **extreme values**


For a local minimum or maximum at an interior point : $f'(c) = 0$ (assuming f' is defined at c)

Note that a local max / min may occur also at an end point of the domain, or at an point where f' is not defined.

A **critical point** of f is an interior point in the domain of f, where f' is either 0 or undefined. Thus the only points where a function can assume extreme values are critical points or endpoints.


```python
from sympy import symbols, diff, solve, plot

# 1. Define the symbol and the function
x = symbols('x')
f = 2*x**3 - 3*x**2 - 12*x + 5

# 2. Find the first derivative
f_prime = diff(f, x)

# 3. Solve f'(x) = 0 to get critical points
critical_points = solve(f_prime, x)
print(f"Critical Points: {critical_points}")
# Output: [-1, 2]

# 4. Classify using the second derivative test
f_double_prime = diff(f, x, 2)

for point in critical_points:
    val = f_double_prime.subs(x, point)
    if val > 0:
        print(f"x = {point} is a Local Minimum (f'' = {val})")
    elif val < 0:
        print(f"x = {point} is a Local Maximum (f'' = {val})")
    else:
        print(f"x = {point} is inconclusive")

plot(f, (x, -3, 4), title="Graph of f(x)", xlabel="x", ylabel="f(x)")
```

## Extreme Value Theorem

**If f is continuous on a closed interval [a,b] then f attains both an absolute maximum M and an absolute minimum value m in [a,b].**

Proof:

The proof depends on 

a) The completeness property of the real number system: Every non-empty subset of the real numbers has a supremum or a "least upper bound". Note this is not true for Q.

b) Bolzano-Weierstrass Property: Ensures that every bounded sequence of real numbers has a convergent subsequence. This in turn depends on the fact Heine-Borel theorem which shows that in R (or R^n), a set is compact if and only if it is both closed and bounded. A compact set has the property that every open cover has a finite subcover.

See Baby Rudin Theorem 4.16 for proof. In brief, it shows that since [a,b] is closed and bounded (and hence compact), f(X) is also compact, and hence closed and bounded. It follows that f attains an absolute maximum and minimum (since it is bounded, and attains the bounds).

**Note** continuity is a must, as well as the interval being closed and bounded.

## Mean Value Theorem

**Rolle's Theorem**

**Suppose y = f(x) is continuous on a closed interval [a,b], and differentiable at every point of its interior (a,b). If f(a) = f(b), then there is at least one point c in (a,b) at which f'(c) = 0.**

Intuitively, the function has to reach some maximum and curve back to f(b) (unless it is completely flat, in which case the theorem is trivially true).

We know f is continuous and hence attains its bounds. The only points in (a,b) where it can attain maximum or minimum is where f' = 0, because there are no points where the derivative is undefined. If the endpoints are both absolute maximum and minimum, then f must be flat, and f'(c) = 0 everywhere.

Note that every condition is necessary:

- if f is discontinuous at the end point of [a,b], then no horizontal tangent may exist 
- if there is a jump anywhere, again same case
- If there is a non-differentiable interior point, again horizontal tangent may not exist e.g. a cusp.


**IVT - Intermediate Value Theorem**

If f is continuous on a closed interval [a,b], then f attains any value $y$ between f(a) and f(b) at some point c in [a,b].

The proof depends on the completeness property of R.

**Mean Value Theorem**

This is the slanted version of Rolle's theorem.

**Suppose y = f(x) is continuous on a closed intevral [a,b], and differentiable at every point of its interior (a,b). Then there is at least one point c in (a,b) at which $\frac{f(b) - f(a)}{b - a} = f'(c)$.**

We define : $g(x) = f(a) + \frac{f(b) - f(a)}{b - a}(x - a)$ as the line joining f(a) to f(b). Since this is just a line, this is continuous and differential.

We then define:
$h(x) = f(x) - g(x) = f(x) - f(a) - \frac{f(b) - f(a)}{b - a}(x - a)$

h is continuous & differentiable (because both f and g are). And h(a) = h(b) = 0. So h satisfies the conditions for Rolle's theorem to apply. Then there is a c in (a,b) where h'(c) = 0. 

$h'(c) = f'(c) - \frac{f(b) - f(a)}{b - a} = 0$. Then:

$f'(c) = \frac{f(b) - f(a)}{b - a}$


Why is it called the mean value theorem? The term $\frac{f(b) - f(a)}{b - a}$ is the mean rate of change over the interval [a,b]. So the theorem is saying there is an interior point where the instantaneous rate of change equals the mean rate of change over the entire interval.

**Corollaries**

**If f'(x) = 0 at each x of (a,b), f(x) = C for all x in (a,b), where C is a constant.**

Note that the interval is open here, we cannot directly apply MVT.

Take any interval [a',b'] in (a,b). Within [a',b'] f satisfies MVT hypothesis. It follows that there is c in (a',b') such that 

$f'(c) = \frac{f(b') - f(a')}{b' - a'} = 0$, since f' is zero everywhere in (a,b).

It follows that $f(b') - f(a') = 0$, i.e. $f(b') = f(a')$. Since a',b' were arbitrary points in (a,b) it follows that f'(x) = C in (a,b), where C is some constant.

**If f'(x) = g'(x) at each x of (a,b), f(x) = C for all x in (a,b), then f(x) = g(x) + C, for all in (a,b) where C is a constant.**

Define h = f - g. Then h is a constant function. It follows f - g = C, or f = g + C.


## Monotonic Functions and the First Derivative Test

**Suppose f is continuous on [a,b], differentiable on (a,b).**

(a) If f'(x) > 0 for each point in (a,b), then f is increasing on [a,b]

(b) If f'(x) < 0 for each point in (a,b), then f is decreasing on [a,b]

Proof:

f satisfies the hypothesis of MVT.

Take any interval $[x_1,x_2]$ in [a,b]. By MVT there is a c in $(x_1, x_2)$ such that, $f(x_2) - f(x_1) = f'(c)(x_2 - x_1)$. 

Since f'(c) > 0, and $x_2 > x_1$, $f(x_2) > f(x_1)$.

We can similarly prove (b), when f'(c) < 0.

**First Derivative Test for Local Extrema**

Suppose c is a critical point of a continuous function f, and that f is differentiable at every point in the interval containing c, except possibly c itself:

Moving from left to right:

- if f' changes from negative to positive, c is a local minimum
- if f' changes from positive to negative, c is a local maximum
- if f' does not change sign at c, f has no local extrema at c



**Second Derivative Test for Concavity**

Let y = f(x) be a twice differentiable function in an interval I.

1. if f'' > 0 on I, f is "concave up" on I, i.e. f' is increasing
2. if f'' < 0 on I, f is "concave down" on I, i.e. f' is decreasing

**Points of Inflection**

A point where the function has a tangent line (even "vertical"), and where concavity changes.

Necessary condition (but not sufficient): At an inflection point c, f''(c) = 0, or f''(c) fails to exist.

## Newton's Method

Choose $x_n$. Take the tangent and get the intercept with x-axis ($x_{n+1}$).

$f'(x_{n}) = \frac{f(x)}{x_n - x_{n+1}}$, or :

$$x_{n+1} = x_n - \frac{f(x)}{f'(x_{n})}$$

```python
import sympy as sp

def sym_newton_method(expr_str, x0, tol=1e-6, max_iter=100):
    """
    Finds the root of an equation using Newton's method with SymPy.
    
    Parameters:
    - expr_str (str): The mathematical equation as a string (e.g., 'x**3 - 2*x - 5')
    - x0 (float): The initial guess
    - tol (float): The convergence tolerance
    - max_iter (int): Maximum number of iterations
    """
    # 1. Define the symbol and parse the expression string
    x = sp.symbols('x')
    f = sp.sympify(expr_str)
    
    # 2. Symbolically calculate the derivative
    df = sp.diff(f, x)
    
    # 3. Compile expressions into fast numerical functions
    f_num = sp.lambdify(x, f, 'numpy')
    df_num = sp.lambdify(x, df, 'numpy')
    
    x_val = float(x0)
    
    # 4. Iteratively update the guess
    for i in range(max_iter):
        f_val = f_num(x_val)
        df_val = df_num(x_val)
        
        # Prevent division by zero if the derivative is too close to zero
        if abs(df_val) < 1e-12:
            raise ValueError(f"Derivative is zero at x = {x_val}. Method fails.")
            
        x_new = x_val - (f_val / df_val)
        
        # Check for convergence
        if abs(x_new - x_val) < tol:
            print(f"Converged to root in {i+1} iterations.")
            return round(float(x_new), 6)
            
        x_val = x_new
        
    raise RuntimeError("Maximum iterations reached without convergence.")

# --- Example Usage ---
equation = "x**3 - 2*x - 5"
initial_guess = 2.0

root = sym_newton_method(equation, initial_guess)
print(f"The root of {equation} starting from guess {initial_guess} is: {root}")

g = sp.sympify(equation)
newroot = sp.nsolve(g, x, 2.0)
print(f"The root is: {newroot:.4f}")
```
