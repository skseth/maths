# Least Upper Bound

Standard Definition of :

- bounded above, bounded below
- upper bound, lower bound
- least upper bound (supremum or sup), greatest lower bound (infimum or inf)

## Least Upper Bound Property of R

If A is a non-empty set of real numbers, and A is bounded above, then A has a **least upper bound**.

Note: The bound may not be in A.

**Corollary: Greatest Lower Bound property**

Suppose if A is a non-empty set of real numbers, and A is bounded below. Let 

$B = \{x | x \text{ is a lower bound of A }\}$

B is not empty - by hypothesis. And B has an upper bound (since A is non-empty), any element of A is an upper bound of B. Thus B has a lub (supremum). Is sup B a lower bound of A? If any element of A is smaller than sup B, there will be a non-empty interval containing upper bounds of B but smaller than sup B, which is contradictory. So sup B is the greatest lower bound of A. Note that it may be in A.

## Theorem 7-1 (Intermediate Value Theorem): If f is continuous on [a,b], and f(a) < 0 < f(b), then there is some x in [a,b] such that f(x) = 0 - Proof

There are one crucial point : We try to find an x, as the **first x** where f(x) i.e. for the interval [a,x), f < 0, and f(x) = 0. We have to prove such an x exists of course. We ignore any larger value where f may be zero. Thus we define a set A:

$A = \{ x | a \le x \le b, \text{ f is negative on interval } [a,x] \}$

With this it becomes simple. 
- A is non-empty: a belongs to A. In fact, using the behaviour of the interval given a right-hand limit at a, we know there is a $\delta > 0$ such that for all x in $[a,a+\delta)$, $f(x) < 0$.

- A is bounded above (b is an upper bound), by definition, and again there is a $\delta > 0$, s.t. for all x in $(b-\delta,b]$, f(x) > 0.

So, we know A is not empty, and that it has an upper bound $\alpha$ strictly smaller than b, and strictly greater than a, i.e. $a \lt \alpha \lt b$. We have to show $f(\alpha) = 0$. We show $f(\alpha)$ cannot be positive or negative.

Suppose $f(\alpha) \lt 0$: Now, by definition of continuous function and the behaviour of a continuous function at an interval, there is a $\delta > 0$, such that for $\alpha - \delta < x < \alpha + \delta$, f(x) < 0. And by definition of a least upper bound of A, there must be a member $x_0$ of A in $(\alpha - \delta,x]$, other x cannot be the least upper bound - which implies f is negative on the interval $[a, x_0]$. But there is also some $x_1 \in [\alpha, \alpha + \delta)$, where f is negative for the entire interval $[x_0, x_1]$ i.e. f is negative on the interval $[a,x_1]$, and $x_1 \in A$, contradicting our assumption that $\alpha$ is the lub. It follows that $f(\alpha) \ge 0$.

Now suppose $f(\alpha) \gt 0$: Again we have an interval $\alpha - \delta < x < \alpha + \delta$, where f(x) > 0. 
We see (by definition of lub) that there is an $x_0 \in (\alpha - \delta, \alpha)$, such that $x_0 \in A$. But this implies f is negative in $[a,x_0]$, and more specifically $f(x_0) < 0$. This is not possible, since $f(x_0) > 0$. Hence, the only possibility left is $f(\alpha) = 0$.

Note the interplay between the lub property and the continuity of the function - both are needed to make the proof work. Without the lub property the domain may have gaps, and theoretically one could skip the x-axis while going from negative to positive values of f.



## Intermediate Result

If f is continuous at a, then there is a number $\delta \gt 0$, such that f is bounded above on the interval $(a - \delta, a + \delta)$. 

Proof: By definition, given an $\epsilon > 0$, we have a $\delta \gt 0$ such that if $|x - a| \lt \delta$, then $|f(x) - f(a)| \lt \epsilon$. This implies $f(x) - f(a) \lt \epsilon$ or $f(x) \lt f(a) + \epsilon$. Equally, $-\epsilon < f(x) - f(a)$ i.e. $f(x) \gt f(a) - \epsilon$. So f is bounded in the interval $(f(a) - \epsilon, f(a) + \epsilon)$, when x is in the open interval $(a - \delta, a + \delta)$

Similar Points: If $\lim_{a^+} f(x) = f(a)$, then there is an half-open interval $a,a + \delta)$ where f is bounded, and if $\lim_{a^-} f(x) = f(a)$, then there is a half-open interval $(a - \delta, a]$, where f is bounded, where $\delta > 0$.

Proof: Again by definition, we get if $0 < x - a \lt \delta$, then $|f(x) - f(a)| \lt \epsilon$ (as per usual). This implies $a \lt x \lt a + \delta$, i.e. when $x \in (a, a + \delta)$ then f(x) is bounded in the interval $(f(a) - \epsilon, f(a) + \epsilon)$. Of course f(a) itself lies in this interval, so we can say f is bounded in the interval $[a, a + \delta)$. Similar proof can be suppled for the left-handed continuity case.

# Theorem 7-2 (Boundedness Theorem): If f is continuous on [a,b], then f is bounded above on [a,b] - Proof. Similarly, f is also bounded below on [a,b] i.e. f is bounded on [a,b].

We define $A = \{x: a \le x \le b, \text{ f is bounded above on }  [a,x]  \}$

A is not empty (a is in A). A is bounded above by b, so A has a sup $\alpha \le b$. We need to show $\alpha = b$.

So suppose $\alpha \lt b$. And we know $\alpha > a$. Then by the intermediate result above, there is a $\delta'$ around $\alpha$ where f is bounded, and we can choose a $0 < \delta < min(\alpha - a, b - \alpha, \delta')$ such that in the interval $(\alpha - \delta, \alpha + \delta)$, f is bounded. Now in the interval $(\alpha - \delta, \alpha)$, there is an $x_0 \in A$, because $\alpha$ is the supremum. It follows that $[a, x_0]$ is bounded above. But we also have an $x_1$ in $(\alpha, \alpha + \delta)$ such that f is bounded above on $[x_0, x_1]$, which means that $[a, x_1]$ is bounded above and $x_1$ is in A. Which is a contradiction because $x_1 \gt \alpha$. This implies $\alpha = b$. 

But does $\alpha = b$ belong to A? We can use the same interval-based argument to show $[a,b]$ is bounded above. Or we can see that [a,b) is in A (since b is the supremum). And f is also bounded above at b, so $[a,b) U \{b\}$ is bounded above i.e. $[a,b]$ is bounded above, which means that b is in A.

To prove the reverse:

We define $B = \{x: a \le x \le b, \text{ f is bounded below on }  [x,b]  \}$. Reversing the argument, B is not empty, B is bounded below. B has an infimum. Reversing the argument above, we will prove that f is bounded below as well, i.e. f is bounded.





## Theorem 7-3 (Extreme Value Theorem): If f is continuous on [a,b], then there is some number y in [a,b] such that $f(y) \ge f(x)$ for all x in [a,b]. - Proof

We have proved f is bounded above. But this theorem states tha a continuous function "attains" its bounds.

Proof: Suppose $\alpha$ is the supremum of f on [a,b]. Assume for every x in [a,b], f(x) is strictly less than $\alpha$. Then $\alpha - f(x) \gt 0$ for every x in [a,b]. 

We define g(x) as :

$$g(x) = \frac{1}{\alpha - f(x)}, x \in [a,b]$$

g(x) should be continuous on $[a,b]$ since denominator is never zero, and hence bounded.

But $\alpha$ is the supremum, which means given an $\epsilon > 0$, there is an $x \in [a,b]$ with $\alpha - f(x) \lt \epsilon$. This implies for every $\epsilon$, there is an x in [a,b] with $g(x) \gt \frac{1}{e}$ i.e. g(x) is not bounded.


## Archimedean Property: N is not bounded above. Or (Rudin): Given x,y in R, x > 0, there is an n in N, such that nx > y.

**Spivak proof**: Suppose N were bounded above (in R). Since $N \ne \emptyset$, there would be a least upper bound $\alpha$ for N. Then:

$\alpha \ge n$, for all n in N, and $\alpha \ge n+1$ for all n in N (since N is inductive).

But this implies $\alpha - 1 \ge n$ for all n in N ie. $\alpha -1$ is a supremum of N - which is a contradiction.

**Rubin's statement proof (Given Spivak)**: Given x and y as above, x > 0, choose n > y/x. Then nx > y. 

**Corollary: For any $\epsilon \gt 0$, there is a natural number n such that $\frac{1}{n} \le \epsilon$** 
Proof: If not, then $\frac{1}{n} \gt \epsilon \implies n \lt \epsilon$, for all n in N. i.e. $\frac{1}{\epsilon}$ is an upper bound of N - which is a contradiction.
