## Chain Rule

If f(x) is differentiable at a, and g(x) is differentiable at f(a) :

$$(f \circ g)(x) = f'(g(a))g'(a)$$

In the leibniz notation, this should really be expressed as :

$$\frac{df(g(x))}{dx} = \frac{df(y)}{dy} \Bigg|_{y=g(x)}\frac{dg(x)}{dx}$$, but it is often expressed as something like this:

$$\frac{dz}{dx} = \frac{dz}{dy}\frac{dy}{dx}$$

Proof:

Roughly, we want to show that given an $\epsilon > 0$, we can find a $\delta'$ such that when $|g(x) - g(a)| \lt \delta'$, if $|f(g(x)) - f(g(a))| \lt \epsilon$. And then we use the $\delta'$ as an $\epsilon$ for g(x), and find a $\delta$ such that if $|x - a| \lt \delta$, then $|g(x) - g(a)| \lt \delta'$. Then we get an $\epsilon-\delta$ relation from x to f(g(x)).

Define 

$$
\phi(h) = 
\begin{cases}
    \displaystyle \frac{f(g(a+h)) - f(g(a))}{g(a+h) - g(a)} &\text{ if } g(a+h) \ne g(a) \\[.5cm]
    f'(g(a)) &\text{ if } g(a+h) = g(a) 
\end{cases}
$$

**Step 1: Prove $\phi$ is continuous at h = 0**

(a) We start with the fact that f is differentiable at g(a). Hence:

$\lim_{k \to 0} \frac{f(g(a) + k) - g(a)}{k} = f'(g(a))$. If $\epsilon \gt 0$, there is a $\delta' > 0$, such that if $0 < |k| < \delta'$, then $|\frac{f(g(a) + k) - g(a)}{k} - f'(g(a))| \lt \epsilon$

(b) Next, we see that g is differentiable (and hence continuous) at a. So given $\delta'$ (important to see the connection here with f), there is a $\delta > 0$, such that for all h, if $|h| < 0$, then $|g(a+h) - g(a)| \lt \delta'$

(c) Now take any h with $|h| \lt \delta$. Define k = g(a+h) - g(a). Note that $k \lt \delta'$, by (b).

(c.1) If $k \ne 0$: $\phi(h) = \dfrac{f(g(a+h)) - f(g(a))}{g(a+h) - g(a)} = \dfrac{f(g(a) + k) - f(g(a)}{k}$. 
Since $|k| \lt \delta$, the condition in (a) applies, and $|\phi(h) - f'(g(a))| \lt \epsilon$

(c.2) If $k = 0$: Then $\phi(h) = f'(g(a))$ by definition.

Thus $\lim_{h \to 0} \phi(h) = f'(g(a))$ i.e. $\phi$ is continuous at h = 0.

**Step 2: Prove Chain Rule using $\phi$**

If $h \ne 0$, 

$\dfrac{f(g(a+h)) - f(g(a))}{g(a+h) - g(a)} = \phi(h)\dfrac{g(a+h) - g(a)}{h}$. Taking limits which we now know exist, we get:

$(f \circ g)'(a) = \lim_{h \to 0}\phi(h)\lim_{h \to 0}\dfrac{g(a+h) - g(a)}{h} = f'(g(a)).g'(a)$
