# Limits

**Definition (Limit)** The function f approaches a limit l near a, if for every $\epsilon$ > 0, there is a $\delta$ > 0, such that for all x, if $0 \lt |x - a| \lt \delta$, then $|f(x) - l| \lt \epsilon$.

**Definition (Right/Left-Hand Limits)** 

We say $lim_{x \to a^+} f(x) = l$ if for every $\epsilon \gt 0$, there exists a $\delta$, such that for all x, if $0 \lt x - a \lt \delta$, then $|f(x) - l| < \epsilon$. Similarly we can define $lim_{x \to a^-} f(x) = l$ if  for every $\epsilon \gt 0$, there exists a $\delta$, such that for all x, if $0 \lt a - x \lt \delta$, then $|f(x) - l| < \epsilon$.

**Theorem (Limit is unique)**

"A function f cannot approach two different limits near a. In other words, if f approaches l near a, and f approaches m near a, then l = m."

Proof: Easy - just choose an $\epsilon$ less that half the distance between the two limits, and you get a contradiction.


**Theorem (Sum, Product, Reciprocal of Limits)**

If $\lim_{x \to a} f(x) = l$ and $\lim_{x \to a} g(x) = m$, then:

$$\tag{1} \lim_{x \to a} (f + g)(x) = l + m$$

$$\tag{2} \lim_{x \to a} (f.g)(x) = lm$$

$$\tag{3} \lim_{x \to a} \left( \frac{1}{g} \right) (x) = \frac{1}{m} \text{ assuming m is not zero }$$

Proof:

For (1) - given any $\epsilon$, find deltas corresponding to $\frac{\epsilon}{2}$ for f and g. |f - l| < e/2, |g - m| < e/2. 
|f+g - (m+l)| = |f-m + g-l| <= |f-m| + |g-l| = e.

For (2) - $|f(x)g(x) - lm| = |f(x)(g(x) - m) + m(f(x) - l)| \le |f(x)||g(x) - m| + |m||f(x) - l|$. 

We choose $\delta_1$ such that $|f(x) - l| \lt \min({1,\frac{\epsilon}{2(|m| + 1)}})$ and $\delta_2$ s.t. $|g(x) - m| \lt \min({1, \frac{\epsilon}{2(|l| + 1)}})$. Then choose $\delta = min(\delta_1, \delta_2)$. Then :

$|f(x)g(x) - lm| \lt (1 + |l|)\frac{\epsilon}{2(|l| + 1)} + (1 + |m|)\frac{\epsilon}{2(|m| + 1)} = \frac{\epsilon}{2} + \frac{\epsilon}{2} = \epsilon$

For (3) : We need : $|(\frac{1}{g})(x) - 1/m| \lt \epsilon$

Let y = g(x). We need a good delta : $|1/y - 1/m| = \frac{|y - m|}{|m||y|}$

We choose $|y - m| \lt \min(\frac{|m|}{2}, \epsilon\frac{m^2}{2})$. Then 

$|m| - |y| < |y - m| \lt \frac{m}{2} \implies |y| \gt \frac{|m|}{2} \implies \frac{1}{|y|} \lt \frac{2}{m}$

Then $|(\frac{1}{y}) - 1/m| = \frac{|y - m|}{|m||y|} \lt \frac{1}{|m|}\frac{2}{|m|}\frac{\epsilon.m^2}{2} = \epsilon$



## Infinite Limit

We define $\lim\limits_{x \to a} f(x) = \infty$ as:

Given any N, no matter how large, there is a $\delta > 0$ such that if $|x - a| < \delta$, then $f(x) > N$.

Ex: Show $\lim\limits_{x \to 3} \dfrac{1}{(x - 3)^2} = \infty$

Given an $N > 0$, we need to find a $\delta$, such that if $|x - 3| \lt \delta$, then $|\dfrac{1}{(x - 3)^2}| > N \implies \dfrac{1}{(x - 3)^2} > N$. We choose $\delta = \dfrac{1}{\sqrt{N}}$. Then we have $|x - 3| \lt \dfrac{1}{\sqrt{N}}$. Then $(x - 3)^2 < \dfrac{1}{N}$. This implies $\dfrac{1}{(x - 3)^2} > N$.

## Limits at Infinity

We define $\lim\limits_{x \to \infty} f(x) = l$ as:

Given any $\epsilon > 0$ there is an N such that if $x > N$, then $|f(x) - l| \lt \epsilon$
