# Continuity

The function f is continuous at a if :

$$\lim_{x \to a} f(x) = f(a)$$

## Theorems: continuity of f+g, f.g, 1/g

If f and g are continuous at a:

(a) f + g is continuous

(b) f.g is continuous

(c) f/g is continuous, assuming $g(a) \ne 0$

## Theorem: Continuity of $f \circ g$

If g is continuous at a, and f is continuous at **g(a)**, then $f \circ g$ is continuous at a.

If f is continuous at g(a), we can say that given a $\epsilon$, there is a $\delta'$ such that for all y :

If $|y - g(a)| \lt \delta'$, then $|f(y) - f(g(a))| \lt \epsilon$. Thus

If $|g(x) - g(a)| \lt \delta'$, then $|f(g(x)) - f(g(a))| \lt \epsilon$.

We now use $\delta'$ as an epsilon: Given a $\delta'$, there is a $\delta$ such that for all x:

If $|x - a|$, then $|g(x) - g(a)| \lt \delta'$, which implies $|f(g(x)) - f(g(a))| \lt \epsilon$

## Problem 6-16 (Right hand Continuity - Interval Behaviour)

Theorem: Suppose $\lim_{x \to a^+} f(x) = f(a)$, and $f(a) \gt 0$. Then there is a $\delta > 0$, s.t. f(x) > 0 for all x satisfying : $0 \le x - a \lt \delta$. And conversely for when we have $f(a) \lt 0$, or for a left-hand continuity.

Proof: By definition of right-hand limit, it follows that given any $\epsilon > 0$, there is a $\delta > 0$, s.t. for all x, if 
$0 \lt x - a \lt \delta$, then $|f(x) - f(a)| < \epsilon$. Choose $\epsilon < f(a)$. 

Then $|f(x) - f(a)| \lt f(a)$ or $-f(a) \lt f(a) - f(x) \lt f(a)$. Add f(a):

$0 \lt f(x) \lt 2f(a)$, i.e. f(x) is positive.

We can similarly prove the left hand continuity side.

## Appendix: Uniform Continuity



**Definition(Uniformly Continuous Function)**

The function f is **uniformly continuous function on an interval A** if for every $\epsilon > 0$ there is some $\delta > 0$ such that, for all x and y in A,

if $|x - y| \lt \delta$, then $|f(x) - f(y)| \lt \epsilon$

**Theorem: If f is continuous on $[a,b]$, then f is uniformly continuous**

**Lemma**: Let a < b < c and let f be continuous on the interval $[a,c]$. Let $\epsilon > 0$, and suppose that there are $\delta_1$ and $\delta_2$ such that if x,y in $[a,b]$, then if $|x - y| \lt \delta_1$, then $|f(x) - f(y)| \lt \epsilon$. Similarly, if x,y in $[b,c]$, then if $|x - y| \lt \delta_2$, then $|f(x) - f(y)| \lt \epsilon$. Then, there is a $\delta$ such that given x,y in $[a,c]$, if $|x - y| \lt \delta$, then $|f(x) - f(y)| \lt \epsilon$.

Proof:

Since f is continuous on $[a,c]$, it is continuous at b. Choose a $\delta_3$ such that given x, y in $[a,c]$, if $|x - b| \lt \delta_3$, then $|f(x) - f(b)| \lt \epsilon/2$. 

Now define $\delta$ as the min of the 3 deltas.

Clearly if x,y are in $[a,b]$ or $[b,c]$, then if $|x - y| \lt \delta$, $|f(x) - f(y)| \lt \epsilon$, since $\delta \le \delta_1$. 

On the other hand, if x is in $[a,b]$ and y in $[b,c]$, if $|x - y| \lt \delta$, $|f(x) - f(y)| \lt \epsilon$, since $\delta \le \delta_2$.

Now suppose x < b < y, or y < b < x. If $|x - y| \lt \delta$, it follows that $|x - b| \lt \delta$ and $|y - b| \lt \delta$. It follows that $|f(x) - f(b)| \lt \epsilon/2$ and $|f(y) - f(b)| \lt \epsilon/2$ i.e. $|f(x) - f(y)| \lt \epsilon$.

Proof:

Spivak defines that given an $\epsilon > 0$, f is $\epsilon-good$ on $[a,b]$ if there is some $\delta > 0$ such that for all y and z in $[a,b]$: if $|y - z| \lt \delta, then |f(y) - f(z)| < \epsilon$. 

We then define a set $A = \{x : a \le x \le b \text{ and } f \text{ is } \epsilon\text{-good on } [a,x]\}$

Obvious $a \in A$, since $|a - a| = 0$, so A is not empty. Obviously also, A is bounded above by b, by definition. So A has a least upper bound $\alpha$.

Suppose $\alpha < b$: Since f is continuous at $\alpha$, there is some $\delta_0 > 0$ such that if $|y - \alpha| \lt \delta_0$, $|f(y) - f(\alpha)| \lt \epsilon/2$

It follows from the triangle inequality, that if $|y - \alpha| < \delta_0$, and $|z - \alpha| < \delta_0$, then $|f(y) - f(z)| \lt \epsilon$. This implies f is $\epsilon\text{-good}$ on the interval $[\alpha - \delta_0, \alpha + \delta_0]$. Also, given $\alpha$ is the supremum of A, f is also $\epsilon\text{-good}$ on $[a, \alpha - \delta_0]$. This implies f is $\epsilon\text{-good}$ on $[a,b]$.
