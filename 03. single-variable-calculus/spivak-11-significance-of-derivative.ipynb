# Significance of the Derivative

**maximum / minimum point, maximum / minimum value** : Let f be a function and A be any subset of the domain of f. A point in is the **maximum point** of f on A, if 

$f(x) \ge f(y)$, for every y of A.

f(x) is the **maximum value** of f on A. 


**local maximum / minimum point** : Let f be a function and A be any subset of the domain of f. A point in is the **local maximum point** of f on A, if there is some $\delta \gt 0$ such that x is a maximum point for f on $A \cap (x - \delta, x + \delta)$. In other words there is some open interval containing x, where 

**critical point / critical value** : A critical point of a function f is a number x such that 

$f'(x) = 0$

The point f(x) is called a critical value of f.

**increasing / decreasing** A function f is said to be increasing on an interval if given any 2 points a and b in the interval (a < b), f(a) < f(b). Decreasing implies f(a) > f(b).



# Theorem (Max/Min and the First Derivative)

Let f be any function defined on (a,b). If x is a maximum or minimum point for f on (a,b), and f is differentiable at x, then $f'(x) = 0$.

Proof:

$\lim_{h \to 0} \dfrac{f(x + h) - f(x)}{h}$ exists. Since x is a maximum point, f(x) >= f(x+h). If we come from the left, h is negative, and hence the limit is negative or zero. If we come from the right, the limit is positive or zero. The only common value is zero, so $f'(x) = 0$ at x.

**Note**: Converse is not true. Just because $f'(x) = 0$, it need not necessarily be a minimum or maximum value.


### Notes on Maximum & Minimum of f on a closed interval [a,b]

3 types of points:

1) Critical points of f
2) End points a and b
3) Points x in (a,b) where f is not differentiable at x (NOTE: Spivak uses [a,b])



## Theorem (Rolle's Theorem) 

If f is continuous on $[a,b]$ and differentiable on $(a,b)$, and $f(a) = f(b)$, then there is an x in (a,b) such that f'(x) = 0

Proof:

Because of continuity of f on $[a,b]$, f is bounded on $[a,b]$ i.e. that it has a maximum and minimum value on $[a,b]$. Suppose $x \in (a,b)$ is a maximum (or a minimum), then f'(x) = 0 by the theorem above. The only other option is that the maximum or minimum values both occur at the endpoints a and b - but this implies f is constant, and f'(x) = 0 on the entire interval (a,b)

Note: If f is not differentiable everywhere on (a,b), we need not have an x such that f'(x) = 0. And if it is not continuous on the entire interval, then the boundedness is not guaranteed.


## Theorem (Mean Value Theorem)

If f is continuous on $[a,b]$, and differentiable on $(a,b)$, then there is an x in $(a,b)$ such that 

$$f'(x) = \left[ \frac{f(b) - f(a)}{b - a} \right]$$

Note that the term in the square bracket is the "mean" or "average" value of the function over the interval. So another way to say is that somewhere in the interval (a,b) the function attains an instantaneous rate of change equal to its average over the whole interval.

Proof:

Define $h(x) = f(x) - \dfrac{f(b) - f(a)}{b - a}(x - a)$

h(a) = h(b) = f(a). And h is also continuous at $[a,b]$, and differentiable at (a,b). Hence Rolle's Theorem applies, and there is an x in (a,b) such that :

$0 = h'(x) = f'(x) - \dfrac{f(b) - f(a)}{b - a}$ i.e. $f'(x) = \dfrac{f(b) - f(a)}{b - a}$


**Corollary (f constant)** 

If f is defined on an interval $[a,b]$, and f'(x) = 0 for all x on (a,b), then f is constant on an interval.

Proof:

Choose any a, b in the interval, a < b. f is continuous on [a,b], and differentiable on (a,b) - as per hypothesis. So there is an x in (a,b) such that:

$f'(x) = \dfrac{f(b) - f(a)}{b - a} = 0$ i.e. f(a) = f(b). Since a,b are any two arbitrary points in the interval, this means f is constant on the interval.

**Corollary (f' = g')**

If f and g are defined on an interval, and $f'(x) = g'(x)$ for all x in the interval, then there is some number c such that $f = g + c$. 

Proof: Define h = f - g i.e. h' = f' - g' = 0. It follows h is constant i.e. h = c = f - g, or f = g + c.



**Corollary(increasing/decreasing)**

If f'(x) > 0 for all x in an interval, then f is increasing in the interval; if f'(x) < 0 for all x in an interval, then f is decreasing in the interval.

Proof: Suppose f'(x) > 0. By definition, when h < 0, f(x + h) - f(x) < 0 i.e. f(x+h) < f(x). And when h > 0, f(x + h) - f(x) > 0 i.e. f(x) < f(x+h). Similarly we can show for f'(x) < 0.


**Example: Polynomial has at most n roots**

Suppose we have a polynomial $f(x) = a_nx^n + a_{n-1}x^{n-1} + \dots + a_0$, then we want to show f has at most n roots. To prove this, let us start with :

a) If f has k roots, then f' has at least k-1 roots: As proof, see that between any two roots of f(x), we can apply Rolle's Theorem on the interval between the roots - so we see that each of these k-1 intervals has a point where f' is 0. 

b) Now, for n = 1, $f(x) = a_1x + a_0$ i.e. f(x) has at most one root at $x = -\dfrac{a_0}{a_1}$. Now assume the statement is true for some n. Then consider a polynomial f of degree n+1. Suppose this polynomial f has more than n+1 roots. If we differentiate, we get a polynomial g of degree n. But by (a), g would have more than n roots, a contradiction of the hypothesis.


**Polynomial Sketching**

Take a polynomial like $f(x) = x^4 - 2x^2$. Then $f'(x) = 4x^3 - 4x = 4x(x - 1)(x + 1)$. 

(i) critical points of f are x = -1, 0, 1, where f(x) = -1, 0 and -1.

(ii) It is reasonable to assume f is increasing in the interval (-1,0) and decreases on (0,1) by seeing sign of f' (f' is +ve between -1 and 0, -ve between 0 and 1)

(iii) f(x) is even (f(x) = f(-x)), so graph is symmetric about y axis

(iv) At extremes, highers powers of x dominate ($x^4$). So reasonable to assume outside of $[-1, 1]$ f(x) is increasing.

(v) f(x) = 0 at x = 0, $x = \pm\sqrt{2}$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from sympy.calculus.util import minimum, maximum, stationary_points

x = symbols('x')

class SymPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        self.isMinimum = False
        self.isMaximum = False
        self.isZero = False
        self.isCritical = False
        self.isEndpoint = False

    def setCritical(self):
        self.isCritical = True
    
    def setMinimum(self):
        self.isMinimum = True

    def setMaximum(self):
        self.isMaximum = True

    def setZero(self):
        self.isZero = True

    def setEndpoint(self):
        self.isEndpoint = True

    def color(self):
        if self.isZero: return 'green'
        if self.isCritical: return 'orange'
        if self.isMinimum: return 'pink'
        if self.isMaximum: return 'blue'
        if self.isEndpoint: return 'black'

    def toString(self):
        descriptors = []
        if (self.isMinimum): descriptors.append("Min") 
        if (self.isMaximum): descriptors.append("Max")  
        if (self.isZero): descriptors.append("Zero")  
        if (self.isCritical): descriptors.append("Crit") 
        if (self.isEndpoint): descriptors.append("EP") 
        desc = "|".join(descriptors)
        
        return f"({float(self.x):.3f},{float(self.y):.3f}) - {desc}"


class SymFunction:
    def __init__(self, f, indVar, domain):
        self.f = f
        self.indVar = indVar
        self.domain = domain
        self.points = []
        self.f_num = sp.lambdify(indVar, f, 'numpy')

    def initPoints(self):
        min_val = minimum(self.f, self.indVar, domain)
        max_val = maximum(self.f, self.indVar, domain)
        
        crit_points = stationary_points(self.f, self.indVar, self.domain)
        crit_points = crit_points - {self.domain.start, self.domain.end}
        for pt in crit_points:
            val = self.f.subs(self.indVar, pt)
            point = SymPoint(pt, val)
            point.setCritical()
            self.points.append(point)
            if val == min_val:
                point.setMinimum()
            if val == max_val:
                point.setMaximum()

        end_points = [self.domain.start, self.domain.end]
        for pt in end_points:
            val = self.f.subs(self.indVar, pt)
            point = SymPoint(pt, val)
            point.setEndpoint()
            self.points.append(point)
            if val == min_val:
                point.setMinimum()
            if val == max_val:
                point.setMaximum()

        try:
            zeros_x = [r.evalf() for r in sp.real_roots(self.f) if self.domain.start <= r.evalf() <= self.domain.end]
        except:
            zeros_x = []
        for pt in zeros_x:
            point = SymPoint(pt, 0)
            point.setZero()
            self.points.append(point)

    def printPoints(self):
        for pt in self.points:
            print(pt.toString())

    def plotPoints(self, ax):
        self.initPoints()
        for pt in self.points:
            ax.scatter(pt.x, pt.y, color=pt.color(), zorder=5)
            ax.annotate(pt.toString(), (pt.x, pt.y), textcoords="offset points", xytext=(0, 10), ha='center', zorder=6)

    def plot(self, ax, color):
        x_vals = np.linspace(float(self.domain.inf), float(self.domain.sup), 400)
        y = self.f_num(x_vals)
        label = f"$f(x) = {sp.latex(self.f)}$"
        ax.plot(x_vals, y, color=color, label=label, zorder=3)
        ax.axhline(0, color='black', linewidth=1.2, zorder=2) # X-axis
        ax.axvline(0, color='black', linewidth=1.2, zorder=2) # Y-axis  
        self.plotPoints(ax)
        # ax.set_title(title)
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.legend()    

def plotsympy(ind_var, equation, plotzeros, plotcriticals, ax, color, label, title, x_vals):
    f_num = sp.lambdify(ind_var, equation, 'numpy')
    y = f_num(x_vals)
    y_masked = y
    x_masked = x_vals
    # mask = (y >= y_min) & (y <= y_max)
    # y_masked = y[mask]
    # x_masked = x_vals[mask]

    ax.plot(x_masked, y_masked, color=color, label=label, zorder=3)
    ax.axhline(0, color='black', linewidth=1.2, zorder=2) # X-axis
    ax.axvline(0, color='black', linewidth=1.2, zorder=2) # Y-axis    
    if plotzeros:
        zeros_x = [float(val) for val in sp.solve(equation, x)]
        zeros_y = [float(equation.subs(x, val)) for val in zeros_x]
        ax.scatter(zeros_x, zeros_y, color='green', zorder=5, label='Zeros')
        for z_x, z_y in zip(zeros_x, zeros_y):
            ax1.annotate(f'({z_x:.2f}, {z_y:.2f})', (z_x, z_y), textcoords="offset points", xytext=(0, 10), ha='center', zorder=6)
        

    if plotcriticals:
        f_prime = sp.diff(f,x)
        crit_x = [float(val) for val in sp.solve(f_prime, x)]
        crit_y = [float(equation.subs(x, val)) for val in crit_x]
        ax1.scatter(crit_x, crit_y, color='orange', marker='s', zorder=5, label='Critical Points')
        for c_x, c_y in zip(crit_x, crit_y):
            ax1.annotate(f'({c_x:.2f}, {c_y:.2f})', (c_x, c_y), textcoords="offset points", xytext=(0, -15), ha='center', zorder=6)
    
    ax.set_title(title)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()    

    
```

## Theorem (Second Derivative Test for Local Maximum / Minimum)

Suppose f'(a) = 0. If f''(a) > 0, then f has a local minimum at a (because f' is increasing), and if f''(a) < 0, then f has a local maximum at a.

Proof: 

If $f''(a)$ is positive, $f'$ is increasing at a. Since f'(a) = 0, then f'(a) < 0 before a, and f'(a) > 0 after a. This in turn means f is decreasing before a, and f is increasing after a. Thus f has a local minimum at a.

**Corollary (Partial converse)**

Suppose $f''(a)$ exists. If f has a local minimum at a, then $f''(a) \ge 0$. If f has a local maximum at a, then $f''(a) \le 0$.

Note: If a local minimum exists, it does not mean $f''(a)$ exists. Secondly, even if it exists, it may be 0 - hence this is a partial converse only. 

Proof: If f has a local minimum at a, and $f''(a) \lt 0$, then f would also have a local maximum at a! This means f is "constant" or flat at a, and $f''(a) = 0$, a contradiction.

## Theorem (Derivative of a limit)

Suppose 
a) f is continuous at a
b) f' is defined for a neighborhood U of a (excepting a itself) i.e. for $U \setminus \{a\}$. 
c) $\lim_{x \to a} f'(x) = L$, say,

then: $f'(a)$ exists and is equal to L.

Note: The implication is that f' cannot have a "jump" discontinuity! If it is discontinuous, it will be an "oscillatory" discontinuity.

Proof:
By definition:

$f'(a) = \lim_{h \to 0} \frac{f(a + h) - f(a)}{h}$. 

Step 1: Applying MVT

For h > 0, there is some h > 0, where f is differentiable in $(a,a+h]$ by hypothesis. This also means f is continuous on $[a,a+h]$, since we are give f is continuous at a.

By MVT then, there is an $\alpha_h$ in $(a, a+h)$, such that $\frac{f(a+h) - f(a)}{h} = f'(\alpha_h)$. It follows that 

Step 2: As $h \to 0$, $\alpha_h \to a$

As $h \to 0$ : given an $\epsilon$, there is a $\delta$ s.t., $|h| < \delta$. But since $\alpha_h \in (a, a+h)$, it follows that $|\alpha_h - a| \lt |h| \lt \delta$

Step 2: Take limiting value of $\alpha_h$

$f'(a) = \lim_{h \to 0} \frac{f(a + h) - f(a)}{h} = \lim_{h \to 0} f'(\alpha_h) = \lim_{\alpha_h \to a} f'(\alpha_h)$ = L. 

Note that, since $\lim_{h \to 0} f'(a+h) = L$ exists, given an $\epsilon > 0$, there is a $\delta > 0$ such that if $|(a+h) - a| = |h| \lt \delta$, then $|f(a+h) - L| \lt \epsilon$.

But whenever $|h| \lt \delta$, then $|\alpha_h - a| \lt \delta$, so $|f(\alpha_h) - f(a)| \lt \epsilon$. It follows that $\lim\limits_{\alpha_h \to a} f'(\alpha_h) = L$.




## Theorem [Cauchy Mean Value Theorem)

If f and g are continuous on $[a,b]$ and differentiable on $(a,b)$, then there is a number x in $(a,b)$ such that :

$[f(b) - f(a)]g'(x) = [g(b) - g(a)]f'(x)$

Proof:

Define : $h(x) = f(x)[g(b) - g(a)] - g(x)[f(b) - f(a)]$

h is continuous on $[a,b]$, and differentiable on $(a,b)$. And h(a) = f(a)g(b) - g(a)f(b) = h(b). Applying Rolle's Theorem, there is x in (a,b) such that $h'(x) = 0$ i.e. $f'(x)[g(b) - g(a)] = g'(x)[f(b) - f(a)]$.

## L'Hopital's Rule

Suppose 
a) that $\lim\limits_{x \to a} f(x) = 0$, and $\lim\limits_{x \to a} g(x) = 0$

b) $\lim\limits_{x \to a} \dfrac{f'(x)}{g'(x)}$ exists

Then : $\lim\limits_{x \to a} \dfrac{f(x)}{g(x)}$ exists and is equal to $\lim\limits_{x \to a} \dfrac{f'(x)}{g'(x)}$


**Proof**

The fact that $\lim\limits_{x \to a} \dfrac{f'(x)}{g'(x)}$ exists implies (a) f'(x) and g'(x) are defined in some interval $(a - \delta, a + \delta) \setminus \{a\}$ and (b) Since the ratio exists, $g'(x) \ne 0$ in this interval. 

We also assume f(a) = g(a) = 0 - this makes f and g continuous at a.

**Step 1: g(x) not zero in $(a,a+\delta)$**

If $a \lt x \lt a + \delta$, then if g(x) = 0, then g is continuous in the interval $[a,x]$ and differentiable in $(a,x)$. This would mean, by MVT, there is some $x_1$ in $(a,x)$ s.t. $g'(x_1) = 0$, violating the hypothesis that $\lim\limits_{x \to a} \dfrac{f'(x)}{g'(x)}$ exists.

The same result can be proven for $a - \delta \lt x \lt a$

**Step 2: Prove the ratio exists**

Applying Cauchy Mean Value Theorem to f and g using the same x as above, there is some $\alpha_x$ in (a,x), such that

$[f(x) - f(a)]g'(\alpha_x) = [g(x) - g(a)]f'(\alpha_x)$, i.e.

$f(x)g'(x) = g(x)f'(x)$, since f(a) = g(a) = 0

It follows that : $\dfrac{f(x)}{g(x)} = \dfrac{f'(\alpha_x)}{g'(\alpha_x)}$

Taking limits: $\lim\limits_{x \to a} \dfrac{f(x)}{g(x)} = \lim\limits_{x \to a} \dfrac{f'(\alpha_x)}{g'(\alpha_x)}$


Now, $\lim\limits_{x \to a} \dfrac{f'(x)}{g'(x)}$ exists, which implies that given an $\epsilon \gt 0$, there is a $\delta \gt 0$, such that if $|x - a| \lt \delta$, $|\frac{f'(x)}{g'(x)} - L| \lt \epsilon$. Now, for the same x, $|\delta_x - a| \lt |x - a|$, since $\delta_x \in (a,x)$. This implies $|\frac{f'(\alpha_x)}{g'(\alpha_x)} - L| \lt \epsilon$. Thus, 

$\lim\limits_{x \to a} \dfrac{f(x)}{g(x)} = \lim\limits_{x \to a} \dfrac{f'(\alpha_x)}{g'(\alpha_x)} = L$.


**Variations of L'Hopital's Rule**

a) If $\lim\limits_{x \to [A]} f(x) = B$, and $\lim\limits_{x \to [A]} g(x) = B$ and $\lim\limits_{x \to [A]} \dfrac{f'(x)}{g'(x)} = C$, then $\lim\limits_{x \to a^+} \dfrac{f(x)}{g(x)} = C$

Here 

- A = $a, a^+, a^-, \infty, -\infty$ - 5 choices
- B = $0, \infty, -\infty$ - 3 choices
- C = $l, \infty, -\infty$ - 3 choices

**Prob 6** : Prove the following: If f is increasing on (a, b) and continuous at a and b, then f is increasing on [a, b]. In particular, if f is continuous on [a, b] and f' > 0 on (a, b), then f is increasing on [a, b].

Proof #1: If f is increasing on (a,b), then it is increasing in (a, a+h). Since f is continuous at a, if a was greater than any given x in (a,a+h), then choosing an $\epsilon < a - x$, we could find points at $(a,a + \epsilon)$ where f is greater than f(x), which is not possible given f is increasing. So a is less that all x in (a,a+h), and hence in (a,b). We can make a similar argument for b. Thus f is increasing on [a,b].

Proof #2: Using IVT: If f(a) > f(x) for any x in (a,b), then by IVT, f attains all values from f(a) to f(x) in the interval [a,x]. This means there is some y in (a,x) s.t. f(a) > f(y) > f(x), a contradiction since f is increasing on (a,x). Thus f(a) < f(x) for all x in (a,b). 

For the second part: Since f' > 0 on (a,b) f is increasing. It follows that f is increasing on [a,b].
