# Three Hard Theorems


**Theorem 1 (Intermediate Value Theorem): If f is continuous on [a,b], and f(a) < 0 < f(b), then there is some x in [a,b] such that f(x) = 0**

Topological equivalent: The continuous image of a connected space is connected.

Intuitively : If you don't take the pen off the paper, then you must cross the x-axis somewhere when you go from f(a) to f(b) 

**Theorem 2 (Boundedness Theorem): If f is continuous on [a,b], then f is bounded above on [a,b]**

i.e. there is some N such that $f(x) \le N$ for all x in [a,b]

Note that it is easy enough to show that f is bounded below as well i.e. f is bounded on [a,b]. 

Topological equivalent: The continuous image of a compact space is compact.

**Theorem 3 (Extreme Value Theorem): If f is continuous on [a,b], then there is some number y in [a,b] such that $f(y) \ge f(x)$ for all x in [a,b]**

Basically, a continuous image of a compact space attains its bounds (it's not just bounded, it also attains the bound).
