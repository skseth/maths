# Inverse Functions

## Definitions

**one-one function** : A function f is **one-one** if $f(a) \ne f(b)$ whenever $a \ne b$.

**inverse function** : For any function f, the **inverse** of f, denoted by **$f^{-1}$**, is the set of all pairs $(a,b)$ for which the pair $(b,a)$ is in f.


## Theorem: A function f has an inverse if and only if it is one-one.

Proof: Assume that (a,b) and (a,c) are two pairs in $f^{-1}$. Then (b,a) and (c,a) are in $f$, which implies b = c. Thus $f^{-1}$ is a function.

**Consequences**

- $b = f(a)$ implies $a = f^{-1}(b)$
- $f(f^{-1}(x)) = x$
- $f^{-1}(f(x)) = x$
- $f \circ f^{-1} = I$
- $f^{-1} \circ f = I$ - may the domain of I may be larger than $f$ or $f^{-1}$


## Theorem: If f is continuous and one-one, then f is either increasing or decreasing on that interval

Proof:

We actually want to prove that given any two points a,b in the interval a < b, either f(a) < f(b), or f(a) > f(b) for all such pairs taken arbitrarily. To do this we have to first establish whether f is increasing or decreasing.

The proof is tricky because we have to prove a fact for 3 points, then for 4 points, and finally use this to show that the theorem is true for any given two points.

**Step 1**: Let a,b,c be any 3 distinct points in the interval, with a < b < c. We have to show that either f(a) < f(b) < f(c), or f(a) > f(b) < f(c). We have either f(a) < f(c), or f(a) > f(c) (equality is not possible since f is one-one). 

Suppose (f(a) < f(c)): Now assume f(b) > f(a). Then applying IVT in the interval $[b,c]$, there is some x in $(b,c)$ where f(x) = f(a), a contradiction, because f is one-one. So our hypothesis is wrong. In the same way we can prove for the case f(a) > f(c).

Thus given three distinct points in the interval, with a < b < c, either f(a) < f(b) < f(c), or f(a) > f(b) > f(c).

**Step 2** Let a < b < c < d in the interval, again we can prove that either f(a) < f(b) < f(c) < f(d) or f(a) > f(b) > f(c) > f(d).

We can use step 1 to prove f is increasing/decreasing between [a,c] and [b,d] i.e. it is increasing/decreasing on the entire interval [a,d].

**Step 3** Now, we take any pair of points $x, y$. Also fix 2 points a,b: a < b. Suppose f(a) < f(b) - i.e. f is increasing.

By taking the points $\{a,b,c,d\}$ in any increasing order, we can apply step 2, and show that f is increasing. Contrariwise, if f(a) > f(b), we can follow the same procedure to show f is decreasing.



## Theorem: If f is continuous and one-one on an interval, then $f^{-1}$ is also continuous

Proof:

Prove that $\lim\limits_{x \to b} f^{-1}(x) = f^{-1}(b)$, for each b in the domain of $f^{-1}$.

**Step 1**: If f is increasing on an interval so is $f^{-1}$ - and conversely.

Suppose f is increasing on some interval. This means given a and b on the interval, if a < b, then f(a) < f(b). Now suppose a' and b' are two points in the domain of $f^{-1}$, a' < b'. Note $f(f^{-1}(a')) = a' < f(f^{-1}(b')) = b$. This implies $f^{-1}(a') < f^{-1}(b')$, since f is increasing, and equality is not an option since both f and the inverse are one-one.

**Step 2** We start by noting that given $\epsilon > 0$. Also note that $f(a) = b$ - we want the limit of $f^{-1}$ at b.

Take the interval : $a - \epsilon < a < a + \epsilon$. Given f is increasing, then: $f(a - \epsilon) < f(a) < f(a + \epsilon$). 

We choose $\delta = \min  \{ f(a) - f(a - \epsilon),f(a + \epsilon$) - f(a) \}. Our choice of $\delta$ ensures that:

$f(a - \epsilon) < f(a) - \delta < f(a) < f(a) + \delta < f(a + \epsilon)$.

Now, if $f(a) - \delta < x < f(a) + \delta$, then $f(a - \epsilon) < x < f(a + \epsilon)$. Since $f^{-1}$ is also increasing, $f^{-1}(f(a - \epsilon)) < f^{-1}(x) < f^{-1}(f(a + \epsilon))$ i.e. 

$a - \epsilon < f^{-1}(x) < a + \epsilon$. This proves the theorem.

## Theorem (Derivative of inverse)

Let f be a continuous one-one function defined on an interval, and suppose f is differentiable at $f^{-1}(b)$, with derivative $f'(f^{-1}(b)) \ne 0$. Then $f^{-1}$ is differentiable at b, and :

$$(f^{-1})'(b) = \dfrac{1}{f'(f^{-1}(b)}$$

Note that the exceptions are the points where f has a **horizontal tangent**, which after reflection becomes a vertical one. In fact the slope of $f^{-1}$ at b is the reciprocal of the slope at the pre-image of b, which is not surprising since we are basically flipping the x and y axis. 

Note the sign of the slope stays the same - so the inverse of an increasing function is still increasing. But a function increasing quickly basically starts increasing slowly and vice versa. The most stunning example is $e^x$ and $ln(x)$. $e^x$ rises faster than any $x^n$, while $ln(x)$ rises slower than any $x^n$.

Proof:

**Step 1** We just show that the $f'(f^{-1}(b)) \ne 0$ condition is necessary

Support, to the contrary, that $f'(f^{-1}(b)) = 0$

$f'(f^{-1}(x)) = x$, by definition of f and the inverse. Suppose $f^{-1}$ were differentiable at b, then, by chain rule:

$f'(f^{-1}(x))f'^{-1}(x) = 1$. Thus, at b, $f'(f^{-1}(b))f^{-1} {'} (b) = 1$ which means $0.f^{-1}{'}(b) = 1$, which is absurd.

This is mainly because a horizontal tangent becomes, after reflection, a vertical one.

**Step 2** 

We want to evaluate $\lim_{h \to 0} \dfrac{f^{-1}(b + h) - f^{-1}(b)}{h}$. 

Let $b = f(a)$ i.e. $f^{-1}(b) = a$. 

And we also define a k for each h, such that b + h = f(a + k). Then $f^{-1}(b + h) = f^{-1}(a + k) = a + k = f^{-1}(b) + k$. Hence, 

$k = f^{-1}(b + h) - f^{-1}(b)$. As $h \to 0$, since $f^{-1}$ is continous, we get $k \to 0$ as well.

Now, we get:

$$
\lim_{h \to 0} \dfrac{f^{-1}(b + h) - f^{-1}(b)}{h} \\
= \lim_{h \to 0} \dfrac{a + k - a}{f(a + k) - f(a)} \\
= \lim_{k \to 0} \dfrac{k}{f(a + k) - f(a)} \\
$$

We are given that $\lim_{k \to 0} \dfrac{f(a + k) - f(a)}{k} = f'(a) = f'(f^{-1}(b)) \ne 0$

Hence : $f^{-1}{'}(b) = \lim_{k \to 0} \dfrac{k}{f(a + k) - f(a)} = \dfrac{1}{f'(f^{-1}(b))}$


## Parametric Representation of Curves

To represent curves on the real plane (e.g. circles etc), you need a parameterized reprentation.

e.g. 

$c:R \to R^2 : \vec{c}(t) = (u(t), v(t))$

$\vec{e}(t) = (cos(t), sin(t))$ - vector valued functions

The normal rules for adding works as expected.

Multiplying with a ordinary (non-vector valued function $\alpha$)

$(\alpha.\vec{c})(t) = \alpha(t)c(t) = (\alpha(t) u(t), \alpha(t) v(t))$ e.g.

$(\alpha.\vec{e})(t) = \alpha(t)\vec{e}(t) = (\alpha(t) cos(t), \alpha(t) sin(t))$ e.g.


Using polar coordinates, we can define:

$\vec{c}(t) = r(t).\vec{e}(\theta(t))$


We can define 2 scalar-valued functions:

$c \cdot d$ - dot product of two vector valued functions, giving a scalar

$det(c,d)$ - determinant of two vector valued functions



### Limits of vector valued functions

Given $c(t) = (u(t), v(t))$

$\lim\limits_{t \to a} c(t) = \lim\limits_{t \to a} (u(t), v(t)) = (\lim\limits_{t \to a} u(t), \lim\limits_{t \to a} v(t))$

$\lim\limits_{t \to a} c + d = \lim\limits_{t \to a} c + \lim\limits_{t \to a} d$

Limit of multiplication of a scalar function with a vector-valued function:

$\lim\limits_{t \to a} \alpha.c = \lim\limits_{t \to a} \alpha(t).\lim\limits_{t \to a} c$


### Derivatives of vector valued functions

Given $c(t) = (u(t), v(t))$

$c'(a) = (u'(a), v'(a))$ - follows from the definitions of limits above

$\text{tangent line} = c(a) + s.c'(a)$. Note that this tangent "line" does not make sense when c'(a) = (0,0). See Problem 1/2, and the issue with reparameterization when c'(a) = (0,0).

$(c + d)'(a) = c'(a) + d'(a)$, or, $(c + d)' = c' + d'$

$(\alpha.c)'(a) = \alpha'(a).c(a) + \alpha(a).c'(a)$, or $(\alpha.c)' = \alpha'.c + \alpha.c'$



### Composition - Chain Rule for vector function over a scalar function

If $d(t) = c(p(t)) = (c \circ p)(t)$, p is a scalar function of t.

$$d = (c \circ p)(t) = (u \circ p, v \circ p)$$

$$d' = (c \circ p)' = p'(c' \circ p)$$

Proof:

$$
\begin{aligned}
d'(a) &= (c \circ p)'(t) \\
    &= ((u \circ p)'(a), (v \circ p)'(a)) \\
    &= (p'(a)u'(p(a)), p'(a)v'(p(a)) \\
    &= (p'(a)u'(p(a)), p'(a)v'(p(a)) \\
    &= p'(a) (u'(p(a)), v'(p(a))
\end{aligned}
$$

### Derivative of Dot Product / Determinant

$$(c \cdot d)'(a) = c(a) \cdot d'(a) + c'(a) \cdot d(a)$$

$$[det(c,d)]'(a) = [det(c',d)](a) + [det(c,d')](a)$$

<b>Prob 2 (Reparameterization Hides a corner)</b>

Note that the graph of a function f(x) can be represented as the the curve (t, f(t)), and the derivatives will also match as expected here. 

But sometimes we can have strange results: Consider for example the function $c(f(t), t^2)$, where $f(t) = t^2 \text{ if } x \ge 0$, and $f(t) = -t^2 \text{ if } x \lt 0$

A little thought will show that this is the exact graph of $h(x) = |x|$. We know this curve has no derivative at x = 0. However, for f(t), f'(0) = 0, and $t^2$ also has derivative 0. So $c'(t=0) = (0,0)$ - instead of what we know for $h(x)$.

Why does this happen? The key is that the parameter acts like a velocity - when we say (t,|t|), we are basically moving along the parameter at constant "speed" of 1. When we hit a corner, things go wrong (as they should). But if we use the reparameterization, we are slowing down the parameter as we approach the "corner" at 0, till we are at a dead stop there. This hides the corner in effect.

**Prob 3 (derivative of curve specified by x = u(t), y = v(t))**

a) Assume u is one-one on some interval. Then the curve lies on the graph of $f = v \circ u^{-1}$.

b) If u is differentiable on this interval and $u'(t) \ne 0$, show that at the point $x = u(t)$ we have :
$$f'(x) = \dfrac{v'(t)}{u'(t)}$$

Or, in Leibniz notation:

$$\dfrac{dy}{dx} = \dfrac{\dfrac{dy}{dt}}{\dfrac{dx}{dt}}$$

c) We also have:
$$f''(x) = \dfrac{u'(t)v''(t) - v'(t)u''(t)}{(u'(t)^3)}$$

</div>

**Proof**

**a) The curve is on the graph of $f = v \circ u^{-1}$**

The curve is of the form: $(u(t), v(t)) = \left( u(t), v(u^{-1}(u(t))) \right) =  \left( u(t), v \circ u^{-1}(u(t)) \right)$

This curve is on the graph of $f = v \circ u^{-1}$ - here, implicitly we are talking of f(x), where x = u(t)

**b) Derivative $f'(x) = \dfrac{v'(t)}{u'(t)}$**

This is the standard chain rule.

$f' = (v \circ u^{-1})' = (v' \circ u^{-1})(u^{-1})'$

$(u^{-1})' = \dfrac{1}{u' \circ u^{-1}}$. Hence, 

$f' = \dfrac{v' \circ u^{-1}}{u' \circ u^{-1}}$

Then, $f'(x) = f'(u(t)) = \dfrac{v'(t)}{u'(t)}$, where x = u(t)

**c) f''(x)**

$f' = \dfrac{v' \circ u^{-1}}{u' \circ u^{-1}}$

$$
\begin{aligned}
f'' &= \dfrac{ (v'' \circ u^{-1})(u^{-1})'(u' \circ u^{-1}) - (v' \circ u^{-1})(u'' \circ u^{-1})(u^{-1})'}{(u' \circ u^{-1})^2} \\
    &= \dfrac{ (v'' \circ u^{-1})(u' \circ u^{-1}) - (v' \circ u^{-1})(u'' \circ u^{-1})}{(u' \circ u^{-1})^3} \\
f''(x)  &= \dfrac{ v''(t) u'(t) - v'(t) u''(t)}{(u'(t))^3} \\
\end{aligned}
$$
