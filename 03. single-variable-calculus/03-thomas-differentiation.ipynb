## Derivative as a limit

- derivative as a limit of the difference quotient

$$f'(x) = \lim_{h \to 0} \frac{f(x_0 + h) - f(x)}{h}$$

- the derivative is the **slope** of f(x) at $x_0$, the **rate of change** of f(x) at $x_0$.
- we can also define **left and right handed derivatives**. The derivative exists if the limits exist and are equal.

## Notations

if f is the function for which a derivative is being taken:

first derivative:
$f', \frac{df}{dx}, Df$

second derivative:
$f''$, $\frac{d^2f}{dx}$, $D^2 f$

nth derivative: $f^{(n)}$, $\frac{d^nf}{dx^n}$, $D^nf$



## When does a derivative not exist

- at a corner
- at a cusp
- a vertical tangent
- a jump or discontinuity
- a oscillation (e.g. sin(1/x)



## Theorem: Differentiable functions are continuous

**If f(x) has a derivative at c, then f is continuous at c. The converse need not be true, as can be seen with $y = |x|$, at x = 0**

$$
\begin{aligned}
f(c+h) &= f(c) + (f(c+h)-f(c)) \\
\lim_{h \to 0} f(c+h) &= \lim_{h \to 0} f(c) + \lim_{h \to 0} (f(c+h)-f(c)) \\
\lim_{h \to 0} f(c+h) &= f(c) + \lim_{h \to 0} \frac{(f(c+h)-f(c))}{h} \lim_{h \to 0} h \\
\lim_{h \to 0} f(c+h) &= f(c) + f'(c).0 = f(c) \\
\end{aligned}
$$

## Differentiation Rules

$$
\begin{aligned}
\frac{d}{dx} (c) &= 0 \text{ : c constant} \\
\frac{d}{dx} (x^r) &= rx^{r-1} \text{ : r real and constant} \\
\frac{d}{dx} (cu) &= c\frac{du}{dx} \text{ : c constant, u as function of x} \\
\frac{d}{dx} (u + v) &= \frac{du}{dx} + \frac{dv}{dx} \text{ : u,v as function of x} \\
\frac{d}{dx} (uv) &= u\frac{dv}{dx} + v\frac{du}{dx} \text{ : u,v as function of x} \\
\frac{d}{dx} (\frac{u}{v}) &= \frac{v\frac{du}{dx} - u\frac{dv}{dx}}{v^2} \text{ : u,v as function of x} \\
\end{aligned}
$$

## Trigonometric Functions

$$
\begin{aligned}
\frac{d}{dx} sin(x) &= cos(x) \\
\frac{d}{dx} cos(x) &= -sin(x) \\
\frac{d}{dx} tan(x) &= sec^2(x) \\
\frac{d}{dx} cot(x) &= -csc^2(x) \\
\frac{d}{dx} sec(x) &= sec(x)tan(x) \\
\frac{d}{dx} csc(x) &= -csc(x)cot(x) \\
\end{aligned}
$$

## Chain Rule

Let f.u be functions of x.

$$
\begin{aligned}
\frac{d}{dx} (f \circ u) &= \frac{df}{du}\frac{du}{dx}
\text{} \\
\text{Assume y = f(g(x))} \\
\frac{dy}{dx} &= f'(g(x)).g'(x) \\
\end{aligned}
$$

Important to note that f is differentiated at the point g(x), not x.

**Derivative of a power of a function**

Suppose $u = x + 1$ (as an example), then, by the chain rule: 

$$
\frac{du^n}{dx} = nu^{n-1}\frac{du}{dx}
$$

Finally, the answer would be $\frac{du^n}{dx} = n(x+1)^{n-1}.1 = n(x+1)^{n-1}$


## Implicit Differentiation



**$y^2 = x$**

Applying chain rule:

$2y\frac{dy}{dx} = 1$, which implies $\frac{dy}{dx} = \frac{1}{2y}$

Thus the derivatives are: $\frac{1}{2\sqrt{x}} \text{ and } -\frac{1}{2\sqrt{x}}$, corresponding to $y = \sqrt{x}$ and $y = -\sqrt{x}$




**$x^2 + y^2 = 25$**

Differentiating:

$2x + 2y\frac{dy}{dx} = 0 \implies \frac{dy}{dx} = -\frac{x}{y}$

**multiple differentiation**

$2x^3 - 3y^2 = 8$

Differentiating, we get: $6x^2 - 6yy' = 0$. Hence, $y' = \frac{x^2}{y}$

Differentiating again:
$y'' = \frac{y2x - x^2y'}{y^2} = \frac{2xy}{y^2} - \frac{x^2}{y^2}y' = \frac{2xy}{y^2} - \frac{x^4}{y^3}$

## Differentials and Linearization

We have discussed $\frac{dy}{dx}$ as a limit, while defining the derivative. 

However, terms like dy, dx also pop-up in a slightly different, but related context, not as a limit, but as real numbers approximating small changes in x, y etc. These are called **differentials**.

Suppose $y = f(x)$. We want to calculate $\Delta{y}$ for a small change in x - $\Delta{x}$ at some x. We can write:

$\Delta{y} = f(x + \Delta{x}) - f(x) = f'(x)\Delta{x} + \text{ ... other non-linear terms }$ - see the Taylor series expansion for a justification. 
 
We can say: $\Delta{y} \approx f'(x)\Delta{x}$ for a reasonable linear approximation.

### Differential

**Linearization of a function**

If f(x) is differentiable at x=a, then the function:

$L(x) = f(a) + f'(a)(x-a)$ is the linearization of f at a.

**Differentials**

We defined dy/dx earlier as the derivative, but now we introduce a new notation. 

Let y = f(x) is a differentiable function. The differential dx as an independent variable. We then define the differential dy as :

$$dy = f'(x)dx$$

**Estimating with Differentials**

We know 

$f(a + dx) = f(a) + \Delta{y}$, where we want to approximate $\Delta{y}$

The differential approximation then gives:

$f(a + dx) \approx f(a) + dy$

Note that $\Delta{y}$ is the actual change in y that we want to calculate. dy is the linear approximation and differs from the actual desired value by the higher order derivative terms of the Taylor series.

**Error Estimation**

$\Delta{f} = f(a + \Delta{x}) - f(a)$ - the true change

$df = f'(a)\Delta{x}$ - the approximate change

Error

$$
\begin{aligned}
error &= \Delta{x} - dx = f(a + \Delta{x}) - f(a) - f'(a)\Delta{x} \\
      &= \left( \frac{f(a + \Delta{x}) - f(a)}{\Delta{x}} - f'(a) \right)\Delta{x} \\
      &= \epsilon\Delta{x}
\end{aligned}
$$

$\epsilon \to 0$ as $\Delta{x} \to 0$, and of course $\Delta{x} \to 0$. So the error tends to 0, the closer x is to a.


**Notes from Higher Maths for Beginners**
At this point, it appears to be just a confusing use of terminology. By definition, $\frac{dy}{dx} = f'(x)$, even though here we are using dy and dx not using the limit definition, but as a true ratio. We have replaced the language of deltas by the language of differentials. But what does it get us?

Suppose we have $y = x^2$. Using deltas, we do :

$\Delta{y} = (x + \Delta{x})^2 - x^2 = 2x\Delta{x} + \Delta{x}^2$. Then $\frac{\Delta{y}}{\Delta{x}} = 2x + \Delta{x}$

Finally, we take limits, and get $y' = 2x$.  

Using differentials, we immediately drop any term with $dx^2$ or higher. So:

$dy = (x + dx)^2 - x^2 = 2xdx$ - having immediately dropped  all higher order terms in dx.

We can express, by definition of differential: $dy = 2x\Delta{x}$, while when using $\Delta{x}$, we get $dy = 2x\Delta{x}$

But an even more important property comes up if x, the independent variable here, is not quite indepedendent but a function of t e.g. $x = t^2$, so, when using deltas, we get $\Delta{x} = 2t\Delta{t} + \Delta{t}^2$. Substituting back, we get :

$dy = 2(t^2)(2t\Delta{t} + \Delta{t}^2) = 4t^3\Delta{t} + 2t^2\Delta{t}$, while what we really want, and get when using differentials:

$dy = 2xdx = 4t^3\Delta{t}$ 

Thus the language of differentials leads to a simple, and more correct linearization expression even when x it itself a function of some other variable.
