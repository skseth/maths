## The Problem of Measure

- Banach Tarski Paradox

Abandon goal of assigning a measure to every subset of $R^d$, and instead try and identify a certain subclass of subsets of R which can be assigned a measure without contradiction, called the "measurable sets".

With an elementary measure like Jordan measure, one is able to satisfy most common requirements for assigning a measure (length, volume etc) to subsets of $R^d$. But the Lebesgue measure works better in analysis, especially where limits are involved, in particular the "Dominated Convergence Theorem" and the "Monotone Convergence Theorem".

## Elementary measure

**Intervals on R** $I = [a,b], [a,b), (a,b], (a,b)$

**length of an interval** |I| = b - a

**box in $R^d$** cartesian product of intervals. In particular, an interval is a one-dimensional box

**volume of a box** $|B| = |I_1| \times |I_2| \times \dots \times |I_d|$

**elementary set** union of a finite number of boxes (including zero)

**Exercise: Boolean Closure of elementary**

If E,F are elementary in $R^d$, show that union, intersection, set difference and symmetric difference are all elementary sets. If $x \in R^d$, show that the translate E+x is also an elementary set.



a) Elementary sets  

Let $E = \bigcup\limits_m E_m$ and $F = \bigcup\limits_n F_n$, m, n finite, $E_m$, $F_n$ are boxes (note that in general E and F will not be boxes).

i) Union: E = union of finite number of boxes, F = union of finite number of boxes. It follows that E u F is also a union of a finite number of boxes and hence elementary.

ii) Intersection: $E \cap F$ = $\bigcup\limits_{i=0}^m \bigcup\limits_{j=0}^{n} E_i \cap F_j$

Since $E_i$ and $F_j$ are boxes there are two possibilities: If they don't intersect, then we get the empty set. If they do intersect, is their intersection a box? Yes - on intersection of two boxes $(I_1, I_2, \dots, I_d) \cap (J_1, J_2, \dots, J_d)$, their intersection is the box $(I_1 \cap J_1, I_2 \cap J_2, \dots, I_d \cap J_d)$. 

So the intersection is a union of multiple boxes or the empty set.

iii) Set Difference: Given two boxes E and E, E - F is not a box, but it is a union of smaller boxes. To see this, using the same notation as before, we split split each interval $I_i$ of E into 3 sub-intervals: $I_{i1}, I_{i2}, I_{13}$, where $I_{i2} = I_1 \cap J_1$. $I_{i1}$ or $I_{i3}$ may be empty, but if $I_{i2}$ is empty for any dimension, then the boxes don't intersect. Now we split E into $3^d$ boxes : only one box $(I_{12}, I_{22}, \dots, I_{d2})$ is not in E - F. So E-F is the union of $3^d - 1$ boxes, of which some may be empty. So the set difference is an elementary set.

Now, the rest is easy. We rewrite E - F as a union of differences in underlying boxes. Each of these is a (finite) union of boxes - hence the overall set difference is a finite union of boxes.

iv) Symmetric Difference: $A \Delta B = (A \cup B) - (A \cap B)$. We have already show that unions, intersections and set difference of elementary sets are elementary sets. So symmetric difference of elementary sets is also an elementary set.

## Measure of an elementary set

Let $E \subset R^d$ be an elementary set.

(i) E can be expressed as the finite union of disjoint boxes

(ii) If E is partitioned as the finite union $B_1 \cup B_2 \cup \dots \cup B_k$ of disjoint boxes, then the quantity 

$$m(E) = |B_1| + |B_2| + \dots + |B_k|$$

is independent of the partition. In other words given any other partition $B_{1'} \cup B_{2'} \cup \dots \cup B_{k'}$ of E. one has :

$$|B_1| + |B_2| + \dots + |B_k| = |B_{1'}| + |B_{2'}| + \dots + |B_{k'}|$$

We refer to m(E) as the **elementary measure** of E.

Proof (i) Given any E, we know E = union of (possibly overlapping) k boxes, where k is some finite number (including 0)

How do we make it disjoint? Looking at one particular dimension, n: Lets say we have $I_n1, I_n2, \dots, I_nk$ are k intervals in this space, possibly overlapping. Let $I = \bigcup I_ni$. 

Now we order the 2k endpoints of these intervals in strictly increasing order, and take the of these k' = 2k-1 intervals with I, to get $J_1, J_2, J_k'$ intervals (some intervals may be empty, which is fine). Clearly these new intervals are disjoint. We repeat this exercise in each dimension, take the cartesian product of the new disjoint intervals, and we get E as a union of disjoint sets.

(ii) We do a similar exercise for the two partitions. For a particular dimension, n: Let's say the first partition intervals are $I_{n1}, I_{n2}, \dots, I_{nk}$ and the second partition are $J_{n1}, J_{n2}, \dots, J_{nk}$. But note that since both are partitions of E, we must have:

$I = \bigcup I_{ni} = \bigcup J_{nj}$, summing over all intervals. We order the 2(k + k') endpoints of both partitions in increasing order (again the endpoints should match), but we ignore that: And for the 2(k + k') - 1 intervals we take an intersection with I. We get a refinement of I. We do this for each dimension, then take the cartesian product with similarly refined intervals of each dimension. It is obvious that this cartesian product (a) forms a partition of E (b) The sum of the volumes of boxes in this partition is equal to the sum of the volumes of both partitions. Hence m(E) is a unique sum of the disjoint partitions of E.

(ii) Book proof: Tao creates a grid of n/N points e.g. given N = 5, we get $\frac{1}{N}Z = \dots,-2/5,-1/5,0/5,1/5,2/5,3/5,4/5,5/5,6/5,\dots$

Suppose I is the interval $[0,1]$, then l(I) = 1. The expression $\frac{1}{N}\#(I \cap \frac{1}{N}Z) => 6/5$ when N = 5. In general, $(N-1)/N <= 1 <= (N+1)/N$, and given an interval of length l, $(N-1)l/N <= l <= (N+1)l/N$. As $N \to \infty$, the sum tends to the length.

In a sense, this is the ultimate extension of (i) and (ii). We simply "discretize" every interval into smaller and smaller chunks, and the two partitions end up with the same boxes. Given E in $R^d$:

$|B| = \lim\limits_{N \to \infty} \frac{1}{N^d}\#(E \cap \frac{1}{N}\mathscr{Z^d})$

Note that $\frac{1}{N}\mathscr{Z^d}$ is not literally  divided by 1/N - it is expressing a 1/N grid of points in d dimensions. It needs to treated as a single expression. So $\frac{1}{5}\mathscr{Z^3}$ will contain $5^3 = 125$ points in a 1x1x..1 box. In general there will be $N^d$ points in a 1x1x..1 box, hence the division by $\frac{1}{N^d}$.

In particular, if for the partition in the statement of the theorem, we get :

$|B_1| + |B_2| + \dots + |B_k| = \lim\limits_{N \to \infty} \frac{1}{N^d}\#(E \cap \frac{1}{N}\mathscr{Z^d})$

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

a = 0
b = np.pi # limits of integration
N = 1000

areas = []

trials = 10000

for i in range(trials):
    xrand = np.random.uniform(a,b, N)
    sin_rad = np.sin(xrand)
    integral = ((b - a)/float(N))*sin_rad.sum()
    areas.append(integral)

plt.title("Distribution of Areas Calculated")
# bins = np.linspace(min(areas), max(areas), 50)
plt.hist(areas, bins=30, ec='black')
plt.xlabel("Areas")
```

## Properties of Elementary Measure

$m(E) \ge 0$

$m(E \cup F) = m(E) + m(F)$, if E and F are disjoint (finite additivity), and by induction it follows for the disjoint union of any finite number of sets.

$m(\emptyset) = 0$ 

$m(B) = |B|$ for all boxes B i.e. it is consistent with our usual definition of length, area, volume etc.

**monotonicity** : Given elementary sets E,F $E \subset F$, then $m(E) \le m(F)$

**translation invariance** : m(E + x) = m(E). This can be easily seen by translating all the disjoint boxes by x and getting the same result. 

Note as we have discussed, these properties all work only on elementary sets. Any other sets, and all bets are off.


Proof: 
a) $m(E) \ge 0$. This follows from our definition of m(E) as the sum of the volume of boxes, which are themselves positive.

b) $F = E + (F - E)$, where E and F-E are disjoint. We have already proved that $F-E$ is an elementary set (during the boolean closure exercise). It follows that :

$m(F) = m(E) + m(F-E)$ or $m(E) = m(F) - m(F-E)$. From the non-negativity of measures, then we conclude that $m(E) \le m(F)$

***Exercise 1.1.3 (Uniqueness of elementary measure)***

Let $m': E(R^d) \to R^{+}$ be a map from the collection of all elementary sets $E(R^d)$ to the non-negative reals - such that m' is:

- finitely additive
- translation invariant
- non-negative

Show that there is a constant $c \in R^{+}$ such that m'(E) = cm(E) for all elementary sets E. In particular, if we impose $m'([0,1]^d) = 1$, them $m' \equiv m$

Ans:

**Step 1 (define c)**
Let $c = m'([0,1]^d)$. Note that by **non-negativity**, $c \ge 0$

**Step 2 (show $|(1/n)^d| = c/n^d$**

Now by **translation invariance**, $m'([0,1/n]^d) = m'([1/n,2/n]^d)$, etc, for all such translations in each of the d dimensions.

But these are disjoint $n^d$ boxes whose union is the box $[0,1]^d$. So $m'([0,1/n]^d) = c/n^d$ (short-circuiting some steps). The result follows from the **finite additivity** property of m'.

**Step 3 (Show all rational boxes have measures as expected)**

Now given a box B with rational endpoints, the length of all intervals is rational. Expressing the interval lengths in a common denominator ($p_1/q, p_2/q$ etc), we get the volume of $m(E) = \frac{p_1p_2\dots p_d}{q^d}$ which implies that :

$m(E) = |(0,1/q)^d|P where (P = p_1p_2\dots p_d)$ i.e E is the sum of P disjoint boxes, each of volume equal to |(0,1/q)^d|. But we have already shown that $m'((0,1/q)^d) = c/q^d$, so it follows that $m'(E) = \frac{c}{q^d}P = cm(E)$ for any rational box.

**Step 3 (Show all boxes have measures as expected)**

Now, we move on to any arbitrary box B. Note that given any 2 real numbers, I can find a rational number between them. I can use this to create inner and outer rational intervals for each interval of the box. Suppose we are given an $\epsilon \gt 0$.

For each side of length $x_i$, i have a scaling factor s < 1, such that i choose outer and inner rational numbers so that :

$x_i(1 - s) < \text{ length of ith interval (rational) inner box } \le x_i \le \text{ length of ith interval of (rational) outer box } <= x_i(1 + s)$

Then $|\text{Volume of outer box in m'}| \lt c|B|(1 + s)^d = cm(B)(1 + s)^d$

$|\text{Volume of inner box in m'}| \gt c|B|(1 - s)^d = cm(B)(1 - s)^d$

Taking the difference, we have |Inner-outer volume difference in m'| \le cm(B)((1+s)^d - (1 - s)^d))

We can express $(1 + s)^d - (1 - s)^d = 2s[\binom{d}{1} + \binom{d}{2}s + \text{ upto d/2+1 odd terms})] \lt 2s[\binom{d}{1} + \binom{d}{2} + \text{ upto d/2+1 odd terms})] = 2sM$, where M is the term inside the bracket. Note that we have used the fact that s < 1. 

Then it follows that |Inner-outer volume difference in m'| $\lt |B|2sM$. Choosing $s = \epsilon/2Mm(B)c$, we basically show that we can reduce the difference less than any given $\epsilon$. Also, as $s \to 0$, volume of either box in m' becomes $cm(B)$. Thus, $m'(B) = c.m(B)$ for any arbitrary box.

**Step 3 (Show all elementary sets have measures as expected)**

Let E be an elementary set. Then $E = B_1 \cup B_2 \dots B_k$, where the $B_i$ are boxes, and 

$m(E) = m(B_1) + m(B_2) + \dots + m(B_k)$

But we also have $m'(E) = m'(B_1) + m'(B_2) + \dots + m'(B_k) = cm(B_1) + cm(B_2) + \dots + c.m(B_k) = cm(E)$

**Step 4 (c = 1)**

If c = 1, it follows that $m'(E) = m(E)$ for any elementary set.


## Exercise 1.1.4 (Product of Elementary sets is Elementary, and measure is the product too)

Let $d_1,d_2 \ge 1$. Let $E_1 \subset R^{d_1}, E_2 \subset R^{d_2}$ be elementary sets. Show that $E_1 \times E_2 \in R^{d_1 + d_2}$ is elementary, and $m^{d_1 + d_2}(E_1 \times E_2) = m^{d_1}(E_1) \times m^{d_2}(E_2)$

Proof:

Every elementary set can be expressed as a union of disjoint boxes. Hence, let us express :

$E_1 = B_1 \cup B_2 \cup \dots B_k$ and $E_2 = C_1 \cup C_2 \dots \cup C_k'$ and $m(E_1) = \sum_{i} |B_1|$, $m(E_2) = \sum_{i} |C_i|$

Since every box in each group is disjoint, so is their product. 

$E_1 \times E_2 = \bigcup\limits_{1 \le i \le k, 1 \le j \le k'} B_i \times C_j$

$B_i \times C_j$ is clearly a "box" - hence $E_1 \times E_2$ is a finite union of boxes, and is therefore an elementary set.

$$
\begin{aligned}
m(E_1 \times E_2) &= \sum\limits_{1 \le i \le k, 1 \le j \le k'} |B_i \times C_j| \\
    &=  \sum\limits_{1 \le i \le k, 1 \le j \le k'} |B_i| |C_j| \\
    &=  \sum\limits_{1 \le i \le k} |B_i| \left(\sum\limits_{1 \le j \le k'}  |C_j| \right) \\ 
    &= \sum\limits_{1 \le i \le k} |B_i| m^{d_2}(E_2) \\
    &= m^{d_2}(E_2) \sum\limits_{1 \le i \le k} |B_i| \\
    &= m^{d_1}(E_1) \times m^{d_2}(E_2)
\end{aligned}
$$.

# Jordan Measure

Inspired by Archimedes's method of exhaustion:

**Definition (Jordan Measure)**

Let $E \subset R^d$ be a bounded set.

The *Jordan inner measure* $m_{*,(J)}(E)$ of E is defined as :

$$m_{*,(J)}(E) = \sup\limits_{A \subset E, A \text{ elementary}} m(A)$$

The *Jordan outer measure* $m_{*,(J)}(E)$ of E is defined as :

$$m^{*,(J)}(E) = \inf\limits_{B \supset A, A \text{ elementary}} m(B)$$

If $m_{*,(J)}(E) = m_{*,(J)}(E) = m(E)$, we say E is *Jordan measurable*, and call m(E) the *Jordan measure* of E.

By convention unbounded sets will not be considered Jordan measurable - they will be deemed to have infinite *Jordan outer measure*.
