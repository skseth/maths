# Continuity & Derivative

## Continuity

A function f(x) is said to continuous at x = a, if given any $\epsilon > 0$, no matter how small, there exists a $\delta > 0$, such that given $|x - a| < \delta$, $|f(x) - f(a)| < \epsilon$.

Note: In other words, there is a small region around x where f is bounded within $\epsilon$ of f(x)

## Derivative

The derivative of a function f(x) :

$$\frac{df}{dx} = \lim_{\Delta{x} \to 0} \frac{f(x + \Delta{x}) - f(x)}{\Delta{x}}$$

Equivalent Notations: $\frac{df}{dx} = f'(x) = f^{(1)} = Df$

You can take the derivative of derivative.

$\frac{d^2f}{dx^2} = f''(x) = f^{(2)} = D^2f$

### Derivative is a Linear operator

$D(a.f(x) + b.g(x)) = a.Df + b.Dg$

### Derivative of a product

$D(fg) = gDf + fDg$


### Chain Rule

$Df(u(x)) = \frac{df}{du}\frac{du}{dx}$

Can be proved easily, but you have to handle the case where $\Delta{x}$ may be zero.

[Proof of Chain Rule](https://www2.math.uconn.edu/~stein/virtual/Notes/Calculus/chainrule.pdf)

## Problem 1.2.1 - Demonstrate product / chain rule from first principles

**Product Rule**
$$
\begin{aligned}
D(fg) &= \lim_{\Delta{x} \to 0} \frac{f(x+\Delta{x})g(x + \Delta{x}) - f(x)g(x)}{\Delta{x}} \\
\\
\text{The *trick*} \\
     &= \lim_{\Delta{x} \to 0} \frac{f(x+\Delta{x})g(x + \Delta{x}) - f(x+\Delta{x})g(x) + f(x+\Delta{x})g(x) - f(x)g(x)}{\Delta{x}} \\
     &= \lim_{\Delta{x} \to 0} g(x)\frac{f(x + \Delta{x}) - f(x)}{\Delta{x}} + f(x + \Delta{x})\frac{g(x + \Delta{x}) - g(x)}{\Delta{x}} \\
     &= g(x)f'(x) + f(x)g'(x) \\
\end{aligned}
$$

**Chain Rule**
$$
\begin{aligned}
D(f(u(x)) &= \lim_{\Delta{x} \to 0} \frac{f(u(x + \Delta{x})) - f(u(x))}{\Delta{x}} \\
          &= \lim_{\Delta{x} \to 0} \frac{f(u(x + \Delta{x})) - f(u(x))}{u(x + \Delta{x}) - u(x)}\frac{u(x + \Delta{x}) - u(x)}{\Delta{x}} \text{ assuming } u(x + \Delta{x}) \ne u(x) \\
          &= \lim_{\Delta{x} \to 0} \frac{f(u(x + \Delta{x})) - f(u(x))}{u(x + \Delta{x}) - u(x)} \lim_{\Delta{x} \to 0} \frac{u(x + \Delta{x}) - u(x)}{\Delta{x}} \\
\text{ let } u(x + \Delta{x}) = u(x) + k \\
          &= \lim_{\Delta{x} \to 0} \frac{f(u+k) - f(u(k))}{k} \lim_{\Delta{x} \to 0} \frac{u(x + \Delta{x}) - u(x)}{\Delta{x}} \\
          &= \lim_{\Delta{k} \to 0} \frac{f(u+k) - f(u(k))}{k} \lim_{\Delta{x} \to 0} \frac{u(x + \Delta{x}) - u(x)}{\Delta{x}} \\
          &= \frac{df}{du}\frac{du}{dx} \\
\end{aligned}
$$

What if $u(x + \Delta{x}) = u(x)$ for sufficiently small $\Delta{x}$, then $f(u(x + \Delta{x})) - f(u(x)) = 0$, which implies the chain rule is true trivially, the derivative being zero.

## Appendix from Rudin: $e^x$ - Exponential Function

From Baby Rudin (pg 178)

Define $E(x) = \sum_{n=0}^{\infty} \frac{x^n}{n!}$

By ratio test (see reference) for series, we see it converges:

$$
|\frac{a_{n+1}}{a_n}| = |\frac{x^{n+1}.n!}{x^n.(n+1)!}| = \frac{|x|}{n+1}
$$

As limit is zero, we know by the ratio test that the series converges **absolutely**.

**Exponential function follows the power rule: $E(x)E(y) = E(x + y)$**

To prove this, we note that by Merten's Theorem, if two series $\sum_{n=0}^{\infty} a_n$, and $\sum_{n=0}^{\infty} b_n$ converge to A and B respectively (with at least one converging absolutely), then the sum of their cauchy product $c_n = \sum_{k=0}^{n} a_kb_{n-k}$ also converges to AB i.e. $\sum_{n=0}^{\infty} \sum_{k=0}^{n} a_kb_{n-k} = AB$. In our case:

$$
\begin{aligned}
E(x)E(y) &= \sum_{n=0}^{\infty} \frac{x^n}{n!}\sum_{m=0}^{\infty} \frac{y^m}{m!} \\
\text{ Using merten's theorem }: \\
        &= \sum_{n=0}^{\infty} \sum_{k=0}^{n} \frac{x^k}{k!}\frac{w^{n-k}}{(n-k)!} \\
        &= \sum_{n=0}^{\infty} \frac{1}{n!} \sum_{k=0}^{n} \binom{n}{k} x^k.w^{n-k} \\
        &= \sum_{n=0}^{\infty} \frac{(x+w)^n}{n!} \\
        &= E(x + y)
\end{aligned}
$$

**E(x) is positive,increasing monotonically**

E(x) > 0, for x > 0 (by definition), since every term is greater than 0.
E(x)E(-x) = 1 => E(-x) = 1/E(x) > 0.

As $x \to \infty, E(x) \to \infty$, by definition of E(x). Since E(-x) = 1/E(x), it follows that $x \to 0, E(x) \to -\infty$.

For $0 < x < y$, $E(x) < E(y)$, by definition. It follows that $E(-y) < E(-x)$, thus E(x) is strictly increasing across R.


**E(x) is differentiable**

$DE(x) = \lim_{\Delta{x} \to 0} \frac{E(x + \Delta{x}) - E(x)}{\Delta{x}} = \lim_{\Delta{x} \to 0} \frac{E(x)E(\Delta{x}) - E(x)}{\Delta{x}} =  E(x)\lim_{\Delta{x} \to 0} \frac{1 + \Delta{x} + .. - 1}{\Delta{x}} = E(x)$

Thus the derivative of E(x) is E(x) itself.

**Defining e and e^n**

We define $e = E(1) = \sum_{n=0}^{\infty} \frac{1}{n!}$

We then see that $e^n = e.e...$, n times.  i.e. $e^n = E(1).E(1)...E(1) = E(n)$, as we have shown earlier.

**Defining $e^p$, p rational**

let $p = \frac{n}{m}$, n > 0, m > 0

$(E(p))^m = E(mp) = E(n) = e^n$. Taking the mth root on both sides:

$E(p) = e^{\frac{n}{m}} = e^p$

Now, E(x)E(-x) = E(0) = 1. it follows that $E(p)E(-p) = e^pE(-p) = 1$. It follows:

$E(-p) = \frac{1}{e^p} \implies e^{-p} = E(-p)$, so now we have power of e for all rational p.

**Define $e^x$, x real**

We define $e^x = \sup_{x > p \in Q,}{e^p} = \sup_{x > p \in Q,}{E(p)}$

We say the supremum is E(x). First, because e is differentiable (and hence continuous) on R, E(x) exists and is greater than 0. 
Since E is strictly increasing, E(x) > E(p), for all $p < x$. Lastly if there is a q < x, such that $\sup_{x > p \in Q,}{E(p)} = E(q)$, we can pick a $q < q' < x$, such that E(q') > E(q), contradicting our assumption. Hence:

$e^x = E(x)$, for all real x.

**$e^x$ rises faster than any power of x**

$lim_{x \to +\infty} \frac{x^n}{e^x} = 0$, for any n.

Proof:

By definition, $e^x > \frac{x^{n+1}}{n+1!}$, for x > 0.

It follows: $x^ne^{-x} < \frac{(n+1)!}{x}$. The limit result follows.

# Appending from Rudin : log(x) - the logarithm

$e^x$ is strictly increasing and differentiable on all of R. Hence the inverse L is differentiable, strictly increasing on its domain of positive reals as well. $ln(x) \to \infty, \text{ as } x \to \infty, ln(x) \to -\infty, \text{ as } x \to 0$

Let $L(x)$ be the inverse of $e^x$ i.e. $L(e^x) = x$.

The domain of L is the set of all positive real numbers, since $0 < e^x < \infty$. The range is R itself. It follows:

$E(L(y)) = y$, or $L(E(x)) = x$.

Differentiating, applying the chain rule:

$L'(E(x))E'(x) = L'(E(x))E(x) = 1$. Setting E(x) = y, we get $L'(y) = \frac{1}{y}$, y > 0

We get $L(E(0)) = L(1) = 0$. We can then define:

$$L(y) = \int_{1}^{y} \frac{dx}{x}$$

Writing $u = E(x), v = E(y)$. Then:

$$L(uv) = L(E(x)E(y)) = L(E(x+y)) = x + y = L(u) + L(v)$$



**Definition of $x^n$, n integer**

$$x^n = E(L(x)) + E(L(x)) ... = E(nL(x)), \text{ assuming $x > 0$, n integer }$$

Given $E(\frac{1}{m}L(x))^m = E(L(x)) = x$. Hence:

$$x^{\frac{1}{m}} = E(\frac{1}{m}L(x))$$

**Definition of $x^p$, p rational**

$$x^p = E(pL(x)) = e^{plog(x)}$$

**Definition $x^r$, r real, r > 0 (Finally)**

By defining $x^r = \sup x^p = \sup_{r > p \in Q} E(pL(x)) = E(rL(x)) = e^{rlog(x)}$. The result follows because E / L are monotonically increasing and continuous.

**Differentiation of $x^r$**

$$D(x^r) = E(rL(x)).L'(x) = x^r.\frac{r}{x} = rx^{r-1}$$
