# Linear Equation Systems

```python
from scipy import linalg as la
from scipy import optimize
import sympy
sympy.init_printing()
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = sympy.Matrix([[2,3], [5, 4]])
b = sympy.Matrix([4,3])
display(A.rank())
display(A.condition_number())
display(A.norm())
```

```python
A = np.array([[2, 3], [5, 4]])
b = np.array([4, 3])
display(np.linalg.matrix_rank(A))
display(np.linalg.cond(A))
display(np.linalg.norm(A))
```

# LU Factorization

```python
A = sympy.Matrix([[2,3], [5, 4]])
b = sympy.Matrix([4,3])
L, U, _ = A.LUdecomposition()
display(L)
display(U)  
display(L*U)
display(A.solve(b))

```

```python
A = np.array([[2, 3], [5, 4]])
b = np.array([4, 3])
P, L, U = la.lu(A)
display(L)
display(U)
display(P)
P.dot(L.dot(U))
la.solve(A, b)
```
