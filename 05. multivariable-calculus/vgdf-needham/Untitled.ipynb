# Tensors

A tensor is a *multilinear* function of multiple vector and co-vector inputs.

**Definition (Tensor, Valence)** 

A **tensor H of valence $\begin{Bmatrix} \mathscr{f} \\ \mathscr{v} \end{Bmatrix}$** at a point p is a multilinear, real-valued function of $\mathscr{f}$ 1-forms, and $\mathscr{v}$ vectors, such that the value of $\boldsymbol{H}$ depends only on the input values.


We can denote the tensor H as :

$$\boldsymbol{H}(\phi_1, \phi_2, \dots,\phi_f \Vert \mathbf{v_1}, \mathbf{v_2}, \dots, \mathbf{v_{\mathscr{v}}})$$

In general, the order of vectors being fed in matters, though there are different types of tensors. 

Examples:

In a **symmetric tensor** (e.g. a dot product), the order of components is immaterial. 

In an **anti-symmetric tensor** (e.g a determinant is a classic example, which represents oriented volume), the flipping the order of two components reverses the sign.

Linear algebra is the study of tensors $\boldsymbol{L}(\phi \Vert \mathbf{v})$, of rank $\{1,1\}$.



## Operations on Tensors

**Addition of Tensors**

$$(\boldsymbol{H} + \boldsymbol{J} )(\phi_1, \phi_2 \Vert \mathbf{v_1}) = \boldsymbol{H}(\phi_1, \phi_2 \Vert \mathbf{v}) + \boldsymbol{J}(\phi_1, \phi_2 \Vert \mathbf{v})$$

**Tensor Product (Multiplication) of Two 1-Forms $\begin{Bmatrix} 0 \\ 2 \end{Bmatrix}$**

Given two 1-forms, $\phi$ and $\psi$, we define their tensor product (acting on two vectors, and hence of valence $\begin{Bmatrix} 0 \\ 2 \end{Bmatrix}$, as follows:

$$\boxed{\quad (\phi \oplus \psi)(\mathbf{v}, \mathbf{w}) = \phi(\mathbf{v})\psi(\mathbf{w}) \quad \text{tensor product}}$$

Note that order matters.

Also called **direct product** or **outer product**

**Tensor Product (Multiplication) of Higher Valences $\begin{Bmatrix} \mathscr{f} \\ \mathscr{v} \end{Bmatrix}$**

Extending the concept, we define a tensor product of a tensor of valence: $\begin{Bmatrix} \mathscr{f} \\ \mathscr{v} \end{Bmatrix}$ as :

$$(\boldsymbol{J} \oplus \boldsymbol{T})(\phi, \psi \Vert \mathbf{u}, \mathbf{v}, \mathbf{w}) \equiv \boldsymbol{J} \oplus \boldsymbol{T})(\phi, \psi \Vert \mathbf{u})  \boldsymbol{T}(\mathbf{v}, \mathbf{w})$$

## Back to 1-forms

Suppose we have a  tensor $T(\mathbf{v}, \mathbf{w})$, of valence $\begin{Bmatrix} 0 \\ 2 \end{Bmatrix}$.

Let us define : $T_{ij} = T(\mathbf{e_i}, \mathbf{e_j})$. Since the unit vectors are fixed, these are effectively constant values for this tensor.

Now we can decompose $T(\mathbf{v}, \mathbf{w})$:

$$
\begin{aligned}
\boldsymbol{T}(\mathbf{v}, \mathbf{w}) &= \boldsymbol{T}(v^i\mathbf{e_i}, w^j\mathbf{e_j}) \quad \text{... this are  two summations} \\
    &= v^iw^j\boldsymbol{T}(\mathbf{e_i}, \mathbf{e_j}) \quad \text{... this is a double summation} \\
    &= T_{ij}v^iw^j \\
    &= T_{ij}[dx^i(\mathbf{v})][dx^j(\mathbf{w})] \\
    &= T_{ij}(dx^i \oplus dx^j)(\mathbf{v}, \mathbf{w}) \\
\end{aligned}
$$

So we can write:

$\boxed{ \boldsymbol{T} = T_{ij}(dx^i \oplus dx^j) \quad \text{ cartesian duals are basis vectors for tensors of valence } \begin{Bmatrix} 0 \\ 2 \end{Bmatrix}} $

In particular, given two one forms $\phi$ and $\psi$, we get :

$$\boxed{ \phi \oplus \psi = \phi_i\psi_j(\boldsymbol{d}x^i \oplus \boldsymbol{d}x^j) \quad \text{ not true for all tensors }}$$

But equally, we can take the tensor product of two vectors:

$$
\begin{aligned}
(\mathbf{v} \oplus \mathbf{w})(\phi, \psi) &= \mathbf{v}(\phi)\mathbf{w}(\psi) \\
    &= v^iw^j(\mathbf{e_i} \oplus {\mathbf{e_j}})(\phi, \psi) \\
\end{aligned}
$$

Extending this technique, we can get for tensors of arbitrary valence $\begin{Bmatrix} \mathscr{f} \\ \mathscr{v} \end{Bmatrix}$,

$$(e_{i_1} \oplus e_{i_2} \dots \oplus e_{i_{f}}) \text{ } \oplus (dx^{j_1} \oplus dx^{j_2} \dots dx^{j_v})$$

Note that the basis vectors are "reversed" - the f "covariant slots" have the standard cartesian basis vectors, and the contravariant slots have the dual basis vectors.
