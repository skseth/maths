# Basis

- a basis is a set of vectors that is linearly independent and spans V
- criteria for a basis: Every vector can be written **uniquely** as a linear combination of basis. The uniquely is the core point.

Thm: Every spanning list contains a basis

Cor: Every finite-dimensional vector space has a basis

Cor: Every linearly independent list extends to a basis

Application: Every subspace of V is part of a direct sum equal to V - i.e you can find another subspace so adding (direct sum) the two will give V



# Dimension

- basis length does not depend on basis, but only on (finite-dimensional) vector space
- dim V definition
- dimension of a subspace U : dim U <= dim V
- linearly independent list of right length is the basis
- subspace U of V: dim U = dim V => U = V
- spanning list of the right length is a basis (and hence also linearly independent?)
- dimension of sum : dim(V1 + V2) = dim(V1) + dim(V2) - dim(V1 n V2)
