# Linear Maps

# Vector Space of Linear Maps

- defn: space of linear maps has additivity and scalar multiplication

- L(V,W), L(V) : set of all linear maps from V to W, or from V to V.

- Examples: 0, identity, differentiation, integration, backward shift(?), from F^n to F^m, polynomial composition

- Lemma: Given basis for V and W, there is a unique map T from V to W, which maps one basis to the other

- L(V,W) is a vector space

- Product of linear maps as composition : T in L(U,V) and S in L(V,W). Then $S \circ T$ is a linear map in L(U,W), written as ST

- product is associative, has identity and is distributive

- Linear maps always map 0 to 0 - y = mx + b is not a linear map


# Null Spaces

- For T in L(V,W), the null space of T is the subset of V consisting of those vectors that map to 0.
- Given T in L(V,W): Null space is a subspace of V
- Injectivity of linear map <=> null space is {0}
- Given T in L(V,W): Range T is a subspace in W

**Thm: Fundamental Theorem of Linear Maps**  
Given V is finite-dimensional and T in L(V,W): Then range T is finite-dimensional and :  
dim V = dim null T + dim range T


Cor: Given L(V,W) where dim V > dim W. Then no linear map is injective   
Cor: Given L(V,W) where dim V < dim W. Then no linear map is surjective  
Cor: **Homogenous system of linear equations with more variables than equations has no solution**
Cor: **Homogenous system of linear equations with more equations than variables has no solution for come choice of the constant terms**


# Matrices

- Defn: matrix (m x n) - m rows, n columns
- Defn: matrix of a linear map (represents the maped basis vectors, for each basis vector in domain). So it is always tied to a specific basis?

Examples
- matrix of a linear map from F2 to F3
- matrix of the differentiation map from P3(R) to P2(R) (polynomials)

Addition and Scalar Multiplication of matrices

- matrix addition
- matrix of the sum of linear maps (both need to have same domain and range)
- scalar multiplication
- Matrix of a scalar times a linear map
- $F^{m,n}$ - set of all mxn matrices
- dim $F^{m,n}$ = mn
- matrix multiplication
- matrix of a product of linear maps

Notation for rows and columns of a matrix: $A_{j,.}$ - jth row. $A_{.,k}$ - kth column

then, given AB : A = m x n, B = n x p, AB => m x p and b = n x 1 matrix:

$(AB)_{j,k} = A_{j,.}B_{.,k}$

$(AB)_{.,k} = A B_{.,k}$ - kth column of AB is the linear combination of each column vector of A with the coefficients of the kth column vector in B

It follows that if A represents the linear map S, then Ab is equivalent to S(b)
