# Inner Products, Dot Products and Norms

$\Vert v \Vert = \sqrt{a^2 + b^2}$ - the norm is not linear

To inject linearity, we create dot product:

$\Vert x \cdot y \Vert = x_1y_1 + x_2y_2 + \dots x_ny_n$

Properties of the dot product:

- x.x >= 0 for all
- x.x = 0 <=> x = 0
- For a given (fixed) y define Sf(x) = x.y$. Then f is linear map (functional) from R^n to R.
- x.y = y.x

An inner product is a generalization of a dot product.

### Complex Dot Product

In C, the absolute value of z is $|z| = \sqrt{a^2 + b^2}$. Then:

$|z|^2 = z\bar{z}$ because $(a + bi)(a - bi) = \sqrt{a^2 + b^2}$


For $z = (z_1, z_2, \dots, z_n) \in C^n$, we define:

$|z| = \sqrt{|z_1|^2 + |z_2|^2 + \dots, |z_n|^2}$. Then it follows:

$|z|^2 = z_1\bar{z_1} + z_2\bar{z_2} + \dots + z_n\bar{z_n}$

And we define the inner product of w with z as: $w_1\bar{z_1} + w_2\bar{z_2} + \dots + w_n\bar{z_n}$

and $\langle w, z \rangle = w_1\bar{z_1} + w_2\bar{z_2} + \dots + w_n\bar{z_n} = \overline{\langle z, w \rangle}$


Note that if w,z are real, this reduces to the dot product.

## Inner Product

An inner product on V is a function that takes each ordered pair $(u, v)$ of elements of V to a number $\langle u, v \rangle$ in F, and has the following properties (given any u,v,w in V) and $\lambda \in F$:

**positivity** $\langle v, v \rangle \ge 0$ for all $v \in V$ - note this means it must be in R specifically

**definiteness** $\langle v, v \rangle = 0$ iff $v = 0$

**additivity (in the first slot)** $\langle u + v, w \rangle = \langle u, w \rangle + \langle v, w \rangle$  

**homogeneity (in the first slot)** $\lambda\langle u + v \rangle = \langle \lambda u + v \rangle$

**conjugate symmetry** 
$\langle u, v \rangle = \overline{\langle v, u \rangle}$


Note that the inner product in the complex space can be understood in terms of waves - the innerproduct measures how much two waves are *in phase*.


**Examples of inner product**  

*Euclidean Inner Product* : As before : $\langle (w_1, w_2,\dots,w_n), (z_1,z_2,\dots,z_n) \rangle = w_1\bar{z_1} + w_2\bar{z_2} + \dots + w_n\bar{z_n}$

*Weighted Inner Product*: If $c_1, c_2, \dots, c_n$ are positive numbers, and inner product can be defined on $F^n$ as $\langle (w_1, w_2,\dots,w_n), (z_1,z_2,\dots,z_n) \rangle = c_1w_1\bar{z_1} + c_2w_2\bar{z_2} + \dots + c_nw_n\bar{z_n}$

*Inner Product for Continuous Real functions on [-1,1]* : $\langle f,g \rangle = \int_1^1 fg$

*Inner Product on P(R)*: $\langle p,q \rangle = p_0q_0 + \int_1^1 pq$ - two polynomials considered close by this inner product are close at the origin, and vary closely together 

*Inner Product on P(R)*: $\langle p,q \rangle = \int_{0}^{\infty} p(x)q(x)e^{-x}dx$ - weighted continuous inner product over an infinite interval - apparently used in the definition of Laguerre polynomials which come up in QM. Since $e^{-x}$ falls off faster than any polynomial, this is used as a damping factor.

### Inner Product Space

This is a vector space and an inner product on it.

Properties:

(a) For each fixed $v \in V$, the function that takes $u \in V$ to $\langle u,v \rangle$ is a linear map from V to F

(b) $\langle 0,v \rangle = 0$ for every $v \in V$

(c) $\langle v,0 \rangle = 0$ for every $v \in V$

(d) $\langle u, v + w \rangle = \langle u, v  \rangle + \langle u, w  \rangle$ for every $u,v,w \in V$

(e) $\langle u, \lambda v \rangle = \bar\lambda\langle u, v  \rangle$ for all $\lambda \in F$ and every $u,v \in V$

Proof:

(a) Since v is the second parameter (which is the conjugate), it follows that $\langle u + w,v \rangle = \langle u, v \rangle + \langle w,v \rangle$ given the additivity and homogeneity of the first slot. So the function is linear.

(b) Follows because this is a linear map, always maps 0 to 0

(c) We use $\langle u, 0 \rangle = \overline{\langle 0, u \rangle} = \bar{0} = 0$ 

(d) $\langle u,v + w \rangle = \overline{\langle v + w, u \rangle} = \overline{\langle v, u \rangle + \langle w, u \rangle} = \overline{\langle v, u \rangle} + \overline{\langle w, u \rangle} = \langle u, v \rangle + \langle u, w \rangle$

(e) $\langle u, \lambda v \rangle = \overline{\langle \lambda v, u \rangle} = \bar{\lambda}\overline{\langle v, u \rangle} = \bar{\lambda}\langle u, v \rangle $

## Norms

**Definition (Norm) $\Vert v \Vert$**

For $v \in V$, the norm of v, denoted by $\Vert v \Vert$, is defined by :

$$\Vert v \Vert = \sqrt{ \langle v, v \rangle }$$

**Properties of Norm**  

Suppose $v \in V$

(a) $\Vert v \Vert = 0 \text{ iff } v = 0$

(b) $\Vert\lambda v \Vert = |\lambda|\Vert v \Vert$ for all $\lambda \in F$


**orthogonal**  

Two vectors $u, v \in V$ are called orthogonal if $\langle u, v \rangle = 0$

Note: 0 is orthogonal to every vector in V. 0 is also the only vector orthogonal to itself.


# Pythagorean Theorem

Suppose $u, v \in V$. If u, v are orthogonal, then:

$\Vert u + v \Vert^2 = \Vert u \Vert^2 + \Vert v \Vert^2$

Proof:

$$
\begin{aligned}
\Vert u + v \Vert^2 &= \langle u + v, u + v \rangle \\
    &= \langle u, u + v \rangle +  \langle v, u + v \rangle \\
    &= \langle u, u \rangle + \langle u, v \rangle + \langle v, u \rangle + \langle v, v \rangle \\
    &= \Vert u \Vert^2 + \Vert v \Vert^2 \text{ because u,v are orthogonal }\\
\end{aligned}
$$

**Orthogonal Decomposition of u in terms of v**  

We want : u = cv + (u - cv). Here we want (u - cv) to be perpendicular to v.

So, $\langle u - cv, v \rangle = \langle u, v \rangle + \langle -cv, v \rangle = \langle u, v \rangle -c \Vert v \Vert^2$. This implies : $c = \dfrac{\langle u, v \rangle}{\Vert v \Vert^2}$. Then we have :

$$u^{||} = \dfrac{\langle u, v \rangle}{\Vert v \Vert^2}v$$

$$u^{\perp} = u - \dfrac{\langle u, v \rangle}{\Vert v \Vert^2}v$$



## Cauchy-Schwarz Inequality

Suppose $u, v \in V$

$$|\langle u, v \rangle| \le \Vert u \Vert \Vert v \Vert$$

This inequality is an equality if and only if one of u, v is a scalar multiple of the other.



Proof:

If v = 0, then $\langle u, v \rangle = 0$ and $\Vert v \Vert = 0$. So the inequality is true trivially.

We decompose u orthogonally :

$$u = \dfrac{\langle u, v \rangle}{\Vert v \Vert^2}v + w$$

Where w is the orthogonal component. By the Pythagorean theorem:

$$
\begin{aligned}
\Vert u \Vert^2 &= \left( \dfrac{\langle u, v \rangle}{\Vert v \Vert^2} \Vert v \Vert \right)^2 + \Vert w \Vert^2 \\
    &= \dfrac{\langle u, v \rangle^2}{\Vert v \Vert^2} + \Vert w \Vert^2 \\
    &\ge \dfrac{\langle u, v \rangle^2}{\Vert v \Vert^2} \\
\langle u, v \rangle^2  &= \Vert u \Vert^2 \Vert v \Vert^2 \\
\langle u, v \rangle  &= \Vert u \Vert \Vert v \Vert \\
\end{aligned}
$$

The inequality becomes an equality if the perpendicular component is zero i.e $u = cv$


**Example : Cauchy-Schwarz**  

a) If $x_1,\dots,x_n,y_1,\dots,y_n \in R$, then

$$(x_1y_1 + x_2y_2 + \dots + x_ny_n)^2 = (x_1^2 + x_2^2 + \dots + x_n^2)^2 + (y_1^2 + y_2^2 + \dots + y_n^2)^2$$


b) If f,g are continuous real-valued functions on $[-1,1]$ then :

$$\Bigg| \int_{-1}^{1} fg \Bigg|^2 = \left( \int_{-1}^{1} f^2 \right) \left( \int_{-1}^{1} g^2 \right)$$

## Triangle Inequality

Suppose $u, v \in V$

$$\Vert u + v \Vert \le \Vert u \Vert + \Vert v \Vert$$

This inequality is an equality if and only if one of u, v is a scalar multiple of the other.


Proof:

$$
\begin{aligned}
\Vert u + v \Vert^2 &= \langle u + v, u + v \rangle \\
                &= \Vert u \Vert^2 +  \Vert v \Vert^2 + \langle u, v \rangle + \langle v, u \rangle \\
                &\le \Vert u \Vert^2 +  \Vert v \Vert^2 + 2\Vert u \Vert\Vert v \Vert \text{ ... by Cauchy-Schwarz } \\   
                &= (\Vert u \Vert + \Vert v \Vert)^2 \\
\Vert u + v \Vert &\le \Vert u \Vert + \Vert v \Vert
\end{aligned}
$$

## Parallelogram Equality

Suppose $u,v \in V$

$$\Vert u + v \Vert + \Vert u - v \Vert = 2(\Vert u \Vert^2 + \Vert v \Vert^2)$$

Proof:

$$
\begin{aligned}
\Vert u + v \Vert + \Vert u - v \Vert &= \langle u + v, u + v \rangle + \langle u - v, u - v \rangle \\
                &= 2\langle u, u \rangle + 2\langle v, v \rangle \\
                &= 2(\Vert u \Vert^2 + \Vert v \Vert^2) 
\end{aligned}
$$
