# Fields $F - R$ or $C$

R and C are fields - let us refer to them as F. Elements of F are called **scalars**. 

Complex numbers are of the form $\{a + bi : a,b \in R \}$

For complex numbers, addition and multiplication are defined by:

$(a + bi) + (c + di) = (a + c) + (b + d)i$  

$(a + bi) - (c + di) = (ac - bd) + (ad + bc)i$  

The above formula can be derived by simply assuming $i^2 = -1$

Properties of Fields:

A set F endowed with addition (+) and multiplication (.) operations:

- **commutativity and associtivity** : both addition and multiplication are commutative and associative. Multiplication distributes over addition i.e. x.(y + z) = x.y + x.z.
- **identities** : 0 is the additive identity, 1 is the multiplicative identity
- **inverses** : every element has a additive inverse i.e. if x is in F, then there is an element -x such that x + (-x) = 0. Every non-zero element has a multiplicative inverse (i.e if x is in F, and non-zero, then there is an element $x^{-1}$, such that $x.x^{-1} = 1$
- **distributive property** : a(x + y) = ax + ay for all a,x,b in F

If x is in F, and m is a positive integer, we can define $x^m = x.x..x$, m times. This is defined using the field definitions, but is not an axiom. We can show that $(z^m)^n = z^{mn}$, $(wz)^m = w^mz^m$ for all w,z in F, and all positive integers m, n.

Note that Q forms a field too, but we are not covering it here.


**Note on Fields of Characteristic 0**

In general a field is any set with 0,1 and endowed with the proeprties above.

A field's characteristic is the positive number of times 1 + 1 + ..  can be added before we get zero. If no such number exists, then the field is said to be of characteristic 0.

Q,R,C are of characteristic 0. In general you could have vector spaces over any field, but some properties would differ between fields of characteristic zero and others.

Q is the smallest possible field of characteristic 0. It is ordered, countable, incomplete, not closed algebraically (i.e. a polynomials may not have a root in Q. Every other field of characteristic 0 has a copy of Q.

R is complete (has the lub property), is ordered, uncountable, not algebraically closed because negative numbers don't have square roots. Any complete ordered field would necessarily be isomorphic to R.

C is algebraically closed, but no ordering that respects multiplication. 

This book is focused on R and C only.

## (Proposition) Uniqueness of Inverses

(a) Suppose a has two inverses, b and c. Then a + b = 0. Adding c on both sides:

c + (a + b) = c  
i.e. (c + a) + b = c (associativity)  
0 + b = c (additive inverse)  
b = c  (identity)

(b) Suppose a is not zero and has two multiplicative inverses b and c.

a.b = 1  
multiplying by c: c(ab) = c  
i.e. (ca)b = c (associativity)  
1.b = c (inverse)  
b = c (identity)

## Definition : $-\alpha$ : subtraction, $\dfrac{1}{\alpha}$ : division, ($\alpha^m) : power$

Given the uniqueness of inverses, we can define two new operations:

**subtraction**

given a,b in F:

$$a - b = a + (-b)$$

**division**

Given a,b in F, $b \ne 0$:

$$a/b = a.b^{-1}$$ 

**power**

Given a in F and m,n are non-negative integers:

We define : $a^0 = 1$ for all a in F.

$a^m = a.a.\dots.a$ m-times, when m > 0

We can then define $a^{-m}$ as the multiplicative inverse of $a^m$

$a^{-m} = (a^m)^{-1} = 1/a^m$


It follows that:

$(a^m)^n = a^{mn}$

$(ab)^m = a^mb^m$

## $F^n$

**Coordinate**

For purposes of this book, we will refer to R or C as F.

Thenr we define $F^n$ as the set of all "lists" of length n of F.

$$F^n = \{(x_1,x_2, \dots, x_n): x_j \in F, \text{ for } j = 1,..,n  \}$$

In particular: $0 = (0,0,\dots,0)$, n times.

Note that $F^n$ are not fields, for n > 1.

**Addition**

We can define addition of two elements of $F^n$ by the addition of the corresponding coordinates.

$(x_1 + x_2 + \dots + x_n) + (y_1, y_2, \dots, y_n) = (x_1 + y_1, x_2 + y_2, \dots, x_n + y_n)$

**Addition is commutative** : because the underlying field F (R or C) is commutative.

**Additive inverse exists and is unique** : again because of underlying field's additive inverse. Thus for any element x we can denote the element -x, and also define subtraction of two elements of $F^n$.

**Scalar Multiplication** 

$a(x_1, x_2, \dots, x_n) = (ax_1, ax_2, \dots, ax^n)$

# Vector Spaces

## Definition (Vector Space over a Field F)

We define a **vector space** motivated by properties of $F^n$. We can easily prove the properties below for $R^n$ and $C^n$, using the field properties defined earlier.

<div class="alert alert-box alert-info">
    
A **Vector Space over a field F** is a set V, along with an addition and scalar multiplication on V which satisfy the following properties:

Addition (+)

- commutativity : u + v = v + u for all $u,v \text{ in } F^n$
- associativity: u + (v + w) = (u + v) + w for all $u,v,w \text{ in } F^n$
- additive identity : there is an element $0 \in V$, such that $v + 0 = v$, for all $v \text{ in } F^n$
- additive inverse: for any v in $F^n$, there is an element -v such that v + (-v) = 0

Scalar Multiplication:

- associativity: (ab)v = a(bv), for all v in $F^n$ and all a,b in $F$.
- multiplicative identity: 1v = v for all v in $F^n$

Distributive Properties:
- a(u + v) = au + av for all u,v in $F^n$ and all a in F
- (a + b)u = au + bu for all u in $F^n$ and all a,b in F

A vector space over R is called a **real vector space** and one over C is called a **complex vector space**. Elements of a vector space are called **vectors** or **points**.

</div>

Note what we don't have: any notion of vector multiplication. 


**Unique Additive Identity**

A vector space has a unique additive identity.

Proof: Suppose 0 and 0' are bot identities. Then 0' = 0' + 0 = 0 + 0' = 0 - this follows from the definition of 0 as identity, then commutativity and finally by the definition of 0' as identity.


**Unique Additive Inverse**

Proof: if a and b are two inverses of x, then x + a = 0, hence b + (x + a) = (0 + b) => 0 + a = 0 + b => a = b.

So we can now define subtraction: u - v = u + (-v)

## Examples of Vector Spaces

**$F^S$ denotes the set of all functions from the set S to the set F**. 
The expression shows that e.g. that if we count all functions from a set of size 3 to a set of size 5, we have 5^3 = 125 functions possible: basically sampling (w/ replacement) the 5 elements 3 times. 

An example is $R^{[0,1]}$ - the set of all functions from the interval $[0,1] \to R$.

We can see that by defining the sum $f + g$ as expected, and also scalar multiplication $af$, we get a vector space.

Note that the 0 element is f, such that f(x) = 0 for all x. With that definition we can define additive inverses and get all the other properties.

**$F^n$ is a special case of $F^S$**

We can think of $F^n$ as being $F^{\{1,2,\dots,n\}}$

**$F^\infty$ - the set of all sequences**

$F^\infty = \{(x_1,x_2, \dots): x_j \text{ in } F \text{ for } j \in 1,2,\dots \}$ - this is an infinite-dimensional vector space

We can think of $F^{\infty}$ also as being $F^{\{1,2,\dots\}}$

**Polynomials**

A function $p:\boldsymbol{F} \to \boldsymbol{F}$ is called a polynomial with coefficients in F if there exist $a_0, a_1, \dots, a_m \in F$ such that $p(z) = a_0 + a_1z + a_2z^2 + a_mz^m$, for all $z \in F$.

Then, we define $P(F)$ as the set of all polynomials with coefficients in F. We can define addition and scalar multiplication as expected, and we get P(F) as a vector space over the field F.

# Subspaces

<div class="alert alert-box alert-info">

Definition: A subset U of a vector space V is called a **subspace** of V if U is also a vector space (using the same definitions of addition and scalar multiplication).

To test a subset U is a subspace, we have the following 3 tests:

i) 0 is in U

ii) (Closed under addition) u,v in U $\implies$ u+v is in U 

iii) (Closed under scalar multiplication) $a \in F \text{ and } u \in U \implies au \in U$

</div>

The motivation to define subspaces comes from a need to be able to define linear subspaces of a space e.g. defining planes, lines etc. As will be seen, the fact that a vector space contains 0 implies these can only be the subspaces containing the origin.

But with this restriction, in some sense we can represent a larger-dimensional space as a collection of smaller-dimensional ones.

We prove later that for $R^2$, the subspaces are precisely : $\{ 0 \}$ and all lines through origin and $R^2$ itself. For $R^3$, we have in addition planes through the origin.

**Example of Subspaces**

b) Set of continuous real-valued functions on the interval $[0,1]$ is subspace of $R^{[0,1]}$

0 is continuous (by definition). Sum of two continuous functions is continuous. And multiplying a continuous function by a scalar is continuous. So this is a subspace.

c) Differentiable real-valued functions on $R^R$

Same as before. What is the derivative of a function which is 0 always? It's 0 everywhere. 

d) Real-valued differentiable functions f on the interval (0,3) such that $f'(2) = b$ is a subspace of $R^{[0,3]}$ iff b = 0  

If f'(2) = b and g'(2) = b => (f + g)'(2) = 2b. So (f + g)(2) is in the subset iff b = 0.

The limit of a (finite) sum is the sum of the limits. Multiplying a sequence by a scalar keeps limit at 0 (but does not work for other limits)




# Sums and Direct Sums

So how do we combine two subspaces to form a another?

Intersections work fine: The intersection of any two subspaces is a subspace: Obviously 0 is an element of the intersection. And if w is in the intersection, then aw must be in the intersection, similarly if u and v are in the intersection, so is u + v. 

But in general unions of subspaces are not subspaces - e.g. if we take 2 lines through origin, their union will not form a subspace. For this we introduce the concept of a sum of subspaces, and in particular the concept of a direct sum. These sums work analogously to unions of sets: Given two subspaces, the smallest subspace containing both is their sum.


<div class="alert alert-info">

**Definition (Sum of Subspaces)**

Suppose $U_1, U_2, ..., U_m$ are subspaces of V. The <b>sum of subspaces</b>  is a subspace, and is denoted by:
$$U_1 + U_2 + \text{ ... } + U_m = {u_1 + u_2 + ... + u_m : u_i \in U_i, i = 1..m}$$

In other words it is the collection of all possible vectors formed by taking one vector from each subset, and adding them together.

Some obvious properties:

- (Commutativity) Since vector addition is commutative, obviously so is the sum. 
- (Associativity) Associativity also follows equally well.
- (Superset of each subspace) Each of $U_1, U_2, ..., U_m$ are subsets of the sum, because we can always pick 0 for m-1 of the elements. 
- (Subspace) The sum is a subspace. This follows because : (Additive Identity) Note that the subspace {0} is an additive identity for the sum of subspaces, since 0 is a member of every subspace. But U is a subspace then U = U + U, since every subspace is closed under addition, U + V = U does not imply V = 0. (Closed under addition) : The sum is closed under addition. If u and v are in the sum, then we can easily see so is u + v, by the fact that each $U_i$ is also closed under addition. (Closed under scalar multiplication) - For the same reason.
- (Smallest Subspace) : Any subspace of V containing $U_1, U_2, ..., U_m$ must also contain their sum, because a subspace contains the (finite) sum of all its elements. So this is the smallest subspace containing all elements of $U_1, U_2, ..., U_m$.
- (No Additive Inverse) For a subspace V, is there as (inverse) subspace W such that V + W = {0}? Since a subspace is contained in the sum, this is only possible if V and W are both {0} i.e. in general there is no inverse.
</div>


If we sum two lines through the origin we get a plane. If we sum a plane and a line through the origin, as can be seen, you will get the whole $R^3$ space (assuming the line is not in the plane). But 2 planes through the origin will also give us the $R^3$ space which is excessive. This fact leads us to the definition of a direct sum :


<div class="alert alert-info">

**Definition (Direct Sum of Subspaces)**

A sum of subspaces is a direct sum if every vector in the direct sum is formed by a unique sum of vectors drawn from each of the subspaces forming the sum. We denote this by $U_1 \oplus U_2 \oplus \text{ ... } \oplus U_m$
</div>


Two planes through the origin do not form a direct sum, but a line and plane not containing the line do. In a sense, we are overspecifying when we use two planes, allowing multiple ways to form any other vector.

<div class="alert alert-box alert-info" >
    
## Proposition (When is a sum a direct sum)?

Suppose $U_1, U_2, ..., U_m$ as subspaces of V. Then V = $U_1 \oplus U_2 \oplus \text{ ... } \oplus U_m$ iff the following conditions hold:

a) $V = U_1 + U_2 + \dots + U_n$

b) the only way to write $0$ as a sum of $u_1 + u_2 + \dots + u_n$, where each $u_j \in U_j$, is by taking $u_j = 0$ for j = 1..n.

</div>

The second condition is simply saying that given any two subspaces for example, i cannot take one non-zero vector from each such subspace such that they cancel out each other other and give the zero vector. 

Think of the 2 planes intersecting at the origin. Their line of intersection goes through the origin. And i can take opposing vectors from this line and get the vector 0. But if i take a plane, and a line perpendicular to the plane, I can only form the zero vector by taking 0 from each subspace. Any non-zero vector will have components either in the plane or the line which cannot cancel.

Proof: 
<= (Suppose V = $U_1 \oplus U_2 \oplus \text{ ... } \oplus U_m$). Both (a) and (b) follow easily from the definition of direct sum. For (b) note that we can write 0 = 0 + 0 + .. + 0, since 0 is an element of every subspace. Then, by definition of direct sum, this is unique.

=> (Suppose conditions (a) and (b) hold)): We have $V = U_1 + U_2 + \dots + U_n$. To prove this is a direct sum, we have to show any v in V can be expressed as a unique combination of vectors from each of the subspaces. 

First note that $0 = 0 + 0 + \dots + 0$ n times, where we take each 0 from one for the $U_i$. So we know this is one way to write 0 as a sum. No other way must exist, by condition (b). 

Now let v be any vector in V such that :

$v = u_1 + u_2 + \dots + u_n$ and $v = w_1 + w_2 + \dots + w_n$, where the u's and w's are from the respective subspaces.

Subtracting, we get $0 = (u_1 - w_1) + (u_2 - w_2) + (u_n - w_n)$. 

Now each $(u_i - w_i) \in U_i$, because $U_i$ is a subspace and closed under addition and scalar multiplication. And by condition (b) - the uniqueness of the representation of 0 means thet each $u_i - w_i = 0$ i.e. $u_i = w_i$ for all i.




<div class="alert alert-box alert-info">

## Proposition (Sum of Two Subspaces is a Direct Sum if their intersection is just the 0 vector)

Suppose U and W are subspaces of V. Then $V = U \oplus W$ iff $V = U + W$ and $U \cap W = \{ 0 \}$ 

</div>


Direct Sums of subspaces are like "disjoint" unions of sets, in that there is some way they do not overlap, with the notable exception of the 0 element.

But note that this statement is true for only 2 subspaces. A collection of 3 spaces where the condition is satisfied pairwise need not be a direct sume.

The restriction to two sets can be surprising. If three subspaces exist which only intersect, pairwise, at 0, then they may still not form a direct sum. Informally this is because it may be possible to form vectors of one subspace by adding vectors of the other two - this breaks the condition for a direct sum. Consider :

$U_1 = (x,y,0), U_2 = (0,0,z) \text{ and } U_3 = (0,y,y)$ 

- $U_1$ is the xy plane
- $U_2$ is the z-axis
- $U_3$ is a slanted line in the yz plane.

Seen pairwise, each of their intersections is just the zero vector. Note that combining z-axis and the slanted yz-line allows us to create the whole y-z plane uniquely (these are just non-perpendicular coordinate axis). Similarly the xy plane and the yz-line allows us to create all of $R^3$ uniquely. But taken all together, we get a contradiction e.g.

$(0,0,0) = (0,1,0) + (0,0,1) + (0,-1,-1)$, and in fact this means there are infinite number of ways to represent the 0 vector.



Proof:

<= (Suppose $V = U \oplus W$) : Then clearly V = U + W (by one part of the definition of direct sum). 

Suppose $v \in U \cap W$. We have to show v = 0. Since v is in U and W, so is -v. Then we can form $0 = v + (-v)$. But $0 = 0 + 0$ already, so this is only possible when v = 0.

=> (Suppose $V = U + W$ and $U \cap W = \{ 0 \}$): Because of the previous proposition, we have to only prove the representation of 0 as a sum is unique and made of all zeros. 

Suppose $0 = u + w$, where u in U and w in W. Then u = -w i.e. u is in W also, because W contains all scalar multiples of w. Then $u \in U \cap W$, which implies u = w = 0, by the hypothesis.

But, 

$\sum\limits_{i=1}^{n} p_iz^i + \sum\limits_{i=1}^{n} q_iz^i = \sum\limits_{i=1}^{n} q_iz^i + \sum\limits_{i=1}^{n} p_iz^i = q(z) + p(z) = (q + p)(z) => p + q = q + p$, by commutativity of addition in F. 


- *associativity*: u + (v + w) = (u + v) + w for all $u,v,w \text{ in } F^n$

We have to show: $p + (q + r) = (p + q) + r$, which easily follows by expanding out the sum as above.

Then calculate $p + 0 = p$

**Direct Sum : $P(F) = U_e \oplus U_o$ - even and odd polynomial subspaces**

We define $U_e$ as the set of all polynomials p of the form:

$p(z) = a_0 + a_2z^2 + \dots a_{2m}z^{2m}$, and 

$U_o$ as the set of all polynomials p of the form:

$p(z) = a_1z + \dots a_{2m+1}z^{2m+1}$

First: $U_e$ is a subspace: 0 is in $U_e$, adding two even polynomials gives an even polynomial, and multiplying by a scalar still keeps the polynomial even. Similarly $U_o$ is a subspace.

Is P(F) a direct sum of $U_e$ and $U_o$: Any p(z) in P(F) can be split into odd and even terms. It is a well known theorem for polynomials that if two polynomials of same degree are equal for all inputs, the coefficients must match. This means the representation of p(z) as a sum of even + odd polynomial is unique, meeting the criteria for a direct sum.
