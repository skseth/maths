# Scalar & Cross Product

**Definitions**

$$\mathbf{A} \cdot \mathbf{B} = A_xB_x + A_yB_y + A_zB_z$$

$$\mathbf{A} \cdot \mathbf{B} = \Vert \mathbf{A} \Vert \Vert \mathbf{B} \Vert \cos(\theta)$$

$$\mathbf{A} \times \mathbf{B} = (A_yB_z - A_zB_y)\mathbf{\hat{i}} + (A_zB_x - A_xB_z)\mathbf{\hat{j}} + (A_xB_y - A_yB_x)\mathbf{\hat{k}}$$

$$\mathbf{A} \times \mathbf{B} = \begin{vmatrix} \hat{i} & \hat{j} & \hat{k} \\ A_x & A_y & A_z \\ B_x & B_y & B_z \end{vmatrix}$$

Note that transposing gives the same result.

**Comparison**

Scalar Product - order does not matter (symmetric). Converts two vectors to one scalar value. It involves the **projection** one one vector on another.

Cross Product - order does matter and in fact reversing the order reverses the sign (anti-symmetric). It involves getting the **oriented area** between two vectors, and is perpendicular to both vectors.

$$\mathbf{A} \times \mathbf{B} = -\mathbf{B} \times \mathbf{A}$$

Also:

$$\mathbf{A} \times \mathbf{B} = \Vert \mathbf{A} \Vert \Vert \mathbf{B} \Vert \sin(\theta)$$


## Cross Product - Unit Vectors

| ijk | ijk-jki | jki-ijk |
| ------------ | --------- | -------- |
| $\mathbf{\hat{i}} \times \mathbf{\hat{i}} = 0$ | $\mathbf{\hat{i}} \times \mathbf{\hat{j}} = \mathbf{\hat{k}}$ | $\mathbf{\hat{j}} \times \mathbf{\hat{i}} = -\mathbf{\hat{k}}$ |
| $\mathbf{\hat{j}} \times \mathbf{\hat{j}} = 0$ | $\mathbf{\hat{j}} \times \mathbf{\hat{k}} = \mathbf{\hat{i}}$ | $\mathbf{\hat{k}} \times \mathbf{\hat{j}} = -\mathbf{\hat{i}}$ |
| $\mathbf{\hat{k}} \times \mathbf{\hat{k}} = 0$ | $\mathbf{\hat{k}} \times \mathbf{\hat{i}} = \mathbf{\hat{j}}$ | $\mathbf{\hat{k}} \times \mathbf{\hat{i}} = -\mathbf{\hat{j}}$ |

Applications include : torque: $\mathbf{\tau} = \mathbf{r} \times \mathbf{F}$, magnetic force problems : $\mathbf{F} = q\mathbf{v} \times \mathbf{B}$

## Scalar Triple Product

$$\mathbf{A} \cdot (\mathbf{B} \times \mathbf{C})$$

$$\mathbf{A} \cdot (\mathbf{B} \times \mathbf{C}) = \begin{vmatrix} A_x & A_y & A_z \\ B_x & B_y & B_z \\ C_x & C_y & C_z \end{vmatrix}$$

A simple way to remember is that $A_x = \mathbf{\hat{i}}.\mathbf{A}$. 

The scalar triple product is the volume of the rectangular parallelepiped which is easy to see if you see that if $\phi$ is the angle between $\mathbf{A}$ and $\mathbf{B} \times \mathbf{C}$, then the value is $(\Vert \mathbf{A} \Vert \cos(\phi)) \Vert \mathbf{B} \times \mathbf{C} \Vert$

## Vector Triple Product



$$\mathbf{A} \times(\mathbf{B} \times \mathbf{C}) = \mathbf{B}(\mathbf{A} \cdot \mathbf{C}) - \mathbf{C}(\mathbf{A} \cdot \mathbf{B})$$

The way to remember is that the result is perpendicular to both $\mathbf{A}$ and $\mathbf{B} \times \mathbf{C}$, so will line in the plane of $\mathbf{B} \times \mathbf{C}$ - hence it's a line 



Proof:

$\mathbf{A} \times(\mathbf{B} \times \mathbf{C}) = a\mathbf{B} + b\mathbf{C}$

We set up $\mathbf{B} = B_x\hat{i}$ and $\mathbf{C} = C_x\hat{i} + C_y\hat{j}$

Then:

$\mathbf{B} \times \mathbf{C} = B_xC_y\hat{k}$

Now the vector product with A gives: $A_yB_xC_y\hat{\mathbf{i}} - A_xB_xC_y\hat{\mathbf{j}} = A_yB_xC_y\hat{\mathbf{i}} + A_xB_xC_x\hat{\mathbf{i}} - A_xB_xC_x\hat{\mathbf{i}} - A_xB_xC_y\hat{\mathbf{j}} = B_x\hat{\mathbf{i}}(A_xC_x + A_yC_y) - (C_x\hat{\mathbf{i}} + C_y\hat{\mathbf{j}})(A_xB_x)$




## Digression: Matrices

A completely different way to look at the cross product $\mathbf{b} \times \mathbf{c}$ involves using $\mathbf{b}$ as an axis, and rotating $\mathbf{c}$. 

For a vector A, you can define its matrix for rotation of another vector around A (in cross product terms) as a "skew-symmetric" matrix:

$\mathbf{A} = [\mathbf{a}]_{\times} = \begin{bmatrix} 0 &  -a_z & a_y \\ a_z & 0 & -a_x \\ -a_y & a_x & 0  \end{bmatrix}$, $\mathbf{B} = [\mathbf{b}]_{\times} = \begin{bmatrix} 0 &  -b_z & b_y \\ b_z & 0 & -b_x \\ -b_y & b_x & 0  \end{bmatrix}$


Then, $\mathbf{b} \times \mathbf{c} \equiv [\mathbf{b}]_{\times}\mathbf{C} = \mathbf{B}\mathbf{C}$. Basically the cross product is an operator, which transforms C. What is more, we can take this rotating system and spin around another axis:

$\mathbf{a} \times (\mathbf{b} \times \mathbf{c}) \equiv [\mathbf{A}]_{\times}[\mathbf{B}]_{\times}\mathbf{C}$

Now :




$$
\begin{aligned}
\mathbf{A}_{\times}\mathbf{B}_{\times} &= \begin{bmatrix} -a_zb_z -a_yb_y & a_yb_x & a_xb_x \\ a_xb_y & -a_zb_z - a_xb_x & a_zb_y  \\ a_xb_x & a_yb_z & -a_yb_y - a_xb_x\end{bmatrix} \\
 &= \begin{bmatrix} a_xb_x & a_yb_x & a_xb_x \\ a_xb_y & a_yb_y & a_zb_y  \\ a_xb_x & a_yb_z & a_zb_z\end{bmatrix} - \mathbf{a} \cdot  \mathbf{b} \mathbf{I} \\
 &= \mathbf{B}\mathbf{A}^T - \mathbf{a} \cdot  \mathbf{b} \mathbf{I} \\
\mathbf{A}_{\times}\mathbf{B}_{\times}\mathbf{C} &= \mathbf{B}\mathbf{A}^T\mathbf{C} - (\mathbf{a} \cdot  \mathbf{b}) \mathbf{I}\mathbf{C} \\
&= \mathbf{B}(\mathbf{a} \cdot \mathbf{c}) - \mathbf{C}(\mathbf{a} \cdot  \mathbf{b})
\end{aligned}
$$

## Partial Derivatives

Assume f is a function of x and y.

Then, at some point P, we can express the rate of change of z (or the function f itself), as :

$$dz = \frac{\partial z}{\partial x}dx + \frac{\partial z}{\partial y}dy$$

IN effect, the change is z is some linear combination of the change in x and y, which depends on the partial derivatives at the point.

You can also take repeated partial derivatives: So $\dfrac{\partial^2 z}{\partial x^2}$ tells the change in the x-direction slope of z, keeping y fixed. 

Similarly, you may have $\dfrac{\partial^2 z}{\partial x\partial y}$ is the change in the y-direction slope as you move in the x direction, keeping y fixed, and $\dfrac{\partial^2 z}{\partial y\partial x}$ represents the change in the x-direction slope as you move in the y-direction, keeping x fixed. Usually, for well behaved functions, this result is the same.

**Chain Rule for Partial Derivatives**

Suppose you want to know change in z w.r.t u and v, where x = g(u,v) abd y = h(u,v). Then:

$\dfrac{\partial z}{\partial u} = \dfrac{\partial z}{\partial x}\dfrac{\partial x}{\partial u} + \dfrac{\partial z}{\partial y}\dfrac{\partial y}{\partial u}$

$\dfrac{\partial z}{\partial v} = \dfrac{\partial z}{\partial x}\dfrac{\partial x}{\partial v} + \dfrac{\partial z}{\partial y}\dfrac{\partial y}{\partial v}$


**Example**

Consider example:

$f(x,y) = 6x^2y + 3x + 5xy + 10$

$\frac{\partial z}{\partial x} = 12xy + 3 + 5y$

$\frac{\partial z}{\partial y} = 6x^2 + 5x$

$\frac{\partial z}{\partial y\partial x} = 12x + 5$

$\frac{\partial z}{\partial x\partial y} = 12x + 5$


```python
import numpy as np
import matplotlib.pyplot as plt

# Enable interactive plots if using Jupyter
%matplotlib inline 

# Define range for x and y from -3 to 3
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

# Create a 2D grid of x and y values
X, Y = np.meshgrid(x, y)

# Compute Z based on the equation
Z = 6 * (X**2) * Y + 3 * X + 5 * X * Y + 10

# Plotting the 3D surface
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')

# Create the surface plot
surface = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Add labels and a color bar
ax.set_title('3D Surface Plot of f(x, y)')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
fig.colorbar(surface, shrink=0.5, aspect=5)

plt.show()
```

## Vectors as Derivatives





Suppose we have f(x,y), where both x and y are functions of some parameter t. Then the directional derivative of f along the curve parameterized by t is :

$\dfrac{df}{dt} = \dfrac{dx}{dt}\dfrac{\partial f}{\partial x} + \dfrac{dy}{dt}\dfrac{\partial f}{\partial y}$

We can then treat the expression as an "operator" (a function which acts on a function, briefly), and plug in any function f.

$\dfrac{d}{dt} = \dfrac{dx}{dt}\dfrac{\partial }{\partial x} + \dfrac{dy}{dt}\dfrac{\partial }{\partial y}$

Here the term $\dfrac{d}{dt}$ represents the tangent vector to the curve, the $\dfrac{dx}{dt}$ and $\dfrac{dy}{dt}$ represent the x and y components of that vector, and the vectors $\dfrac{\partial }{\partial x}$ and $\dfrac{\partial }{\partial x}$ represent the basis vectors in the x and y direction, effectively scaled by the function in question.




## Nabla (Del) Operator

The Del Operator is :

$$\nabla \equiv \hat{i}\dfrac{\partial }{\partial x} + \hat{j}\dfrac{\partial }{\partial y} + \hat{k}\dfrac{\partial }{\partial z}$$

## Gradient of a Scalar Field

**Cartesian Coordinates**

$$grad(\Psi) = \nabla \Psi \equiv \hat{i}\dfrac{\partial \Psi}{\partial x} + \hat{j}\dfrac{\partial \Psi}{\partial y} + \hat{k}\dfrac{\partial \Psi}{\partial z}$$

**Cylindrical Coordinates $(r, \phi, z)$**

Using chain rule, we get:

$$grad(\Psi) = \nabla \Psi \equiv \hat{r}\dfrac{\partial \Psi}{\partial r} + \hat{\phi}\dfrac{1}{r}\dfrac{\partial \Psi}{\partial \phi } + \hat{z}\dfrac{\partial \Psi}{\partial z} \quad \text{ ... cylindrical }$$


Proof: 

$\dfrac{\partial \Psi}{\partial r} = \dfrac{\partial \Psi}{\partial x}\dfrac{\partial x}{\partial r} +  \dfrac{\partial \Psi}{\partial y}\dfrac{\partial y}{\partial r} = \dfrac{\partial \Psi}{\partial x}\cos(\phi) +  \dfrac{\partial \Psi}{\partial y}\sin{\phi}$

$\dfrac{\partial \Psi}{\partial \phi} = \dfrac{\partial \Psi}{\partial x}\dfrac{\partial x}{\partial \phi} +  \dfrac{\partial \Psi}{\partial y}\dfrac{\partial y}{\partial \phi} = -\dfrac{\partial \Psi}{\partial x}r\sin(\phi) +  \dfrac{\partial \Psi}{\partial y}r\cos{\phi}$

z stays same. Simplifying, we get:

$\dfrac{\partial \Psi}{\partial x} = \dfrac{\partial \Psi}{\partial r}\cos(\phi) - \dfrac{\partial \Psi}{\partial \phi}\dfrac{\sin(\phi)}{r}$

$\dfrac{\partial \Psi}{\partial y} = \dfrac{\partial \Psi}{\partial r}\sin(\phi) + \dfrac{\partial \Psi}{\partial \phi}\dfrac{\cos(\phi)}{r}$

$$
\begin{aligned}
\nabla \Psi &\equiv \hat{i}\dfrac{\partial \Psi}{\partial x} + \hat{j}\dfrac{\partial \Psi}{\partial y} + \hat{k}\dfrac{\partial \Psi}{\partial z} \\
&= \dfrac{\partial \Psi}{\partial r}(\cos(\phi)\hat{i} + \sin(\phi)\hat{j}) + \dfrac{\partial \Psi}{\partial \phi }\dfrac{1}{r}(-\sin(\phi)\hat{i} + \cos(\phi)\hat{j}) + \hat{k}\dfrac{\partial \Psi}{\partial z} \\
&= \hat{r}\dfrac{\partial \Psi}{\partial r} + \hat{\phi}\dfrac{\partial \Psi}{\partial \phi }\dfrac{1}{r} + \hat{z}\dfrac{\partial \Psi}{\partial z}
\end{aligned}
$$

**Spherical Coordinates $(r, \theta, \phi)$**


$$grad(\Psi) = \nabla \Psi \equiv \hat{r}\dfrac{\partial \Psi}{\partial r} + \hat{\theta}\dfrac{1}{r}\dfrac{\partial \Psi}{\partial \theta} + \hat{\phi}\dfrac{1}{r\sin(\theta)}\dfrac{\partial \Psi}{\partial \phi} \quad \text{ ... spherical }$$


```python
import sympy as sp

# 1. Initialize pretty printing for Jupyter Notebook output
sp.init_printing(use_latex='mathjax')

# 2. Define our symbolic variables
r, theta, phi = sp.symbols('r theta phi', positive=True)

# 3. Construct the Jacobian matrix M that maps Cartesian derivatives to Spherical derivatives
# [df/dr, df/dtheta, df/dphi]^T = M * [df/dx, df/dy, df/dz]^T
M = sp.Matrix([
    [sp.sin(theta)*sp.cos(phi),   sp.sin(theta)*sp.sin(phi),   sp.cos(theta)],
    [r*sp.cos(theta)*sp.cos(phi), r*sp.cos(theta)*sp.sin(phi), -r*sp.sin(theta)],
    [-r*sp.sin(theta)*sp.sin(phi), r*sp.sin(theta)*sp.cos(phi),  0]
])

# 4. Display the original forward transformation matrix
print("Forward Transformation Matrix (M):")
display(M)

# 5. Invert the matrix and simplify the trigonometric expressions
M_inverse = M.inv()
M_inverse_simplified = sp.simplify(M_inverse)

# 6. Display the inverted matrix
print("\nInverted Matrix (M^-1) mapping Spherical to Cartesian derivatives:")
display(M_inverse_simplified)


```

Based on the conversion matrix above, we get:

### Inverted Cartesian Derivative Equations

$$
\frac{\partial f}{\partial x} = \sin\theta\cos\phi\frac{\partial f}{\partial r} + \frac{\cos\theta\cos\phi}{r}\frac{\partial f}{\partial \theta} - \frac{\sin\phi}{r\sin\theta}\frac{\partial f}{\partial \phi}
$$

$$
\frac{\partial f}{\partial y} = \sin\theta\sin\phi\frac{\partial f}{\partial r} + \frac{\cos\theta\sin\phi}{r}\frac{\partial f}{\partial \theta} + \frac{\cos\phi}{r\sin\theta}\frac{\partial f}{\partial \phi}
$$

$$
\frac{\partial f}{\partial z} = \cos\theta\frac{\partial f}{\partial r} - \frac{\sin\theta}{r}\frac{\partial f}{\partial \theta}
$$

Substituting, we get:

### Collecting and Regrouping Terms

$$
\mathbf{\nabla} f = \frac{\partial f}{\partial r} \underbrace{(\sin\theta\cos\phi\,\hat{i} + \sin\theta\sin\phi\,\hat{j} + \cos\theta\,\hat{k})}_{\hat{r}} + \frac{1}{r}\frac{\partial f}{\partial \theta} \underbrace{(\cos\theta\cos\phi\,\hat{i} + \cos\theta\sin\phi\,\hat{j} - \sin\theta\,\hat{k})}_{\hat{\theta}} + \frac{1}{r\sin\theta}\frac{\partial f}{\partial \phi} \underbrace{(-\sin\phi\,\hat{i} + \cos\phi\,\hat{j})}_{\hat{\phi}}
$$



**Example**

$\Psi(x,y) = 5x + 2y$

This is a simple plane through origin. We can expect a constant gradient.

Indeed : $\nabla \Psi = 5\hat{\mathbf{i}} + 2\hat{\mathbf{j}}$

Which shows the gradient is constant everywhere (does not depend on x and y). 



```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider

def plot_dashboard(elevation=30, azimuth=-45, num_contours=15):
    # 1. Setup coordinate grid
    x = np.linspace(-5, 5, 100)
    y = np.linspace(-5, 5, 100)
    X, Y = np.meshgrid(x, y)
    
    # 2. Define the scalar field Psi(x, y) = 5x + 2y
    Psi = 5 * X + 2 * Y
    
    # 3. Setup a sparser grid for the vector fields
    x_q = np.linspace(-4, 4, 5)
    y_q = np.linspace(-4, 4, 5)
    X_q, Y_q = np.meshgrid(x_q, y_q)
    U = np.ones_like(X_q) * 5
    V = np.ones_like(Y_q) * 2
    
    # 4. Create a side-by-side figure layout
    fig = plt.figure(figsize=(18, 8))
    
    # ----------------------------------------------------
    # LEFT PLOT: 3D Surface Field
    # ----------------------------------------------------
    ax1 = fig.add_subplot(121, projection='3d')
    surf = ax1.plot_surface(X, Y, Psi, cmap='viridis', alpha=0.6, edgecolor='none')
    
    # Plot prominent 3D vectors sitting on the surface
    Z_q = 5 * X_q + 2 * Y_q
    ax1.quiver3D(
        X_q, Y_q, Z_q, U, V, np.zeros_like(X_q), 
        length=0.5, normalize=True, color='red', linewidth=2.5, arrow_length_ratio=0.35
    )
    
    ax1.set_title(r'3D Surface: $\Psi(x,y) = 5x + 2y$', fontsize=14)
    ax1.set_xlabel('X Axis')
    ax1.set_ylabel('Y Axis')
    ax1.set_zlabel(r'$\Psi(x,y)$')
    ax1.view_init(elev=elevation, azim=azimuth)
    fig.colorbar(surf, ax=ax1, shrink=0.5, aspect=10, pad=0.1, label='Field Intensity')
    
    # ----------------------------------------------------
    # RIGHT PLOT: 2D Gradient Field
    # ----------------------------------------------------
    ax2 = fig.add_subplot(122)
    contours = ax2.contour(X, Y, Psi, levels=num_contours, cmap='viridis', linewidths=1.5)
    ax2.clabel(contours, inline=True, fontsize=9, fmt='%1.1f')
    
    # Plot prominent 2D vectors
    ax2.quiver(
        X_q, Y_q, U, V,
        color='red', angles='xy', scale_units='xy', scale=12,
        width=0.007, headwidth=5, headlength=6, headaxislength=4.5,
        label=r'Gradient $\nabla\Psi = \langle 5, 2 \rangle$'
    )
    
    ax2.set_title(r'2D Projection: Gradient & Contours', fontsize=14)
    ax2.set_xlabel('X Axis', fontsize=12)
    ax2.set_ylabel('Y Axis', fontsize=12)
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.set_aspect('equal') # Keeps perpendicular angles visually accurate
    ax2.legend(loc='upper left')
    
    plt.tight_layout()
    plt.show()

# Interactive controls for the unified dashboard
interact(
    plot_dashboard,
    elevation=FloatSlider(min=-90, max=90, step=5, value=30, description='3D Elev'),
    azimuth=FloatSlider(min=-180, max=180, step=5, value=-45, description='3D Azim'),
    num_contours=IntSlider(min=5, max=30, step=1, value=15, description='2D Contours')
);
```
