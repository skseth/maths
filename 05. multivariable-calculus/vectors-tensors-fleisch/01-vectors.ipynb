## Vectors As Magnitude and Direction



A vector is shown as an arrow.

A set of vectors at each point represent a vector field. These vectors are "bound" in some sense, to the point they represent.

Are there "free" vectors - it depends on the application. There may also be vectors which can be "sliding".

A scalar is characterized by magnitude only.

Tensors are entities characterized by a magnitude and multiple directions.

Scalars represented by one number. Vectors (in three dimensions) by 3 numbers. Tensors by $3^R$ numbers in 3 dimensions.

| Scalar | Vector | Tensor |
| --- | --- | --- |
| $(x)$ | $(x_1, x_2, x_3)$ or $\begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}$ | $\begin{pmatrix} x_{11} & x_{12} & x_{13} \\ x_{21} & x_{22} & x_{23} \\ x_{31} & x_{32} & x_{33} \end{pmatrix}$ |

## Components and Cartesian Unit Vectors

Any vector can be expressed as a combination of unit vectors and components.


$\mathbf{A} = A_x\mathbf{\hat{i}} + A_y \mathbf{\hat{j}} + A_z \mathbf{\hat{k}}$

The components represent **projections" of the vector on each of the basis vectors.

You get the length, angles etc in the usual way. Addition, scalar multiplication works as expected.


## Polar Unit Vectors

Polar co-ordinates:

$$r = \sqrt{x^2 + y^2}$$

$$\theta = arctan(\frac{y}{x})$$

Conversion back:

$$x = r \cos(\theta)$$

$$y = r \sin(\theta)$$

Unit Vectors:

$$\hat{r} = \cos(\theta)\hat{i} + \sin(\theta)\hat{j}$$

$$\hat{\theta} = -\sin(\theta)\hat{i} + \cos(\theta)\hat{j}$$

Note that these "unit vectors" change direction, when moved. So they are not free in the sense discussed earlier.

The r unit vector is easy to see. But that about $\hat{\theta}$. It's basically achieved by rotating $\hat{r}$ by $\pi/2$. So we get $\hat{\theta} = \cos(\pi/2 + \theta)\hat{i} + \sin(\pi/2 + \theta)\hat{j} = -\sin(\theta)\hat{i} + \cos(\theta)\hat{j}$


## Cylindrical Unit Vectors - $R^3$

Cylindrical coordinates:

$$r = \sqrt{x^2 + y^2}$$

$$\phi = arctan(\frac{y}{x})$$

$$z = z$$

The other way:

$$x = r \cos(\phi)$$

$$y = r \sin(\phi)$$

$$z = z$$

Unit Vectors:

$$\hat{r} = \cos(\phi)\hat{i} + \sin(\phi)\hat{j}$$

$$\hat{\phi} = -\sin(\phi)\hat{i} + \cos(\phi)\hat{j}$$

$$\hat{z} = \hat{z}$$



## Spherical Unit Vectors - $R^3$

Spherical coordinates:

$$r = \sqrt{x^2 + y^2 + z^2}$$

$$\theta = \arccos(\frac{z}{\sqrt{x^2 + y^2 + z^2}}) \quad \text{polar angle}$$

$$\phi = \arctan(\frac{y}{x}) \quad \text{azimuthal angle}$$

The other way (note that $r\sin(\theta)$ is the projection of r on the x-y plane.

$$x = r\sin(\theta) \cos(\phi)$$

$$y = r\sin(\theta) \sin(\phi)$$

$$z = r\cos(\theta)$$

Unit Vectors:

$$\hat{r} = \sin(\theta)\cos(\phi)\hat{i} + \sin(\theta)\sin(\phi)\hat{j} + \cos(\theta)\hat{k}$$

$$\hat{\theta} = \cos(\theta)\cos(\phi)\hat{i} + \cos(\theta)\sin(\phi)\hat{j} - \sin(\theta)\hat{k}$$

$$\hat{\phi} = -\sin(\phi)\hat{i} + \cos(\phi)\hat{j}$$


**Notes**:

In just the x-y plane, we can see that $\hat{\phi} = -\sin(\phi)\hat{i} + \cos(\phi)\hat{j}$. 

Note that $\mathbf{r_{xy}} = \cos(\phi)\hat{i} + \sin(\phi)\hat{j}$

$\mathbf{r}$, projected on $\mathbf{r_{xy}}$ at an angle of $\pi/2 - \theta$, and so the component in the x-y plane of $\hat{r}$ is : $\sin(\theta)\cos(\phi)\hat{i} + \sin(\theta)\sin(\phi)\hat{j}$

The z-axis component is : $\sin(\pi/2 - \theta)\hat{k} = \cos(\theta)\hat{k}$

$\hat{\theta}$ is perpendicular to $\mathbf{r}$, in the plane containing $\mathbf{r}$ and $\mathbf{r_{xy}}$. So if we turn $\hat{r}$ by $\pi/2$, we get $\hat{\theta}$.

## Basis Vectors - Conversion



$$
\begin{aligned}
\mathbf{e_r} &= \dfrac{\partial x}{\partial r}\mathbf{\hat{i}} + \dfrac{\partial y}{\partial r}\mathbf{\hat{j}} + \dfrac{\partial z}{\partial r}\mathbf{\hat{k}} \\
        &= \sin(\theta)\cos(\phi)\mathbf{\hat{i}} + \sin(\theta)\sin(\phi)\mathbf{\hat{j}} + \cos(\theta)\mathbf{\hat{k}} \\
\mathbf{e_{\theta}} &= \dfrac{\partial x}{\partial \theta}\mathbf{\hat{i}} + \dfrac{\partial y}{\partial \theta}\mathbf{\hat{j}} + \dfrac{\partial z}{\partial \theta}\mathbf{\hat{k}} \\
        &= r\cos(\theta)\cos(\phi)\mathbf{\hat{i}} + r\cos(\theta)\sin(\phi)\mathbf{\hat{j}} - \sin(\theta)\mathbf{\hat{k}} \quad \text{ ... not a unit vector } \\
\mathbf{e_{\phi}} &= \dfrac{\partial x}{\partial \phi}\mathbf{\hat{i}} + \dfrac{\partial y}{\partial \phi}\mathbf{\hat{j}} + \dfrac{\partial z}{\partial \phi}\mathbf{\hat{k}} \\
        &= -r\sin(\theta)\sin(\phi)\mathbf{\hat{i}} + r\sin(\theta)\cos(\phi)\mathbf{\hat{j}} \quad \text{ ... not a unit vector } \\
\text{} \\
\text{ Note } \\
|\mathbf{e_{\theta}}| = r, |\mathbf{e_{\phi}}| = rsin(\theta) \\
\text{ This follows, because } r_{xy} \text{ has length }  rsin(\theta)
\end{aligned}
$$
