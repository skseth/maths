# Transpose of a Matrix

Turns a matrix on its side - rows and columns are interchanged

$(A^T)_{ij} = a_{ji}$

If A is an mxn matrix, this becomes an n x m matrix.

We could write a dot product as a matrix multiplication:

We can say (roughly) : $\mathbf{x} \cdot \mathbf{y} = \mathbf{x}^T\mathbf{y}$, assuming we treat a 1x1 matrix as a scalar.


$(AB)^T = B^TA^T$

This is necessary just by seeing the dimensions: AB is m x n - n x p = m x p.

$A^T$ is n x m, $B^T$ is p x n

The transpose has to p x m - which is only possible if we have p x n - n x m, which is $B^TA^T$

transpose and dot product:

Given A: mxn, x in $R^n$, and y in $R^m$

$$A\mathbf{x} \cdot_m y = x \cdot_n A^T y$$

Proof:

$A\mathbf{x} \cdot_m y = (A\mathbf{x})^Ty = (\mathbf{x}^TA^T)y = \mathbf{x}^T(A^Ty) = \mathbf{x}.(A^Ty)$

**Economics use-case for Transpose**


$p = (p_1, p_2, \dots, p_n)$ - prices for cookie ingredients - say 20x1 matrix

x - no of cookies of each type of cookie  5x1 (5 types of cookies)
A - recipe matrix, giving quantities of ingredients needed for each type of cookie  20x5 matrix
y - quantities needing ordering (20x1)

Then:

$Ax = y$

Cost: $p \cdot y = p \cdot Ax = A^Tp.x$

Here : Ax = qty of ingredients needed totally = 20x1 matrix. When dot product by p, we get total cost.

$A^T$ = 5 x 20 matrix, p - 20 x 1 matrix: 5x1 matrix giving unit price per cookie type

Dot Multiply by quantity needed of each cookie (x) and we get total cost







**Inverse of a Matrix**

A is an nxn matrix. We say A is invertible if there is an nxn matrix satisfying AB = I, and BA = I.

If such a B exists, B is called the inverse of A.
