- If diagonals of parallelogram are perp, it's a rhombus, and vice versa.
(x - y).(x + y) = x.x - y.y = 0 (perp)
|x|^2 - |y|^2 = 0. Then |x| = |y| - because |x|, |y| >= 0. 
You can also work backwards, proving the reverse.

- Law of Cosines:
  We have : |x + y|^2 = |x|^2 + |y|^2 + 2x.y = |x|^2 + |y|^2 + 2|x||y|cos(t)
  But law of cosines uses a minus sign, making "obtuse" angles +ve. So if we replace t by pi - t' (the other angle in the //'gram, then we get:

    |x + y|^2 = |x|^2 + |y|^2 + 2x.y = |x|^2 + |y|^2 - 2|x||y|cos(t')
  
- Cauchy-Schwartz Theorem: Prove w/o geometry: given any 2 vectors x and y
  We look at the f(t) |x - ty|^2 and show it has a minimum value.
  Basically proof from book

- Triangle inequality: For any x and y in R^n
   |x + y| <= |x| + |y|, equality holds when either x or y is zero, of one is a positive scalar multiple of the other. Proof is simple
