# Partial Derivatives

- definition of partial derivatives
- function $f(x,y) = \dfrac{xy}{x^2 + y^2}$, f(0) = 0

Is not continuous, but has partial derivatives.

Warning sign : single variable rules don't apply.

Slope of the function - rate of change in a particular direction

Directional derivative of f at a in the direction of v:

$$D_{\mathbf{v}}f(a) = \lim_{t \to 0} \dfrac{f(\mathbf{a} + t\mathbf{v}) - f(\mathbf{a})}{t}$$

If you want slope, you have to use unit vectors.



```python
import sympy as sp
from IPython.display import display, Math

# Define the symbols
x, y = sp.symbols('x y')

def partdiff(f, var_x, var_y):
     df_dx = sp.diff(f, var_x)
     df_dy = sp.diff(f, var_y)
     sp_display = f"{sp.latex(f)} \\quad \\frac{{\\partial f}}{{\\partial x}} = {sp.latex(df_dx)} \\quad \\frac{{\\partial f}}{{\\partial y}} = {sp.latex(df_dy)}"
     display(Math(sp_display))


partdiff(x**3 + 3*x*(y**2) - 2*y + 7, x, y)

partdiff(sp.sqrt(x**2 + y**2), x, y)

partdiff(sp.atan(y / x), x, y)

partdiff(sp.E**-(x**2 + y**2), x, y)
```
