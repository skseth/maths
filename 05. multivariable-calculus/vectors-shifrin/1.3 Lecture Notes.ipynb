Generalizing lines and planes through the origin in Rn

Properties of a plane:

i) 0 in V  
ii) for all v in V and scalar c, cv in V  (v is closed under scalar multiplication)
iii) for all v,w in V, v + w in V  (closed under addition)

Any subset supporting these properties is called a subspace of Rn. Note {0} is a subspace.

Homogeneous linear differential equations have same property. Tangent lines are generalized using this concept as well. So subspaces are fundamental.

Counter-examples: 
- (ii) => (i). So only way a subspace cannot have (i) is if V is empty. So empty set is a subset which is not a subspace because it violates (i)
- 2 lines through origin (violates iii)
- V = [x1,x2], x1,x2 are in Z or Q




Examples: Line

(i) Pick any $\vec{v} \in R^n$. Line through origin: $l = \{\vec{x} \in R^n: \vec{x} = t\vec{v} \text{ for some } t \in R \}$. If $\vec{v}$ is not $\vec{0}$, l is a line.

Prove l is a line:  
(i) taking t = 0, x = 0v = 0  
(ii) Let x = tv is in l. Then cx = (ct)x which is in l, since ct is in R.  
(iii) let x = cv and y = dv be in l. Then x + y = (c + d)v which is in l, because (c+d) is in R.  

Plane: Pick and v,w non-parallel and consider vectors of form x = sv + tw. Non-parallel implies v and w are non-zero.

(i) Pick x = 0v + 0w = 0 + 0 = 0.  
(ii) x = sv + tw. cx = csv + ctw = (cs)v + (ct)w  - cs, ct in R => cx is in V  
(iii) x = sv + tw, y = s'v + t'w. x + y = (s + s')v + (t + t')w => x + y is in V  




**Linear combinations, spans**

- linear combination definition
- definition of span(v1,v2,..vk) = all linear combinations of v1,v2,...

Theorem: span(v1,v2..,vk) is a subspace of Rn, and is called the subspace spanned by v1,v2..vk - a span is the **smallest** subspace containing these vectors. k is unrelated to n.

Another prescription of a subspace: Take any vector A in R^n. Then 

V = {x in R^n: A.x = 0} - when A is not zero, V is called the hyperplane, with normal vector A. Hyperplane is a subspace of dimension n-1.

Language: **standard basis vectors** for Rn

- define $e_1, e_2, \dots, e_n$ in $R^n$.

Span($e_1, e_2, \dots, e_n$) = R^n. Any vector $\vec{x} = (x_1, x_2, \dots, x_n)$  can be expressed as $\vec{x} = x_1\vec{e_1} + ... + x_n\vec{e_n}$.

Given an $\vec{x}$ with unknown coordinates, you can recover the coordinates by taking dot products with respective unit vectors e.g. $x_1 = \vec{x} \cdot \vec{e_1}$



Given a subspace V, define $V^{\perp}$ as :

$V^{\perp} = \{ \vec{x} \in R^n: \vec{x}.\vec{v} = 0 \text{ for all }  \vec{v} \text{ in } V \}$

NOTE: it is not enough to be perpendicular to just one vector in V. Every vector in $V^{\perp}$ is perpendicular to every vector in V.

e.g in $R^4$, if V is a span of $e_1, e_2$, then $V^{\perp}$ is a span of $e_3,e_4$.

$V^{\perp}$ is a subspace. 

(i) 0.v = 0 (since 0 is in V)

(ii) if x is in $V^{\perp}$.  x.v = 0. (cx).v = c(x.v) = c.0 = 0. so cx is in $V^{\perp}$

iii) x,y in $V^{\perp}$: (x + y).v = x.v + y.v = 0 + 0 = 0




Example: Span([1,-1,1]^T) is a line

Then $V^{\perp} = \{ x \in R^3: x_1 - x_2 + x_3 = 0 \}$

$x_1 - x_2 + x_3$ represents the normal to the plane and is called an "implicit representation of the plane". 

An explicit representation involves finding two vectors which span the plane.
