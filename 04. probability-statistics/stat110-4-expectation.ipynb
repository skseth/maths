# Expectation

**arithmetic mean**

Given a list of numbers $x_1, x_2, \dots, x_n$, the arithmetic mean, is defined by:

$\displaystyle \overline{x} = \frac{1}{n}\sum\limits_{j=1}^{n} x_j$

**weighted mean**

Given a list of numbers $x_1, x_2, \dots, x_n$, and weights $p_1, p_2, \dots, p_n$ the weighted mean, is defined by:

$\displaystyle \text{weighted-mean}(x) = \sum\limits_{j=1}^{n} x_jp_j$

The arithmetic mean = weighted mean where $p_j = 1/n$ for all j.

## Definition (Expectation of a discrete r.v.)

The expected value of a discrete r.v. X, with distinct possible values $x_1, x_2, \dots$ is defined by:

$$E(X) = \sum\limits_{j=1}^{\infty} x_jP(X = x_j)$$

Note that if the range or the support of E(X) is finite, this reduces to a finite sum.

## Definition (Expectation of a discrete r.v. using sample space)

The expected value of a discrete r.v. X, on a sample space S is defined as :

$$E(X) = \sum\limits_{s \in S} X(s)P(\{s\})$$

Proof (of equivalence):

Given $X:S \to R$, it follows that for any $x_j \in range(X)$, $s \in X^{-1}(x_j) \implies X(s) = x_j$. What's more,

$P(X = x_j) = \sum\limits_{s \in X^{-1}(x_j)} P(\{s\})$

Also, for all j, $X^{-1}(x_j)$ form a partition of S i.e. $S = \bigcup_j X^{-1}(x_j)$

$E(X) = \sum\limits_{j=1}^{\infty} x_jP(X = x_j) = \sum\limits_{j=1}^{\infty} \sum\limits_{s \in X^{-1}(x_j)} x_j P(\{s\}) = \sum\limits_{j=1}^{\infty}  \sum\limits_{s \in X^{-1}(x_j)} X(s) P(\{s\}) = \sum\limits_{s \in S} X(s) P(\{s\})$

**Divergence**

If the sum $\sum\limits_{j=1}^{\infty} |x_j|P(X = x_j)$ diverges, then E(X) is undefined, as the series then diverges or is order-dependent.


```python
import itertools
import numpy as np
from sympy import Eq, Symbol, LessThan, GreaterThan, And, Rational
from scipy.stats import randint
from sympy.stats import Die, Coin, density, P, E, sample, Bernoulli, Poisson, Binomial
import matplotlib.pyplot as plt

def plot_discrete_pmf(ax, xaxis, xlabels, pmf, *, ms:int=8, lw:int=5):
    if xlabels is not None:
        ax.set_xticks(xaxis, xlabels)
    ax.plot(xaxis, pmf, 'bo', ms=ms, label='randint pmf')
    ax.vlines(xaxis, 0, pmf, colors='b', lw=lw, alpha=0.5)


def plot_discrete_pmf_rv(ax, rv, **kw_args):
    den = density(rv)
    xlabels = np.array(list(den.keys()))
    xaxis = np.arange(0,xlabels.size)
    pmf = np.array(list(den.values()))
    plot_discrete_pmf(ax, xaxis, xlabels, pmf, **kw_args)


def plot_discrete_pmf_distribution(ax, distribution, *, start_index=None, end_index=None, **kw_args):
    den = density(distribution).dict
    pmf = np.array(list(den.values())[start_index: end_index])
    xaxis = list(den.keys())[start_index:end_index]

    plot_discrete_pmf(ax, xaxis, None, pmf, **kw_args)


def plot_discrete_cdf(ax, cdf, *, ms:int=8, lw:int=5):
    xaxis = list(cdf.keys())
    values = list(cdf.values())

    xaxis.insert(0, xaxis[0])
    values.insert(0, 0)

    ax.plot(xaxis, values, ms=ms, label='cdf', drawstyle='steps-post')
    # ax.vlines(xaxis, 0, values, colors='b', lw=lw, alpha=0.5)

def discrete_cdf_from_rv(rv):
    pmf = density(rv).dict
    return dict(zip(pmf.keys(), np.cumsum([float(i) for i in pmf.values()])))

def expect(distribution):
    s = 0
    for x,p in density(distribution).dict.items():
        s += x*p
    return s
```

## Proposition (discrete r.v.'s with same distribution have same expected value)

If X and Y have the same distribution, then E(X) = E(Y). 

Proof: Follows immediately from the definition of distribution, and expected value.

**NOTATION**

We often write :

$$EX = E(X)$$

$$EX^2 = E(X^2) - \text{ not } E(X)^2$$

$$EX^n = E(X^n)$$



## Theorem (Linearity of Expectation): For any r.v.'s X,Y and any constant c:

$$E(X + Y) = E(X) + E(Y)$$

$$E(cX) = cE(X)$$

Proof:
Using the sample space definition:

$E(X+Y) = \sum_{s \in S} (X+Y)(s)P(\{s\}) = \sum_{s \in S} (X(s)+Y(s))P(\{s\}) = E(X) + E(Y)$

Similar proof for the other condition.


**NOTE: Expectation of Product of Independent r.v.'s**

Let X and Y be independent r.v.'s. Then:

$E(XY) = \sum_{x,y} xyP(X=x)P(Y=y) =  \sum_{x,y} xP(X=x)yP(Y=y) = \sum_{x} xP(X=x) \sum_y yP(Y=y) = EY \sum_{x} xP(X=x) = EX.EY$

See : https://stats.stackexchange.com/questions/578090/expectation-of-the-product-of-iid-random-variables for a great explanation

## Theorem (Binomial Expectation) : For $X \thicksim Bin(n,p), E(X) = np$

Given $I_1, I_2, \dots, I_n$ as indicator Bern(p) r.v.'s for each of the n (independent) trials.

$X = I_1 + I_2 + \dots + I_n$. Then $E(X) = E(I_1 + I_2 + \dots + I_n) = E(I_1) + E(I_2) + \dots + E(I_n) = np$

## Theorem (Hypergeometric Expectation) : For $X \thicksim HGeom(w,b,n), E(X) = n\frac{w}{w+b}$

Given $I_1, I_2, \dots, I_n$ as indicator Bern(p) r.v.'s for each of the n (dependent) trials, but basically the same approach works, thanks to the linearity of expectations.

$X = I_1 + I_2 + \dots + I_n$. Then $E(X) = E(I_1 + I_2 + \dots + I_n) = E(I_1) + E(I_2) + \dots + E(I_n) = n\frac{w}{w+b}$

This is somewhat surprising: Why are the $I_n \thicksim Bern(n,p)$. Because (as discussed earlier), whether we pick n balls one by one, or altogether makes no difference, and every ball has equal opportunity of being in any selection - hence the Bern(n,p). The fact that the Binom / Hypergeometric distributions have such similar expectations is surprising, but correct.

## Proposition (Monotonicity of Expectations)

Let X and Y be r.v.'s such that $X \ge Y$ with probability 1. Then $E(X) \ge E(Y)$, with equality holding iff P(X = Y) = 1.

Proof(Discrete r.v.'s only):

Define Z = X - Y. Then $P(Z \ge 0) = 1 \implies E(Z) \ge 0$. Thus $E(X - Y) = E(X) - E(Y) \ge 0 \implies E(X) \ge E(Y)$.

If E(X) = E(Y), then E(Z) = 0. Since value of Z is greater than or equal to 0 everywhere, Z must be zero everywhere i.e. P(X - Y = 0) = 1.



# Geometric and Negative Binomial

## Geometric Distribution

### Definition (Geometric Distribution, PMF, CDF - Geom(p))**

A sequence of independent, Bern(p) trials (Sampling with replacement). Let X be an r.v. for the number of failures till the first success. Then $X \thicksim Geom(p)$.

PMF of Geom(p): r.v = $q^kp$

CDF of Geom(p): $F(x) = \begin{cases} 1 - q^{\lfloor x \rfloor + 1} &\text{ if } x \ge 0 \\ 0 &\text{ if } x \lt 0\end{cases}$

$E(X) = \frac{q}{p}$

**Proof(PMF is valid):**

Note that $\sum_{k=0}^{\infty} q^k$ is a geometric series, whose sum is $\frac{1}{1-q}$, when q < 0

$sum_{k=0}^{\infty} q^kp = p sum_{k=0}^{\infty} q^k = p.\frac{1}{1-q} = p.\frac{1}{p} = 1$.

**Proof(CDF)**

Note: $\sum\limits_{k=0}^{k} q^k = \frac{1 - q^{k + 1}}{1-q}$ - sum of a geometric series with  k + 1 terms. The +1 comes from the fact that the series starts from 0.

$F(x) = \sum\limits_{k=0}^{\lfloor x \rfloor} q^{\lfloor x \rfloor}p = p . \frac{1 - q^{\lfloor x \rfloor + 1}}{1-q} = 1 - q^{\lfloor x \rfloor + 1}$

**Proof(Expectation)**

First, note that:

$\sum_{k=0}^{\infty} q^k = \frac{1}{1-q}$. Differentiating, we get  

$\sum_{k=0}^{\infty} kq^{k-1} = \frac{1}{(1-q)^2}$. 

$E(X) = \sum\limits_{k=0}^{\infty} kq^kp = pq \sum\limits_{k=0}^{\infty} kq^{k-1} = pq\frac{1}{(1-q)^2} = \frac{q}{p}$

### Alternate Definition (Geometric Distribution - First Success)

Sometimes Geometric distribution is defined as the number of failures + the one success. If we represent Y as this alternate definition, and X as the number of failures only, then:

Y = X + 1 i.e.  $Y - 1 \thicksim Geom(p)$

$P(Y = k) = P( Y-1 = k-1 ) = q^{k-1}p$

CDF of Y : $P(Y \le x) = P(Y-1 \le x-1) = \begin{cases} 1 - q^{\lfloor x \rfloor } &\text{ if } x \ge 1 \\ 0 &\text{ if } x \lt 1\end{cases}$

Expectation: $E(Y) = E((Y-1) + 1) = \frac{q}{p} + 1 = \frac{1}{p}$

## Negative Binomial Distribution

Generalizes Geometric Distribution - we wait for r successes. Given independent Bern(p) trials, X = number of failures till rth success. $X \thicksim NBin(r,p)$

We can look at the negative binomial as a sum of r geometric r.v.'s. 

$X = X_1 + X_2 + \dots X_r$, where each $X_i$, i = 1 to r is an r.v. Geom(p)


PMF : $\displaystyle P(X = n) = \binom{n+r-1}{r-1} p^rq^n$

E(X) = $\displaystyle \frac{rq}{p}$


**Proof(Expectation)**:

$E(X) = E(X_1) + E(X_2) + \dots + E(X_r) = \frac{rq}{p}$

### Coupon Collector Problem

n types of coupons. X = number of (random, equally likely) purchases till you get all n. Find E(X).

Let $N_1, N_2, ...N_n$ represent number of purchases till the ith item type. Note these are not i.i.d.

Each $N_i$ is a Geometric Distribution (w/ first success included). 

$X = N_1 + N_2 + \dots + N_n$.

$E(N_i) = \frac{1}{p_i} = \frac{n}{n-i+1}$

$E(X) = \sum_{i=1}^{n} \frac{n}{n-i+1} = n \sum_{i=1}^{n} \frac{1}{i}$

Note that sum of harmonic series : $\sum_{i=1}^{n} \frac{1}{i}$, for which we only have approximations, no closed form.

### St. Petersburg Paradox (Expectation of a non-linear function of r.v.)

Suppose wealthy stranger offers to play following game - flip fair coin - 1st head - 2$, 2nd head - 4$, nth head - 2^n dollars. Let X be the amount you win (payoff). How much would you pay to play?


Ans: 

Let N = number of rounds the game lasts. i.e. till the first tails. This is basically a Geometric (first-success) dist., so

$E(N) = 1/(1/2) = 2$ - thus, you may get $2^{2} = 4$ dollars.  

$E(X) = \sum\limits{k=1}^{\infty} 2^k(1/2)^k = \sum\limits{k=1}^{\infty} 1 = \infty$

This shows $E(g(X)) \ne g(E(x))$, for non-linear functions. Here $X = 2^N$ is a non-linear function over N. So $E(g(N)) = E(2^N) \ne g(E(N)) = 2^{E(N)}$.

Also, note a few things - suppose we cap the game at 40 rounds:

$E(X) = \sum\limits_{k=1}^{40} 2^k(1/2)^k + \sum\limits_{k=41}^{\infty} 0.(1/2)^k = 40$ - remember 40 rounds would be close to a trillion dollars, and in any case is extremely unlikely. 





# Indicator r.v.s

## Theorem (Indicator r.v. properties)

Let A, B be events on some sample space. Then:

1. $(I_A)^k = I_A$
2. $I_{A^c} = 1 - I_A$
3. $I_{A \cap B} = I_AI_B$
4. $I_{A \cup B} = I_A + I_B - I_AI_B$

Proof: Easy, from definition.

## Fundamental Bridge (between Probability and Expectation)

$P(A) = E(I_A)$

Proof: $I_A \thicksim Bern(P(A))$, thus $E(I_A) = P(A)$

### Boole, Bonferroni, and the Principle of inclusion-exclusion

**Boole/Bonferroni inequality**

Given events $A_1, A_2, \dots, A_n$ in a sample space:

Note that   

$I(A_1 \cup A_2 \dots A_n) \le I_{A_1} + I_{A_2} + \dots + I_{A_n}$,  

since whenever the outcome is in one of the events, at least one (and maybe more) of the right side may be one. But when the left side is 0, the right must be zero also.

Thus: $E(I(A_1 \cup A_2 \dots A_n)) \le E(I_{A_1}) + E(I_{A_2}) + \dots + E(I_{A_n})$,  using linearity of expectation.

Finally: $P(A_1 \cup A_2 \dots A_n) \le P(A_1) + P(A_2) + \dots + P(A_n)$,  using the "fundamental bridge".

**Principle of Inclusion-Exclusion**

$$\begin{align*}
1 - I(A_1 \cup A_2 \dots A_n) &= I(A_1^c \cap A_2^c \dots A_n^c) \\
                              &= I(A_1^c)I(A_2^c) \dots I(A_n^c) \\
                              &= (1 - I(A_1))(1-I(A_2))\dots(1-I_{A_n}) \\
                              &= 1 - \sum_{1 \le i \le n} I(A_i) + \sum_{1 \le i \lt j \le n} I(A_i)I(A_j) + \dots + (-1)^n I(A_1)I(A_2)\dots I(A_n) \\
\implies 1 - E(I(A_1 \cup A_2 \dots A_n)) &= 1 - \sum_{1 \le i \le n} E(I(A_i)) + \sum_{1 \le i \lt j \le n} E(I(A_i)I(A_j)) + \dots + (-1)^n E(I(A_1)I(A_2)\dots I(A_n)) \\
                              
\implies 1 - P(A_1 \cup A_2 \dots A_n) &= 1 - \sum_{1 \le i \le n} P(A_i) + \sum_{1 \le i \lt j \le n} P(A_i \cap A_j) + \dots + (-1)^n P(A_1 \cap A_2 \dots \cap A_n) \\

\implies P(A_1 \cup A_2 \dots A_n) &= \sum_{1 \le i \le n} P(A_i) - \sum_{1 \le i \lt j \le n} P(A_i \cap A_j) + \dots + (-1)^{n+1} P(A_1 \cap A_2 \dots \cap A_n) \\

\end{align*}$$









### Montmort's Problem with Indicator r.v.'s

X = number of matches of well-shuffled deck. Find E(X)

Let $I_i$ be the event the ith card matches. $E(I_j) = 1/n$. Then $X = I_1 + I_2 + \dots + I_n$ and $E(X) = \sum_j E(I_j) = 1$.

### Birthday Problem with Indicator r.v.'s

a) n people, how many distinct birthdays? X = number of distinct days people were born.

Let $I_j$ represent anyone being born on jth day. Then $E(I_j)$ = P(someone is born on jth day) = 1 - P(no one is born on jth day) = $1-(\frac{364}{365})^n$

$E(X) = \sum_{1 \le j \le 365} 1-(\frac{364}{365})^n = 365(1-(\frac{364}{365})^n)$

b) How may pairs of people have same birthday? Let X = no of pairs with same birthday.

There are $N = \binom{n}{2}$ distinct pairs. Let $I_j$ represent probability of jth pair being born on the same day. Then 

$E(I_j) = P(jth pair being born on same day) = 1/365$

$E(X) = \sum_{1\le j \le N} 1/365 = \frac{\binom{n}{2}}{365}$

### Putnam Problem

A permutation $a_1,a_2,\dots,a_n$ of 1,2,...,n has a local maximum at j if $a_j$ is greater than its neighbors. For n >= 2, what is the average number of local maxima of a random permutation of 1,2,...,n, with all n! permutations equally likely.

Let $I_j$ be indicator variable for the jth position being a local maxima. Then $E(I_j) = 1/3$, for j = 2 to n-1. $E(I_1) = E(I_n) = 1/2$.

$E(X) = 1/2*2 + sum_{1 < j < n} 1/3$ = 1 + (n-2)/3 = n+1/3


# Negative Hypergeometric Distribution

From w white and b black balls, sample without replacement. Let X be the number of black balls sampled till we get r white balls. Then $X \thicksim NHGeom(w,b,r)$.

PMF: $\displaystyle P(X = k) = \frac{\binom{w}{r-1}\binom{b}{k}}{\binom{w+b}{k+r-1}}.\frac{w-r+1}{w+b-k-r+1}$, k=0,1..b

$E(X) = \frac{rb}{w+1}$

**Proof (PMF)**

Pick r-1 white balls, k black balls and then select one more white ball.

PMF: $\displaystyle P(X = k) = \frac{\binom{w}{r-1}\binom{b}{k}}{\binom{w+b}{k+r-1}}.\frac{w-r+1}{w+b-k-r+1}$

Alternatively: focusing just on the white balls: Out of the w+b locations, we can put w white balls anywhere we want. But for the desired k, in the first k+r-1 balls, there are exactly r-1 white balls, in the last w+b-k-r balls, there are w-r white balls. Thus:

**Proof (Expectation)**

Imagine a line of w+b spots. If there are w white balls, then we have w+1 spots where a black ball can possibly go $X_1, X_2, X_{w+1}$ denote the number of black balls in each location. $E(X_1) = \frac{b}{w+1}$ - there are b black balls and w+1 locations (the +1 is the location after the last white ball).

$\displaystyle E(X) = E(X_1) + E(X_2) + \dots + E(X_r) = \frac{rb}{w+1}$


# Survival Function and Expectation

**Definition (Survival Function)**

Let X be a **non-negative, integer-valued** r.v. Let F be CDF of X, and $G(X) = 1 - F(X) = P(X > x)$. Then G is called the **survival function** of X.

**Expectation (via Survival Function)**

$E(X) = \sum\limits_{n=0}^{\infty} G(n)$

Intuitively - the numbers appearing most often in the survivor's list will have highest weightage.

Proof:

Suppose $I_j = I(X \gt j)$, j=0,1,..  Then :

$X = \sum_{j=0}^{\infty} I_j$

For example, if X = 3, then $I_0,I_1,I_2$ will be 1, the rest will be 0. (This is why we are using the integer-valued condition, and the non-negative condition).

Then, by linearity of expectation: $E(X) = \sum_{j=0}^{\infty} E(I_j) = \sum_{j=0}^{\infty} P(X \gt j) = \sum_{j=0}^{\infty} G(j)$

**Geometric Mean Expectation (via Survival Function)**

$X \thicksim Geom(p)$. Then $X \gt k$ is the event that first k+1 trials are failures.$X \gt k = q^{k+1}$.

$\displaystyle E(X) = \sum_{k=0}^{\infty} q^{k+1} = \frac{q}{1-q} = \frac{q}{p}$

# LOTUS - Law of the Unconscious Statistician

### Theorem (LOTUS): If X is a discrete r.v. and $g:R \to R$ is a function, then:

$$E(g(X)) = \sum_x g(x) P(X = x)$$

**Proof:**

Let $x_1, x_2,\dots$ form the support of X.

$$\begin{align*}
E(g(X)) &= \sum_{s \in S} g(X(s)) P(\{s\}) \\
        &= \sum_x \sum_{s:X(s)= x} g(X(s)) P(\{s\}) \\
        &= \sum_x \sum_{s:X(s)= x} g(x) P(\{s\}) \\
        &= \sum_x g(x) \sum_{s:X(s)= x} P(\{s\}) \\
        &= \sum_x g(x) P(X = x) \\
\end{align*}$$

# Variance and Standard Deviation


**Definition (Variance and Standard Deviation)**

The variance of an r.v. X is:

$$Var(X) = E((X - EX)^2)$$

i.e. the expected value of the square of the difference of the r.v. from the mean.

The standard deviation:

$$SB(X) = \sqrt{Var(X)}$$

**Points to note:**

- Why not use $E(X - EX) = E(X) - E(E(X)) = E(X) - E(X) = 0$. Note that E(E(X)) = expected value of a constant. This is 0 because on averge (by definition) values of X are equally distributed around the mean and cancel out.
- Why not use $E|X - EX|$?  Lack of continuity makes it harder to work with than variance.
- Why SD? Because it gets back to the right units of whatever X is.

## Theorem (Variance): For any r.v. X, $Var(X) = EX^2 - (EX)^2$

Proof:

$$\begin{align*} 
Var(X)  &= E((X - EX)^2) \\
        &= E(X^2 + (EX)^2 - 2X.EX) \\
        &= EX^2 + E((EX)^2) - E(2X.EX) \\
        &= EX^2 + (EX)^2 - 2(EX)^2 \\
        &= EX^2 - (EX)^2 \\
\end{align*}$$


**Variance properties**

- $Var(X + c) = Var(X)$, for any constant c

- $Var(cX) = c^2Var(X)$, for any constant c

- Var(X) >= 0, with equality only when P(X = a) = 1 for some constant a in R.

- If X and Y are independent: Var(X + Y) = Var(X) + Var(Y). In general, if X and Y are dependent, Var(X + Y) is not the sum of the variances. E.g. if Y = X, then Var(X + Y) = Var(2X) = 4Var(X), provided Var(X) > 0.

Proof:

- $E(X + c)^2 - (E(X + c))^2 = E(X^2 + 2cX + c^2) - (E(X) + E(c))^2 = EX^2 + 2cEX + c^2 - (EX)^2 - 2cE(X) - c^2 = EX^2 = (EX)^2 = Var(X)$
- $E(cX)^2 - (E(cX))^2 = E(c^2X^2) - (cE(x))^2 = c^2(EX^2 - (EX)^2) = c^2Var(X)$
- $Var(X) = E(X - E(X))^2$, so necessarily $Var(X) \ge 0$. Suppose $Var(X) = E(X - E(X))^2 = 0$, it implies that for every x in the support of X, x - E(X) = 0, other if even one x was different from E(X), the sum would be positive. Thus, P(X = a) = 1, where a is some real number. On the other hand, suppose X is a constant, say X = a, then clearly E(X) = a, and EX^2 = a^2, so Var(X) = 0.
- This will be proved later.


Note:
- Variance is not linear.


## Variances of discrete distributions

### Geometric Variance = $\displaystyle \frac{q}{p^2}$

This is also the variance for the "first success" variance, since adding a constant does not change variance.

Let $X \thicksim Geom(p)$. E(X) = q/p. 

$$\begin{align*}
E(X^2) &= \sum\limits_{k=0}^{\infty} k^2P(X = k) &\text{ applying LOTUS}\\
       &= \sum\limits_{k=1}^{\infty} k^2pq^k &\text{ for k=0, value is zero} \\
       &= p.\sum\limits_{k=1}^{\infty} k^2q^k &\text{ for k=0, value is zero}
\end{align*}$$

Now:

$$\begin{align*}
\sum\limits_{k=1}^{\infty} q^k  &= \frac{1}{1-q} &\text{ geometric series sum} \\
\sum\limits_{k=1}^{\infty} kq^{k-1}  &= \frac{1}{(1-q)^2} &\text{ taking derivatives of an infinite series ...} \\
\sum\limits_{k=1}^{\infty} kq^{k}  &= \frac{q}{(1-q)^2} &\text{ multiplying by q ...} \\
\sum\limits_{k=1}^{\infty} k^2q^{k-1}  &= \frac{1+q}{(1-q)^3} &\text{ once more ...} \\
\end{align*}$$

Substituting, we get
$$\begin{align*}
E(X^2)  &= pq\frac{1+q}{(1-q)^3} = \frac{q(1+q)}{p^2} &\text{ since p = 1-q} \\
Var(X)  &= \frac{q(1+q)}{p^2} - (q/p)^2 \\
        &= \frac{q(1+q) - q^2}{p^2} \\
        &= \frac{q}{p^2}
\end{align*}$$


## Negative Binomial Variance = $\displaystyle \frac{rq}{p^2}$

Proof: If X has an NBin(r,p) distribution, it is the sum of r i.i.d Geom(p) r.v.s, hence we can use the additivity property. 

## Binomial Variance = $\displaystyle np(1-p)$

Proof:

# Poisson Distribution

Poisson distribution provides a distribution for the probability of occurence of a large number of events, each individuall of very small probability.

e.g. number of customers arriving at a store per hour. Here the arrival of each possible customer 

**Poisson Distribution**

An r.v. has the Poisson distribution with parameter $\lambda > 0$, if the PMF of X is:

$$P(X = k) = \frac{e^{-\lambda}\lambda^k}{k!}$$

We write $X \thicksim Pois(\lambda)$

$E(X) = \lambda$

$Var(X) = \lambda$

**Proof(Valid PMF)**

$\displaystyle \sum\limits_{k=0}^{\infty} \frac{e^{-\lambda}\lambda^k}{k!} = e^{-\lambda} \sum\limits_{k=0}^{\infty} \frac{\lambda^k}{k!} = e^{-\lambda}e^{\lambda} = 1$

**Proof(Expectation)**

$$\begin{align*}
E(X) = \sum\limits_{k=0}^{\infty} \frac{ke^{-\lambda}\lambda^k}{k!} = \lambda e^{-\lambda} \sum\limits_{k=1}^{\infty} \frac{\lambda^{k-1}}{k-1!} = \lambda e^{-\lambda} e^{\lambda} = \lambda

\end{align*}$$

**Proof(Variance)**

First a lemma:

$$\begin{align*}

\text{ Starting with } \\
\sum\limits_{k=0}^{\infty} \frac{\lambda^k}{k!} &= e^{\lambda} \\
\sum\limits_{k=0}^{\infty} \frac{k\lambda^{k-1}}{k!} &= e^{\lambda} &\text{differentiating} \\
\sum\limits_{k=0}^{\infty} \frac{k\lambda^k}{k!} &= \lambda e^{\lambda} &\text{ multiply} \\
\sum\limits_{k=0}^{\infty} \frac{k^2\lambda^{k-1}}{k!} &= (1 + \lambda) e^{\lambda} &\text{ differentiate} \\
\sum\limits_{k=0}^{\infty} \frac{k^2\lambda^k}{k!} &= \lambda(1 + \lambda) e^{\lambda} &\text{ multiply}
\end{align*}$$

Now:
$$\begin{align*}
E(X^2)  &= \sum\limits_{k=0}^{\infty} \frac{k^2e^{-\lambda}\lambda^k}{k!} \\
        &= e^{-\lambda} \sum\limits_{k=1}^{\infty} \frac{\lambda^{k-1}}{k-1!} \\
        &= e^{-\lambda} e^{\lambda} \lambda(1 + \lambda) \\
        &= \lambda(1 + \lambda) \\
Var(X)  &= \lambda(1 + \lambda) - \lambda^2 = \lambda
\end{align*}$$

See: [Can one differentiate an infinite sum](https://math.stackexchange.com/questions/1139739/can-one-differentiate-an-infinite-sum)

### Poisson Paradigm

Let $A_1, A_2, \dots, A_n$ be events with $p_j = P(A_j)$, where n is large, the $p_j$ are small, and the $A_j$ are independent or weakly dependent. Let:

$$X = \sum\limits_{j=1}^{n} I(A_j)$$

count how many of the $A_j$ occur. Then X is **approximately distributed** as $Pois(\lambda)$ with $\lambda = \sum_{j=1}^{n} p_j$


**Theorem (Upper bound on Error From Poisson Distribution - Law of rare events)**

$$|P(X \in B) - P(N \in B)| \le \min(1,\frac{1}{\lambda}) \sum\limits_{j=1}^{n} p_j^2$$

A proof of this requires the Stein-Chen method - see : https://dornsife.usc.edu/larry-goldstein/wp-content/uploads/sites/221/2023/06/pacs-1.pdf

The basic requirement is that the sum of the square of the probabilities remains much smaller than $\lambda$, to keep the error low. "Rare" refers to the $p_j$'s being low, not $\lambda$.

Note that a Poisson distribution in theory has no upper bound, but in practice we do use it in situations that have bounds. It is a good model for **count data** i.e. distributions whose values are non-negative integers, and where the probability of any number occuring is very low.

## Connections between Binomial and Poisson

### Theorem (Sum of Independent Poisson's)

If $X \thicksim Pois(\lambda_1)$, and $Y \thicksim Pois(\lambda_2)$, and X is independent of Y, then: $X + Y \thicksim Pois(\lambda_1 + \lambda_2)$

Intuitively, if two independent events are happening at rate $\lambda_1$ and $\lambda_2$, the overall rate is $\lambda_1 + \lambda_2$

**Proof**

$$\begin{align*}
P(X + Y = k)    &= \sum_{j=0}^{k} P(X + Y = k| X = j)P(X = j) &\text{LOTP}\\
                &= \sum_{j=0}^{k} P(Y = k-j|X =j)P(X = j) \\
                &= \sum_{j=0}^{k} P(Y = k-j)P(X = j) &\text{independence}\\
                &= \sum_{j=0}^{k} P(Y = k-j)P(X = j) &\text{independence}\\
                &= \sum_{j=0}^{k} \frac{e^{-\lambda_1}\lambda_1^{k-j}}{(k-j)!}\frac{e^{-\lambda_2}\lambda_2^j}{j!} &\text{independence}\\
                &= e^{-\lambda_1 - \lambda_2}\sum_{j=0}^{k} \frac{\lambda_1^{k-j}}{(k-j)!}\frac{\lambda_2^j}{j!} \\
                &= \frac{e^{-\lambda_1 - \lambda_2}}{k!}\sum_{j=0}^{k} \binom{k}{j} \lambda_1^{k-j}\lambda_2^j \\
                &= \frac{e^{-\lambda_1 - \lambda_2}(\lambda_1 + \lambda_2)^k}{k!} &\text{binomial theorem}\\
\end{align*}$$


## Theorem (Binomial approaches Poisson at high n, low p)

If $X \thicksim Bin(n,p)$ and we let $n \to \infty$ and $p \to 0$ such that $\lambda = np$ stays fixed, then PMF of X converges to PMF of $Pois(\lambda)$. More generally, the same conclusion holds if $n \to \infty$ and $p \to 0$, such that np converges to a constant $\lambda$.

Proof:

$$\begin{align*}

\text{Note}: p &= \frac{\lambda}{n} \\
\text{Also}: \binom{n}{k}p^k &= p^k\frac{n(n-1)\dots(n-k+1)}{k!} = \frac{\lambda^k}{k!}1.(1-1/n)\dots(1-(k-1)/n)\\


P(X = k)    &= \binom{n}{k} p^k (1-p)^{n-k} = \frac{\lambda^k}{k!}1.(1-1/n)\dots(1-(k-1)/n)(1 - \lambda/n)^{n-k} \\
            &= \frac{\lambda^k}{k!}1.(1-1/n)\dots(1-(k-1)/n)(1 - \lambda/n)^n(1 - \lambda/n)^{-k} \\
\lim_{n \to \infty, p \to 0} P(X = k) &= \frac{\lambda^k}{k!}1.(1)\dots(1)e^{\lambda}.1 =  e^{\lambda}\frac{\lambda^k}{k!}\\

\end{align*}$$

# Information Theory Application

A code is a function that assigns to each k-bit input message an equivalent n-bit message - a codeword (rate of the code = k/n). A decoder takes c(x), possibly corrupted, and retrieves the correct x.

Channel flips bit with probability p. 

**Hamming Distance**: Given 2 binary vectors, the Hamming distance is the number of bits where they differ : d(v,w) = \sum_i |v_i - w_i|

**Binary Entropy function H**: For $0 < p < 1, H(p) = -plog_2p -(1-p)log_2(1-p)$

Interpretation of H(p) : Amount of information you get from a Bern(p) trial r.v. 

H(1/2) = -1/2(-1) - (1/2)(-1) = 1 => A fair coin gives us 1-bit of information
H(1) = -1.0 - 0.x = 0. => A coin always landing head gives us no information

**Theorem (Shannon)** Consider a channel where each bit gets flipped with probability p, independently. Let 0 < p < 1/2 and $\epsilon > 0$. There exists a code with rate at least $1 - H(p) - \epsilon$, that can be decoded with error probability less than $\epsilon$.


Proof:

*prelimnaries*
Assume $1 - H(p) - \epsilon > 0$, else this would be pointless. 
Let n be a very large number (discussed below).
Define $k = \lceil n(1 - H(p) - \epsilon) \rceil + 1$. Choose $p' \in (p,1/2)$, such that $|H(p') - H(p)| < \epsilon/2$, this can be done since H is continuous.

*Enter a random code*
Consider a *random code* C. For each k-bit input vector x, choose C(x) to be a uniformly random n-bit vector, By definition $k/n > 1 - H(p) - \epsilon$.

When a k-bit input x is sent, it is sent as an n-bit C(x) message, and received as an n-bit message Y. We hope that d(C(x), Y) < d(C(m), Y) for all other messages m. To do this, we follow the following decode rule: If there is a unique z such that d(C(z), Y) < np', decode Y to that z, else declare failure.

If n is large enough, error rate will be less than $\epsilon$.

*Choosing the right n*

Let $B_1, B_2, \dot, B_n$ represent the event that a bit is flipped (with prob p) - each is a Bern(p), independent event. Thus:

$d(C(x), Y) = B_1 + B_2 + \dots + B_n$. Thus, $d(C(x), Y) \thicksim Binom(n,p)$. 

By law of large numbers, $d(C(x), Y) \to np$, it's expected value i.e. d(C(x), Y)/n will get very close to p, and hence be less than p'. 

$P(d(C(x), Y) > np') = P(\frac{B_1 + \dots + B_n}{n} > p') \to 0 \text{ as } n \to \infty$. This implies, choosing n large enough, that:

$P(d(C(x), Y) > np') < \epsilon/4$

*Getting n big enough, to make other messages far enough*

Now we need to make sure no impostor message gets within distance np' of Y. 

Consider any other random message $z \ne x$. $d(C(z), Y) \thicksim Bin(n,1/2)$, for all $x \ne z$. 


$$\begin{align*}
P(d(C(z), Y) \le np' \text{ for some } z \ne x) &<= \sum\limits_{\text{message} z \ne x} P(d(C(z), Y) <= np') &\text{by Boole's inequality} \\
                                                &<= (2^k - 1) \sum\limits_{j=0}^{np'} \frac{\binom{n}{j}}{2^n} \\
                                                &<= (2^k - 1) \frac{np' + 1}{2^n}{\binom{n}{np'}} \\
                                                &<= (2^k - 1)(np' + 1)(2^{nH(p')-n}) &\text{by lemma below}\\
                                                &<= (2^k)(np' + 1)(2^{nH(p')-n}) \\ 
                                                &<= (np' + 1)(2^{k + nH(p')-n}) \\ 
                                                &<= (np' + 1)(2^{n(1 - H(p) - \epsilon) + 2 + n(H(p) + \epsilon/2)-n}) &\text{since } k < n(1 - H(p) - \epsilon) + 2\\ 
                                                &= 4(np' + 1)2^{-n\epsilon/2}

\end{align*}$$

Thus, as $n \to \infty$, $P(d(C(z), Y) \le np' \text{ for some } z \ne x) \to 0$. Thus, we can choose n such that $P(d(C(z), Y) \le np' \text{ for some } z \ne x) \lt \epsilon/4$. 


lemma for $\binom{n}{np'}$:

$$\begin{align*}
\text{By stirling's approximation} \\

n! \approx (\frac{n}{e})^n\sqrt{2\pi n} \\

\binom{n}{np'} &\approx \frac{n!}{(n-np')! np'!} \\
               &= \frac{(\frac{n}{e})^n\sqrt{2\pi n}}{(\frac{n-np'}{e})^{n-np'}\sqrt{2\pi (n-np')}(\frac{np'}{e})^{np'}\sqrt{2\pi np'}} \\
               &= \frac{({n})^n\sqrt{2\pi n}}{(n-np')^{n-np'}\sqrt{2\pi (n-np')}(np')^{np'}\sqrt{2\pi np'}} \\
               &= \frac{1}{p'^{np'}q'^{nq'}}\frac{1}{\sqrt{2\pi npq}} \\
               &<= \frac{1}{p'^{np'}q'^{nq'}} \\

log_2{\frac{1}{p'^{np'}q'^{nq'}}} &= np'log(p') + nq'log(q') = nH(p') \\
\binom{n}{np'} &<= 2^{nH(p')}

\end{align*}$$
Above is From: https://saadquader.wordpress.com/2014/01/17/upper-bounds-on-binomial-coefficients-using-stirlings-approximation/


*There exists a good-enough code*

Assume we have chosen k, n such that, for a fixed x,

$P(d(C(x), Y) > np') < \epsilon/4, P(d(C(z), Y) \le np' \text{ for some } z \ne x) \lt \epsilon/4$

Let F(C,x) denote a failure event for x. This can happen in the two ways shown above - either $d(C(x), Y) > np'$, or $d(C(z), Y) \le np'$ for some other message. Thus: $P(F(C,x)) = \epsilon/2$.

Thus, for a given x there is a code which gives the desired low-error rate. But we want to show that a single code exists that works well for all x. 

Let X be a uniformly randome k-bit message, independent of C. By LOTP:

$P(F(C,X)) = \sum\limits_x P(F(C,x))P(X = x) \lt \epsilon/2$

But using LOTP, but conditioning on C:

$\sum_c P(F(c,X)) P(C = c) = P(F(C,X)) < \epsilon/2$

Thus, there exists a code c with probability of failure for a random input message is less than $\epsilon/2$. But we have one more step.

*Purge the bad messages*

For this code, we will purge the worst 50% of input messages i.e. we will remove as legal input messages $2^{k-1}$ values of x, with the highest failure probabilities. Now, for all remaining input messages, $P(F(c,x)) < \epsilon$. Because if not, then suppose one remaining message had failure probability $\epsilon$. Now, slightly more than half ($2^{k-1} + 1$) input messages will have failure rates at least equal to double the average - i.e. P(F(c, X)) > $\frac{(2^{k-1}+1)\epsilon}{2^k} > \epsilon/2$, which contradicts our assumption that $P(F(c,X)) < \epsilon/2$.

Now, for the remaining, we have a set of (k-1)-bit vectors, and a code c' which encodes them such that the failure rate is $< \epsilon$ for every message.

The rate = k-1/n. By our original definition : $k = \lceil n(1 - H(p) - \epsilon) \rceil + 1 \le n(1 - H(p) - \epsilon) + 1.$. It follows that :

$$\frac{k-1}{n} \le 1 - H(p) - \epsilon$$
