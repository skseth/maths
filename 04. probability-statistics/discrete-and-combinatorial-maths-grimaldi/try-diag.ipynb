# Introduction to Combinatorics: Rule of Sum and Rule of Product

In combinatorics, the **Rule of Sum** and the **Rule of Product** are the foundational building blocks for counting possibilities. This notebook breaks down both rules using math equations and visual diagrams.

## 1. The Rule of Sum (The "OR" Rule)

### Definition
If you have $A$ ways to do one action, and $B$ ways to do another action, and these actions **cannot** be done at the same time (they are mutually exclusive), then there are $A + B$ total ways to choose one of the actions.

### Mathematical Equation
$$\text{Total Ways} = |A \cup B| = |A| + |B| \quad \text{where } A \cap B = \emptyset$$

```python
# Rule of Sum Diagram: Choosing 1 item from 2 disjoint sets
import matplotlib.pyplot as plt
import networkx as nx

fig, ax = plt.subplots(figsize=(6, 4))
G = nx.DiGraph()

# Create nodes representing choices
edges = [
    ('Choose 1 Item', 'Set A (3 options)'),
    ('Choose 1 Item', 'Set B (2 options)'),
    ('Set A (3 options)', 'Option A1'),
    ('Set A (3 options)', 'Option A2'),
    ('Set A (3 options)', 'Option A3'),
    ('Set B (2 options)', 'Option B1'),
    ('Set B (2 options)', 'Option B2')
]
G.add_edges_from(edges)

# Define a hierarchy layout manually
pos = {
    'Choose 1 Item': (0, 2),
    'Set A (3 options)': (-1.5, 1),
    'Set B (2 options)': (1.5, 1),
    'Option A1': (-2.5, 0),
    'Option A2': (-1.5, 0),
    'Option A3': (-0.5, 0),
    'Option B1': (1, 0),
    'Option B2': (2, 0)
}

nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=2000, ax=ax)
nx.draw_networkx_edges(G, pos, arrows=True, edge_color='gray', ax=ax)
nx.draw_networkx_labels(G, pos, font_size=9, font_family='sans-serif', ax=ax)

plt.title("Rule of Sum: Total Options = 3 + 2 = 5", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```

## 2. The Rule of Product (The "AND" Rule)

### Definition
If a procedure can be broken down into a sequence of $k$ steps, where the first step has $n_1$ outcomes, the second step has $n_2$ outcomes, and so on, then the total number of outcomes for the sequence is the product of the outcomes of each step.

### Mathematical Equation
$$\text{Total Ways} = |A \times B| = |A| \cdot |B| = \prod_{i=1}^{k} n_i$$

```python
# Rule of Product Diagram: Sequential Choices (Tree Diagram)
import matplotlib.pyplot as plt
import networkx as nx

fig, ax = plt.subplots(figsize=(8, 5))
G_prod = nx.DiGraph()

# Step 1 choices (2 options) -> Step 2 choices (3 options each)
prod_edges = [
    ('Start', 'Step 1: A1'), ('Start', 'Step 1: A2'),
    ('Step 1: A1', 'A1 -> B1'), ('Step 1: A1', 'A1 -> B2'), ('Step 1: A1', 'A1 -> B3'),
    ('Step 1: A2', 'A2 -> B1'), ('Step 1: A2', 'A2 -> B2'), ('Step 1: A2', 'A2 -> B3')
]
G_prod.add_edges_from(prod_edges)

pos_prod = {
    'Start': (0, 2),
    'Step 1: A1': (-2, 1),
    'Step 1: A2': (2, 1),
    'A1 -> B1': (-3, 0), 'A1 -> B2': (-2, 0), 'A1 -> B3': (-1, 0),
    'A2 -> B1': (1, 0), 'A2 -> B2': (2, 0), 'A2 -> B3': (3, 0)
}

nx.draw_networkx_nodes(G_prod, pos_prod, node_color='lightgreen', node_size=1800, ax=ax)
nx.draw_networkx_edges(G_prod, pos_prod, arrows=True, edge_color='gray', ax=ax)
nx.draw_networkx_labels(G_prod, pos_prod, font_size=8, font_family='sans-serif', ax=ax)

plt.title("Rule of Product: Total Outcomes = 2 × 3 = 6", fontsize=14, fontweight='bold')
plt.axis('off')
plt.show()
```
