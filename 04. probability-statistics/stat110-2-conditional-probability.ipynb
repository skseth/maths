# Conditional Probability

**Definition (Conditional Probability)**

If A and B are events in some sample space, with P(B) > 0, the conditional probability of A given B is:

$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$

In this context, $P(A)$ is called the **prior** probability of A (i.e. prior to **evidence** B), and $P(A|B)$ is called the **posterior** probability of A given B.

Note that $A|B$ is NOT an event. A is the event, and $A \cap B$ is an event.

It follows :

$P(A|A) = \frac{P(A \cap A)}{P(A)} = 1$

## Example (2-cards)

2 cards picked from a deck  
A = event first card is a heart  
B = event second card is red  

Note: Sample Space Size: 52.51

P(A) = 13.51/52.51 = 1/4  
P(B) = 26.(26 + 25)/52.51 = 26/52 = 1/2  

P(A \cap B) = 13.25 / 52.51 = 25/204  

P(A|B) = $\frac{25/204}{1/2} = \frac{50}{204} = \frac{25}{102}$

P(B|A) = $\frac{25/204}{1/4} = \frac{100}{204} = \frac{50}{102}$


## Points of Note

- Avoid the **prosecutor's fallacy** - $P(A|B) \ne P(B|A)$


- "Conditioning" is not causation - B does not cause A, but the fact that B has occured provides some more information on whether A may have occured.

- In a sense P(A|B) - amounts to rescaling probabilities as if B is the sample space, and then seeing what is the probability of A. This can be related to the **frequentist interpretation** - if we are doing repeated trials, we restrict our attention to when B occured, and then see if A occured.

# Bayes Rule and Law of Total Probability

## Theorem : (Intersection of Two Events):

$$P(A \cap B) = P(B)P(A|B) = P(A)P(B|A)$$

Proof: This follows immediately from the following two equations:

$P(A|B) = \frac{P(A \cap B)}{P(B)}$, $P(B|A) = \frac{P(B \cap A)}{P(A)} = \frac{P(A \cap B)}{P(A)}$

## Theorem : (Intersection of n Events):

$$P(A_1, A_2, \dots, A_n) = P(A_1)P(A_2|A_1)P(A_3|A_1A_2)\dots P(A_n|A_1,A_2,\dots,A_{n-1})$$

Here the commas refer to intersection. 

Proof: for n = 2, we have proved the result above. By induction we can prove the rest. Assume the result is true for some k (hypothesis).

$$P(A_1, A_2, \dots, A_k) = P(A_1)P(A_2|A_1)P(A_3|A_1A_2)\dots P(A_k|A_1,A_2,\dots,A_{k-1})$$

For k+1 (the induction step): 

$$
\begin{aligned}
P(A_1, A_2, \dots, A_k, A_{k+1}) &= P(A_1, A_2, \dots, A_k \cap A_{k+1}) \\
                                 &= P(A_1)P(A_2|A_1)P(A_3|A_1A_2)\dots P(A_k \cap A_{k+1}|A_1,A_2,\dots,A_{k-1}) \text{ by hypothesis} \\[1cm]
\text{ ... However, setting } C=A_1 \cap A_2 \cap \dots \cap A_{k-1} \text{ ,} \\                               
P(A_k,A_{k+1}|C) &= \frac{P(C,A_k,A_{k+1})}{P(C)} = \frac{P(C,A_{k})P(A_{k+1}|C,A_k)}{P(C)} = \frac{P(C)P(C|A_{k})P(A_{k+1}|C,A_k)}{P(C)} = P(A_k|C)P(A_{k+1}|C,A_k) \\
\text{ NOTE: We have used the n=2 formula twice in the expression above } \\[1cm]                               
\text{ ... Substituting back } \\                               
P(A_1, A_2, \dots, A_k, A_{k+1}) &= P(A_1)P(A_2|A_1)P(A_3|A_1A_2)\dots P(A_k|A_1,A_2,\dots,A_{k-1})P(A_{k+1}|A_1,A_2,\dots,A_{k-1},A_k)
\end{aligned}
$$





## Theorem: (Bayes' Rule):

$$P(A|B) = \frac{P(A)P(B|A)}{P(B)}$$

Think of it as : P(A|B) = P(B|A) rescaled by ratio of P(A) to P(B)

Proof: is trivial

$P(A \cap B) = P(B)P(A|B) = P(A)P(B|A) \implies P(A|B) = \frac{P(B|A)P(A)}{P(B)}$


**Definition (Odds)**

The odds of an event A are:

$$odds(A) = \frac{P(A)}{P(A^c)}$$

Note that $P(A) = odds(A).P(A^c) = odds(A)(1 - P(A))$, resulting in:

$$P(A) = \frac{odds(A)}{1 + odds(A)}$$



```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Generate an array of odds (e.g., from 0.05 to 10)
odds = np.linspace(0.05, 10, 200)

# 2. Convert odds to probability: P = Odds / (1 + Odds)
prob = odds / (1 + odds)

# 3. Create the plot
plt.figure(figsize=(8, 5))
plt.plot(odds, prob, color='blue', linewidth=2, label='Probability = Odds / (1 + Odds)')

# 4. Format the graph
plt.title('Odds to Probability Conversion', fontsize=14)
plt.xlabel('Odds', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.axhline(0.5, color='gray', linestyle='--', alpha=0.7) # 50% line
plt.axvline(1.0, color='gray', linestyle='--', alpha=0.7) # Even odds line
plt.grid(True, alpha=0.3)
plt.legend()

# 5. Render the plot in Jupyter
plt.show()
```

## Theorem: Bayes' Rule (Odds Form)

$$\frac{P(A|B)}{P(A^c|B)} = \frac{P(B|A)}{P(B|A^c)} \frac{P(A)}{P(A^c)}$$

$$\text{posterior odds} = \text{likelihood ratio} * \text{prior odds}$$


Posterior Odds of A (given B) = likelihood ratio of B given A * prior odds of A

**Side note on sensitivity, specificity, likelihood ratio**

Suppose $A$ = prevalence = probability of having a disease, $A^c$ = probability of not having it (before the test).

Let $B$ = test is positive. Now:

true positive rate = sensitivity = P(testing positive when you have the disease) = P(B|A) - 

true negative rate = specificity = P(testing negative when you do not have the disease = $P(B^c|A^c)$

false negative rate = 1 - specificity = $P(B|A^c)$

likelihood ratio = $\frac{\text{true positive}}{\text{false postive}} = \frac{P(B|A)}{P(B|A^c)}$


## Theorem: (Law of Total Probability - LOTP)

Assuming $A_1, A_2, \dots, A_n$ are a partition of the sample space, with $P(A_i) > 0$ for all i:

$$P(B) = \sum\limits_{i=1}^{n} P(B|A_i)P(A_i)$$

Visualize the A's breaking S into n partitions. We are then summing B across each of these partitions.

Proof:

$B = \bigcup_{i=1}^{n} B \cap A_i$ 

RHS is a union of disjoint sets, since the $A_i$ are disjoint i.e. sum of 

$P(B) = P(\bigcup_{i=1}^{n} B \cap A_i) = \sum_{i=1}^{n} P(B \cap A_i) = \sum_{i=1}^{n} P(B|A_i)P(A_i)$

# Conditional Probabilities are Probabilities

Let E be some event, P(E) > 0. $\hat{P} = P(\cdot|E)$ is then a function from subsets of E to $[0,1]_R$. 

**Axioms**
When we say P(A|E), A|E is not an event. Instead the event is $A \cap E$, and we calculate the scaled probability of $A \cap E$ by scaling the probability of P(E) to be 1, and calculating $P(A|E) = \frac{P(A \cap E)}{P(E)}$. Thus:

A) $\hat{P}(\emptyset) = P(\emptyset|E) = \frac{P(\emptyset \cap E)}{P(E)} = \frac{P(\emptyset)}{P(E)} = 0$

B) $\hat{P}(S) = P(S|E) = \frac{P(S \cap E)}{P(E)} = \frac{P(E)}{P(E)} = 1$

C) If $A_1, A_2, \dots, A_n$ are disjoint, then 

$\hat{P}(\bigcup\limits_{i=1}^{n} A_i) = P(\bigcup\limits_{i=1}^{n} A_i|E) = \sum\limits_{i=1}^{n} P(A_i|E) = \sum\limits_{i=1}^{n} \hat{P}(A_i)$


A,B,C shows that $\hat{P}$ meets the axioms of probability, which means all other results we have made follow.


**Complements**
- $P(A^c|E) = 1 - P(A|E)$

**Inclusion-Exclusion**

$P(A \cup B|E) = P(A|E) + P(B|E) - P(A \cap B|E)$

**Extra Conditioning**

Note that:

$\hat{P}(A|B) = \frac{\hat{P}(A \cap B)}{\hat{P}(B)} = \frac{P(A \cap B \cap E)}{P(B \cap E)} = P(A|B,E)$


$P(A|B,E) = \frac{P(A \cap B \cap E)}{P(B \cap E)} = \frac{P(A \cap B|E)P(E)}{P(B|E)P(E)} = \frac{P(A \cap B|E)}{P(B|E)}$

What this means is that with this new conditional probability "measure" $\hat{P}$ further conditions work as expected. To translate them into the original measure P, simply treat them as multiple conditions.


**Theorem : (Bayes' Rule w/ Conditioning)**

If A,B,E are events, P(E) > 0,

$\displaystyle P(A|B,E) = \frac{P(A|E)P(B|A,E)}{P(B,E)}$

**Theorem: (LOTP w/ Conditioning)**

E is an event in a sample space S, P(E) > 0  
$A_1, A_2, \dots, A_n$ are partitions of S. 

It follows that:
$\displaystyle \hat{P}(B) = \sum\limits_{i}^{n} \hat{P}(B|A_i)\hat{P}(A_i)$

By definition it follows:

$\displaystyle P(B|E) = \sum\limits_{i}^{n} P(B|A_i,E)P(A_i|E)$

Actually we can also work with partitions of E, since we can take $A_{n+1} = E^c$ as the final partition. Now $P(E^c|E) = 0$, so the result above still stands.

# Independence

**Definition: Independence of Events**

Events A and B are independent if :

$P(A \cap B) = P(A)P(B)$  

$P(A|B) = P(A)$ - assuming P(B) > 0, P(A) > 0 

Relation is symmetric : If $P(A|B) = P(A)$, $P(B|A) = P(B)$

Proof: $P(A|B)P(B) = P(B|A)P(A) \implies P(A)P(B) = P(A)P(B|A) \implies P(B|A) = P(B)$

NOTE: P(A) > 0 assumption is needed only for applying symmetry. Otherwise only P(B) > 0.

NOTE: INDEPENDENCE $\ne$ DISJOINTNESS: A common misconception is to think of independence in terms of sets - two sets being "independent" in the sense of having no common elements. But in probability it is almost the opposite. If two events A and B are disjoint, the P(B|A) = 0 - if A has happened, B could not have happened. In independence, outcomes favourable to B appear in A in the same ratio as in the overall sample space. Thus P(B|A) = P(B).

## Proposition: If A and B are independent, A and $B^c$ are independent, $A^c$ and B are independent, and $A^c$ and $B^C$ are independent.

Proof: Assume P(A), P(B) > 0. 

$P(A|B^c) = \frac{P(B^c|A)P(A)}{P(B^c)} = \frac{(1-P(B|A))P(A)}{1-P(B)} = \frac{(1-P(B))P(A)}{1-P(B)} = P(A)$. $A \text{ and } B^c$ are independent.

$P(B^c|A) = 1 - P(B|A) = 1 - P(B) = P(B^c)$. Thus $B^C$ and $A$ are independent.

Reversing the roles of A and B, we get the other results e.g.

$P(A^c|B^c) = 1 - P(A|B^c) = 1 - P(A) = P(A^c)$, so $A^c$ and $B^c$ are independent.

$P(A^c|B) = 1 - P(A|B) = 1 - P(A) = P(A^c)$ - so $A^c$ and $B$ are independent.

If P(A) = 0, then obviously P(A|B^c) = P(A) = 0. And P(A^c) = 1. So P(B^c|A^c) = P(B^c). P(B|A^c) = P(B). But note that P(B|A) is not P(B), since no event is independent of the empty set, while the empty set is independent of every event.

**Independence of Three events**

Three events A,B,C are independent iff A,B,C are pairwise independent, AND $P(A \cap B \cap C) = P(A)P(B)P(C)$

Example (Pairwise independence != independence):

Consider 2 independent fair, independent tosses. A = first is H, B = second is H, C = both first and second are same

S = {HH, TH, HT, TT}

A = {HH, HT}

B = {HH, TH}

C = {HH, TT}

All the intersections have probability 1/4 (including P(A n B n C)), but P(A)P(B)P(C) = 1/8. Why? Because knowing A and B tells us all about C, even though individually A and C, and B and C are independent.

**Definition: Independence of Many, Even Infinite Events**  

n events $A_1, A_2, \dots, A_n$, are said to be independent if all combinations of subsets of size k = 2 to n are independent.

For infinite events, we say that they are independent if any finite combination of subsets is independent.

**Definition: Conditional Independence**  

$P(A \cap B|E) = P(A|E).P(B|E)$

NOTE: There is no connection between A and B being independent, and being conditionally independent based on some event E. There is no connection between being conditionally independent based on event E, and being conditionally independent based on $E^c$. However, within the context of a condition E, the earlier independence propositions hold. e.g.

If $P(A|B,E) = P(A|E)$, then $P(A^c|B,E) = P(A^c|E)$ etc.

## Ex (Conditional Independence doesn't imply independence)

Consider the case of one fair, one biased coin. 

F = event we chose a fair coin  
$A_1$ = event first toss is heads  
$A_2$ = event second toss is heads  

$S = \{F,U\} \times \{HH,HT,TH,TT\}$

$A_1 = \{F,U\} \times \{HH,HT\}$

$A_2 = \{F,U\} \times \{HH,TH\}$

$A_1 \cap A_2 = \{F,U\} \times \{HH\}$

Note that F & U options are not equally likely.

Then $P(A_2|A_1,F) = P(A_2|F)= 1/2$, $P(A_2|A_1,F^c) = P(A_2|F^c)= 3/4$, but $P(A_2|A_1) \ne P(A_2)$

$A_1, A_2$ are conditionally independent given either F or $F^c$. But they are not unconditionally independent, because $A_1$ can provide some information whether F has occured or not, which impacts the probability of $A_2$.

## Ex (Independence does not imply conditional independence)

A and B are independent.

Then P(B|C) > 0, P(A|C) > 0, but P(B|A,C) = 0, and P(A|B,C) = 0. Thus A and B are not conditionally independent given C.

# Coherency of Bayes' Rule

If we have multiple pieces of information to update our priors, it does not matter whether we do it one by one or together.

## Ex (Testing for a rare disease, continued)

$P(D|T_1) = 0.19/1+0.19 = 0.16$

Using the odds form:

$\frac{P(D|T_1 \cap T_2)}{P(D^c|T_1 \cap T_2)} = \frac{P(D)}{P(D^c)}\frac{P(T_1 \cap T_2|D)}{P(T_1 \cap T_2|D^c)} = \frac{0.01}{.99}(\frac{0.95}{0.05})^2 = 361/99 = 3.646$

$P(D|T_1 \cap T_2) = 3.646/(1 + 3.646) = 78\%$


# Conditioning as Problem-Solving Tool

## (Monty Hall)

3 doors hiding 2 goats, one car. You guess one door. Host opens one of the other doors with no goat. Q. So you switch?

Important point:
- You want a car  
- Host always opens door with a goat, and with equal probability when he has a choice  

To simply we call the first selected door 1, the other two as 2/3.  

$C_i$ = car is behind door i, i = 1,2,3. $P(C_i) = 1/3$. S = event that i switch.  
$W$ = Probability of winning a car by unconditionally switching after host selects a door.

$P(W) = P(W_u|C_1)P(C_1) + P(W|C_2)P(C_2) + P(W|C_3)P(C_3) = 0.1/3 + 1.1/3 + 1.1/3 = 2/3$.

Let $M_j$ be event that host opens door j, j = 2 or 3. 

$P(W) = P(W|M_1)P(M_1) + P(W|M_2)P(M_2)$

By symmetry: $P(M_2) = P(M_3) = 1/2$, and $P(W|M_1) = P(W|M_2) = x$, say. And we know P(W) = 2/3

$2/3 = x/2 + x/2 = x$. Thus $P(W|M_1) = P(W|M_2) = P(W)$

## (Branching Strategy)

Amoeba splits every minute into 2 amoebas, or dies, or stays the same, all 1/3 probability. What is P(amoeba population will die out)

Solution:

B_i = event that at end of first minute you have i=0,1,2 amoebas. P(B_i) = 1/3. D be event population dies out (eventually)

$P(D) = P(D|B_0)P(B_0) + P(D|B_1)P(B_1) + P(D|B_2)P(B_2)$ - LOTP

$P(D|B_0) = 1, P(D|B_1) = P(D), P(D|B_2) = P(D)^2$. The last case happens because now we have 2 amoebas, and each one has to die out.

## (Gambler's Ruin)

2 gamblers make a series of random bets. Gambler A starts with money i, and has probability p of winning each bet. B has N-i dollars and probability q = 1-p of winning each bet. This is a random walk problem. Game ends when random walk ends i.e. Game reaches 0 or N.

A_1 = gambler 1 wins  
A_2 = gambler 2 wins  

$p_i$ = probability A wins game, given A starts with i dollars, W = event A wins the game.

$p_i = P(W|\text{A starts round at i, wins round i}).p + P(W|\text{A starts at round i, loses round i}).q = p_{i+1}.p + p_{i-1}.q$

The solution is shown in the next sections.


Summarizing:

$$
\begin{align*}

p_i &= 
    \begin{cases}
        \frac{1-(q/p)^i}{1-(q/p)^N} &\text{ if q <> p} \\
        \frac{i}{N} &\text{ if q = p}
    \end{cases}
\end{align*}

$$

By a similar logic, the probability of B starting from N-i is :

$$
\begin{align*}

p'_{N-i} &= 
    \begin{cases}
        \frac{1-(p/q)^{N-i}}{1-(p/q)^N} = \frac{q^{N-i}-p^{N-i}}{q^N-p^N} &\text{ if q <> p} \\
        \frac{N-i}{N} &\text{ if q = p}
    \end{cases} \\

\text{Adding up, we see :} \\
p'_{N-i} + p_i &= 
    \begin{cases}
        \frac{1-(q/p)^i}{1-(q/p)^N} + \frac{1-(p/q)^{N-i}}{1-(p/q)^N} &\text{ if q <> p} \\
        \frac{N-i}{N} &\text{ if q = p}
    \end{cases} \\


\end{align*}

$$



In brief, if p < q, or p = q, but the B has more much more money, A is likely to lose - in any case the chances of the game going on forever is 0.


### Proof using Homogenous, Second-order Linear Difference Equation
See: [Introduction to Difference Equations - MIT](https://courses.csail.mit.edu/6.01/fall07/lectures/lecture4.pdf) - esp the section on second order difference equations.


Given a sequence, a homogenous second-order linear difference equation depends on two other values of the sequence. In our case (Gambler's ruin), we have:

$p_i = p.p_{i+1} + q.p_{i-1}$

In general, if we set this to $p_i = cx^i$, we get the characteristic equation:

$p.x^2 - x + q = 0$.

Solving we get possible values for the x:

$x = \frac{1 \pm \sqrt{1 - 4qp}}{2p} = \frac{1 \pm \sqrt{1 - 4(1-p)p}}{2p} = \frac{1 \pm \sqrt{4p^2-4p + 1}}{2p} = \frac{1 \pm \sqrt{(2p-1)^2}}{2p}$.

Thus: x = 1 + 2p - 1/2p = 1, or x = 1 - 2p + 1/2p = q/p. If p = 1/2, x = 1 always.


When roots are different:

$p_i = a + b(q/p)^i$

Since $p_0 = 0$, we get $a = -b$.

Since $p_N = 1 = -b + b.(q/p)^N \implies a = -b = \frac{1}{1 - (q/p)^N}$

Substituting, we get:

$p_i = \frac{1}{1 - (q/p)^N} - \frac{1}{1 - (q/p)^N} (q/p)^i = \frac{1 - (q/p)^i}{1 - (q/p)^N}$

For the case where roots are same, then $p_i = a + ib$ (see difference equation link)

$p_0 = 0 \implies a = 0$  
$p_N = 1 \implies b = 1/N$  
Thus, $p_i = i/N$  

Summarizing:

$$
\begin{aligned}
p_i &= 
    \begin{cases}
        \frac{1-(q/p)^i}{1-(q/p)^N} &\text{ if q <> p} \\
        \frac{i}{N} &\text{ if q = p}
    \end{cases}
\end{aligned}
$$

Reference:

- https://www.cimt.org.uk/projects/mepres/alevel/discrete_ch14.pdf
- https://www.cimt.org.uk/projects/mepres/alevel/discrete_ch15.pdf



### Proof using Alternate Way (Using sums)

$p_i = p.p_{i+1} + q.p_{i-1}$

We rewrite as : $p.p_i + q.p_i = p.p_{i+1} + q.p_{i-1}$, since p + q = 1. Rearranging, we get:

$p_{i+1} - p_{i} = q/p(p_i - p_{i-1})$

Thus, $p_2 - p_1 = q/p(p_1 - p_0) = q/p(p_1), p_3 - p_2 = (q/p)^2p_1$, and more generally:

$p_{i+1} - p_i = (q/p)^i p_1$. Thus,

$p_{i+1} - p_1 = \sum\limits_{k=1}^{i} (q/p)^k.p_1$

$p_{i+1} = p_1 + \sum\limits_{k=1}^{i} (q/p)^k.p_1 = p_1\sum\limits_{k=0}^{i} (q/p)^k$

Given the sum of geometric series: $\sum_{n=0}^{i} a^i = \frac{1 - a^{i+1}}{1 - a}$

For the case where root is same, we see that q = p. Thus:

$p_{i+1} - p_i = p_{i} - p_{i-1}$ - we have a constant difference between terms.




$$
\begin{aligned}
p_{i+1} &= 
    \begin{cases}
    p_1.\frac{1-(q/p)^{i+1}}{1-q/p}  &\text{ if q <> p} \\
    p_1.(i + 1) &\text{ if q = p} 
    \end{cases} \\
\text{But $p_N = 1$, so:} \\
p_1 &= 
    \begin{cases}
        \frac{1-q/p}{1-(q/p)^N} &\text{ if q <> p} \\
        \frac{1}{N} &\text{ if q = p}
    \end{cases} \\
\text{ and finally } \\
p_i &= 
    \begin{cases}
        \frac{1-(q/p)^i}{1-(q/p)^N} &\text{ if q <> p} \\
        \frac{i}{N} &\text{ if q = p}
    \end{cases}  \\
\end{aligned} 
$$

# Pitfalls and Paradoxes

## (Prosecutor's Fallacy - Sally Clark Case)

2 children died of SID. P(Child dying of SID) = 1/8500. 

Prosecutor's argument: $(\frac{1}{8500})^2 = 1/(73 \text{ million})$. Probability of innocence was one in 73 million. The claim was that the probability of innocence, given the evidence was too small.

Fallacy: a) Simply multiplying together the probabilities is wrong - this only works if the two events are independent, which may not be the case when there might be underlying genetic or other causes.

b) The prosecutor was calculating P(evidence|innocence). What we need is P(innocence|evidence).

$P(I|E) = \frac{P(E|I)P(I)}{P(E)} = \frac{P(E|I)P(I)}{P(E|I)P(I) + P(E|Guilt)P(Guilt)} \approx \frac{P(E|I)P(I)}{P(E|I)P(I)} = 1$. 

P(Guilt) - the prior probability of guilt is so low compared to P(Innocence), that it is neglible, and P(I|E) is almost 1. 

In other words, the probability of 2 children in the same family dying of SIDS is far more likely than the probability of any mother killing two of her children in the first place. 

See: 
- [Simple Bayesian Problems: The Sally Clark Case](https://www.mcgrayne.com/disc.htm)
- [Beyond Reasonable Doubt](https://plus.maths.org/beyond-reasonable-doubt)



## (Defence Attorney's Fallacy)

Woman murdered, with evidence of abuse. Defence argument - this is irrelevant, as only 1/10000 men who abuse actually kill their wives. Should judge grant the defence attorney's motion to exclude evidence? Also given: 1 in 10 husbands abuse their wives. 1 in 5 wives murdered, are murdered by their husbands, and 50% of husbands who murdered their wives, also abused them. How should the prosecutor argue?

A - wife is abused
M - wife is murdered

Defence argues that P(G|A) = 1/10000.

But what matters is P(G|A,M)

$$P(G|A,M) = \frac{P(G|M)P(A|G,M)}{P(A|M)} =  \frac{P(A|G,M)P(G|M)}{P(A|G,M)P(G|M) + P(A|G^c,M)P(G^c|M)}$$

P(A|G,M) = 1/2

P(A) = 1/10

$P(A|G^c,M) = P(A) = \frac{1}{10}$

P(G|M) = 1/5

P(G|M) = 0.2, P(G|A,M) = 0.56 - so abuse is very relevant to the case. P(G|A) was not used, because it is not conditioned on M.

## (Simpson's Paradox)

Dr A

|     | Heart | Band-Aid |
| --- | --- | --- |
| Success | 70 | 10 |
| Failure | 20 | 0  |


Dr B
|     | Heart | Band-Aid |
| --- | --- | --- |
| Success | 2 | 81 |
| Failure | 8 | 9 |

Dr A is better than Dr B at both Heart surgery and band-aids, yet his overall success rate (80/20), is worse than Dr B(83/17). This is the counter-intuitive result which shows up in many scenarios. The reason is simple - Dr A is performing the harder type of surgery more often than Dr B is. We should be wary of aggregate comparisons like this.

The paradox:

For events S - success, A - Dr. A, H - Heart surgery:

$P(S|A,H) > P(S|A^c,H)$  
$P(S|A,H^c) > P(S|A^c,H^c)$  


But:

$P(S|A) < P(S|A^c)$  

The reason this happens is because Dr A is doing the more difficult surgery more often. Let's see LOTP:

$P(S|A) = P(S|A,H)P(H|A) + P(S|A,H^c)P(H^c|A) = (7/9)(9/10) + 1.(1/10)$

$P(S|A^c) = P(S|A^c,H)P(H|A^c) + P(S|A^c,H^c)P(H^c|A^c) = (2/10)(1/10) + (.9)(9/10)$

A **confounding variable** like surgery type is what explains the apparent paradox.

**28) (Disease Testing - PPV): For a rare disease and a reasonably good test, specificity matters more than sensitivity**  


(a) Express posterior odds in terms of specificity and sensitivity.

$\text{posterior odds(have disease | positive test)} = \dfrac{\text{sensitivity}}{1 - \text{specificity}} * \text{prevalence(prior odd of disease)}$

The **positive predictive value** is the probability $P(\text{have disease | positive test})$. For a rare disease and a reasonably good test, specificity matters much more than sensitivity, because even with a high specificity, the large pool ensures false positives may outnumber true positives. 

Let p = prior odds of having disease, q = 1-p, D - have disease. T - positive test

$$\boxed{odds(D|T) = \dfrac{p*sensitivity}{q(1 - specificity)} = \dfrac{sensitivity}{\frac{q}{p}(1 - specificity)}}$$

$$\boxed{PPV = P(D|T) = \dfrac{sensitivity}{sensitivity + \frac{q}{p}(1 - specificity)}}$$

Suppose sensitivity = specificity = 0.95, and q/p = 99. Then the denominator is $0.95 + 99(0.05) = 0.95 + 4.95 = 5.9$.

Suppose sensitivity = specificity = 0.99, and q/p = 99. Then the denominator is $0.99 + 99(0.01) = 0.99 + 0.99 = 1.98$.

Assume disease affects 1% of people, with a population of 1000. Take two tests A and B, with sensitivity 90% and 95%, but specificity reversed.

</div>


b) Problem
Suppose 100 people are tested, and within this 100, P(D=disease) = 5%.

Let A = event test A is positive, let B = event test B is positive.

$P(A|D) = 0.9, p(B|D) = 0.95, P(A^c|D^c) = 0.95, P(B^c|D^c) = 0.9,P(D) = 0.05, P(D^c) = 0.95$

Given 100 people are tested, how many false positives and false negatives can we expect?

P(D^c|A) = P(D^c)P(A|D^c)/P(A|D)P(D)  + P(A|D^c)P(D^c) = 0.95*0.05 / (0.9*0.05 + 0.95*0.05) = 0.95/1.85 = 19/37 = 51%

P(D^c|B) = P(D^c)P(B|D^c)/P(B|D)P(D)  + P(B|D^c)P(D^c) = 0.95*0.10 / (0.95*0.05 + 0.95*0.10) = 0.1/0.15 = 2/3 = 24/36 = 66%


P(D|A^c) = P(D)P(A^c|D)/P(A^c|D)P(D) + P(A^c|D^c)P(D^c) = 0.05*0.1/(0.05*0.1 + 0.95*0.95) = 1/181.5 = 0.55%

Test A will report: around 48.5 people false positive, almost 0.1275 people as false negative

```python
def false_positives(sensitivity: float, specificity: float, prevalence: float):
    nonprevalence = 1 - prevalence
    nonspecificity = 1 - specificity
    return (nonprevalence*nonspecificity)/(nonprevalence*nonspecificity + prevalence*sensitivity)

def false_negatives(sensitivity: float, specificity: float, prevalence: float):
    nonprevalence = 1 - prevalence
    nonsensitivity = 1 - sensitivity
    return (prevalence*nonsensitivity)/(prevalence*nonsensitivity + nonprevalence*specificity)

print(f"A: False positives - {round(false_positives(0.9,0.95,0.05)*100,2)}")
print(f"B: False positives - {round(false_positives(0.95,0.9,0.05)*100,2)}")
print(f"A: False negatives - {round(false_negatives(0.9,0.95,0.05)*100,2)}")
print(f"B: False negatives - {round(false_negatives(0.95,0.9,0.05)*100,2)}")
```

In solution:

$P(A) = P(A|C)P(C) + P(A|C^c)P(C^c) < P(B|C)P(C) + P(B|C^c)P(C^C) = P(B)$

b) Simpson's Paradox:

P(A|B,C) > P(A|B^c,C)  
P(A|B,C^c) > P(A|B^c,C^c)  

But

P(A|B) < P(A|B^c)

Unlike the case in (a), the weights $P(C|B)$ and $P(C|B^c)$, as well as $P(C^c|B)$ and $P(C^c|B^c)$ are different, which is why the equality can flip.

(b) Let $C_j$ be the event that the first Ace is at position j. 

$P(B|C_j) = \frac{3}{52-j}$

$P(C_j) = \frac{\binom{48}{j-1}}{\binom{52}{j-1}}.\frac{4}{52-j+1}$

$P(B) = \sum\limits_{j=1}^{49} P(B|C_j)P(C_j) = \sum\limits_{j=1}^{49} \frac{\binom{48}{j-1}}{\binom{52}{j-1}}.\frac{4}{52-j+1}\frac{3}{52-j}$

(d) Partly From solution:

Let L = probability an ace is the last card. We can easily see P(L) = 1/13.

But suppose we ask what is the probability that the last card is an ace, given the first ace is at the jth position? And we find $P(L|C_j) = P(B|C_j)$!. But by LOTP,

$P(L) = \sum_{j=1 \to 49} P(L|C_j)P(C_j) = \sum_{j=1 \to 49} P(B|C_j)P(C_j) = P(B) = 1/13$

Why are the two cases similar? Suppose we had said: After the first ace is found, we start counting backwards from the bottom. What is the probability the last card is an ace? After the jth card, all permutations of the remaining 52-j cards is equally likely, so swapping the 1st and the last card, or even taking the reverse permutation so the first card becomes the last - should be the same. 

A more complicated way to phrase the problem would be : D = event that the middle card in the remaining pack (calculated as $\lfloor \frac{52-j}{2} \rfloor$), is an ace? This removes the mysterious requirement that the "next" card be an ace. We find that $P(D|C_j) = \frac{3}{52-j}$, 

Or more simply E = pick a random card from the remaining 52-j cards. What is the probability it is an ace? Again - $P(E|C_j) = \frac{3}{52-j}$. And then we can apply the same argument as the last card argument. 

There is nothing special per-se about the last card. We can ask - what is the probability that the "nth" card is an ace? Let's call this event N. Then P(N) = 1/13 (obviously). The issue is just that as stated, it is harder to frame the problem properly. So we can frame this as : Count till the first ace, skipping position "N", if n is before the first ace. Let $C_j$ = event that the jth open card is an ace - j still ranges from 1 to 49. Now what is the probability the nth card is an ace (N)? 3/52-j!. Using LOTP, we see that  P(N) = 1/13 = $\sum\limits_{j=1}^{49} P(N|C_j)P(C_j)$ = P(B).
