## Fallacy - Reasoning on the Mean

How many throws of a fair die do we need to get a fair (>= 50% chance) of getting a six.

A **fallacious** argument says: each throw has a 1/6 chance of getting one six. So the probability of getting at least one six in 3 tries is np = 3.(1/6) = 1/2. 

In reality, $1-(5/6)^3 = 0.421$ is the probability of getting at least one six in 3 tries. Probability of getting exactly one 6 is even less: $((25*3)/6^3) = 0.347$







```python

x1 = round(((25*3)/(6.0 ** 3.0)), 6)
x2 = round(((5*3)/(6.0 ** 3.0)), 6)
x3 = round(((1)/(6.0 ** 3.0)), 6)
P = x1 + x2 + x3
E = x1 + 2*x2 + 3*x3

print(x1, x2, x3, P, E)
```

As can be seen above - the fallacy is based on confusing the expected number of sixes (which is 1/2), and the expected number of groups of 3 tries containing at a six. The error comes because we are counting the number of sixes, and not the number of tries of 3 which contain sixes.

To quote: "If an event has a probability p in one trial of an experiment, then in n trials, the event will occur on **np** times on average, which is then wrongly taken to represent the **probability** that the event will occur at least once in n trials. 

# Fallacy - Old Gambler's Rule

The old gambler's rule says that if there are two games, where probability of winning is $p_1$ and $p_2$ respectively, and with "critical values" of $C_1$ and $C_2$ respectively, such that 

$$C_1p_1 = C_2p_2$$

then 

$$C_2 = C_1\frac{p_1}{p_2}$$

One key problem with the reasoning is that this does not take into account the fact that "C_1" could be fractional, leading to errors.


A better way to do this:

Let x be such that $(1-p)^x = 0.5$, where x is any real number. $1-p$ is the probability of the event not happening. 

Then $x = \frac{-ln(2)}{ln(1-p)}$, and the number of trials needed $C_1 = \lceil x \rceil$


```python
import math

p_double_six = round(-math.log(2)/math.log(35/36), 5)

print(p_double_six, math.ceil(p_double_six))

```
