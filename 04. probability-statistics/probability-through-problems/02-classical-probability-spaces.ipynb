**2.13** lottery : t tickets, w winning, i buy b tickets. What is the probably g tickets (out of b) win?

Total ways to select b from t is : $\binom{t}{b}$

There are t-w losing tickets and w winning tickets. I want to select g from w, and b-g from (t-w) = $\binom{t-w}{b-g}\binom{w}{g}$, 

P(g wins out of b) = $\frac{\binom{t-w}{b-g}\binom{w}{g}}{\binom{t}{b}}$

**2.14** 1000 light bulbs, 2% are defective. What is the probability that out of 20, 2 (i.e. 10%) are defective.

#S = $\binom{1000}{20}$

#E = $\binom{20}{2}\binom{980}{18}$

```python
from math import comb

print((comb(20,2)*comb(980,18))/comb(1000,20))
```

**2.16** Bridge. I have 4 spades in one hand, 8 hearts in opponents. Probability my partner has at least 3 spades?

We know our 13 cards + 8 hearts - leaving 31 cards to select from include 9 spades. 

#S = $\binom{31}{13}$

#not-E = $\binom{22}{13} + \binom{22}{12}\binom{9}{1} + \binom{22}{11}\binom{9}{2}$ 


```python
from math import comb

print(1-(comb(22,13) + (comb(22,12)*comb(9,1)) + (comb(22,11)*comb(9,2)))/comb(31,13))
```
