**1.3** A die is rolled. What about 2 dice?

S = {1,2,3,4,5,6}, P({w}) = 1/6

For 2 dice, we can take the ordered pairs : S = {(1,1),(1,2)...(6,6)}, with P({w}) = 1/36.

Or we can take the sum: S = {2,...,12}, where probabilities can we obtained by counting.


```python
import numpy as np

# 1. Create the single die array
die = np.array([1, 2, 3, 4, 5, 6])

# 2. Use broadcasting to sum all 36 pairs (creates a 6x6 grid)
sums_grid = die[:, None] + die[None, :]

# 3. Find the unique sums and count their exact frequencies
unique_sums, frequencies = np.unique(sums_grid, return_counts=True)
probabilities = frequencies / sums_grid.size

# 4. Print the exact counts
print("Sums:       ", unique_sums)
print("Frequencies:", frequencies)
print("Probabilities:", np.round(probabilities, 4))
```

**1.6** Draw 2 cards from 52

If sampling with replacement (draw one, replace, draw two), we have |S| = 52x52 - all ordered pairs of $C \times C$, where C is the deck of cards.

If we are sampling without replacement (two cards at once), |S| = all subsets of 2 cards from 52: $\binom{52}{2}$



```python
import random

suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']
ranks = ['2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']

# 1. Create a card and deck representation
deck = [{'rank': r, 'suit': s} for s in suits for r in ranks]


# 3. You can use SymPy combinatorics if you need to calculate 
#    complex poker hand permutations (e.g., "52 choose 5")
from sympy.utilities.iterables import multiset_permutations, combinations
total_combinations = len(list(combinations(deck, 2)))

print(f"Total possible 2-card hands: {total_combinations}")

# 2. Draw a hand (e.g., 5 cards)
hand = random.sample(deck, 5)
print(hand)
```
