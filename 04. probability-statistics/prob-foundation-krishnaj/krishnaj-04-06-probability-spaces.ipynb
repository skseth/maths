**Structure of Events**

An collection $\mathscr{F}_0$ of subsets of $\Omega$ is an algebra if 

(i) $\emptyset$ is in $\mathscr{F}_0$

(ii) If $A \in \mathscr{F}_0$, then $A^c \in \mathscr{F}_0$

(iii) If $A \in \mathscr{F}_0 \text{ and } B \in \mathscr{F}_0$, then $A \cup B \in \mathscr{F}_0$

By induction, we can prove that, given a collection of n elements $A_1, A_2, \dots, A_n$ of $\mathscr{F}_0$,  $n \in \mathscr{N}$,

$\bigcup\limits_{i=1}^{n} A_i \in \mathscr{F}_0$

Note that the union is not a limiting operation in any sense - it is simply a set which contains all $\omega$ which belong to at least one of the $A_i$.

i.e. a $\mathscr{F}_0$ is closed under union of a finite number of its elements.

Similary, we can show that 

$\bigcap\limits_{i=1}^{n} A_i \in \mathscr{F}_0$. 

To prove this, note that applying De Morgan's law, and the previous identity, we get:

$\left( \bigcup\limits_{i=1}^{n} A_i^c \right)^c = \bigcap\limits_{i=1}^{n} A_i$

Since each of $A_i^c$ are $\mathscr{F}_0$ (by (ii), it follows that their finite union $\bigcup\limits_{i=1}^{n} A_i^c$ is also in $\mathscr{F}_0$. And taking the complement (which must also be in the collection by (ii)), we get the required result.

A $\sigma$-algebra : An algebra closed under countably infinite union of its elements.

By de Morgan's laws it follows immediately that it's closed under infinite intersection.

Note that the algebra may contain uncountably infinite elements. It just needs to be closed under countably infinite union.

**Obviously a sigma-algebra is closed under finite unions i.e. it is an algebra** : For any finite collection of n sets $A_1, A_2, \dots, A_n$, extend it to be infinite by adding the empty set, and you can then show that the sigma-algebra is closed under finite union.



Homework: Show every algebra is not a sigma-algebra

$\Omega$ is a sample space. $\mathscr{F}$ is a $\sigma$-algebra of subsets of $\Omega$. Subsets of $\Omega$ which belong to $\mathscr{F}$ are called **$\mathscr{F}-\mathbf{measurable}$ sets**, and $(\Omega,\mathscr{F})$ is called a **measurable space**.


**Some possible sigma-algebras on $\Omega$**

The smallest sigma algebra = $\{\emptyset, \Omega \}$

Smallest Sigma algebra with event A = $\{\emptyset, A, A^c, \Omega \}$

$2^{\Omega}$ is the largest sigma-algebra

For finite and countably infinite sample spaces, we can actually use $2^{\Omega}$ as a sigma-algebra, and assign probabilities to every subset of $\Omega$.

The problem is for uncountably infinite sample spaces, this fails i.e. $2^{\Omega}$ is too large in some sense.


## Measure

Probability is a special case of a more general concept of measures.

Given a measurable space, $(\Omega,\mathscr{F})$, a **measure** is a function $\mu: \mathscr{F} \to [0, \infty]$ such that :

(i) $\mu(\emptyset) = 0$

(ii) (countable additivity) : Given a countable collection $A_1, A_2, \dots$ of **disjoint** $\mathscr{F}$-measurable sets, then:

$$\mu({\bigcup\limits_{i=1}^{\infty} A_i}) = \sum\limits_{i=1}^{\infty} \mu(A_i)$$

## Measure Space

A measurable space  $(\Omega,\mathscr{F})$ along with a measure $\mu$, is a measure space (ugh!), $(\Omega,\mathscr{F}, \mu)$

If $\mu(\Omega) < \infty$, it's called a **finite measure**, else it is called an **infinite measure**.

If $\mu(\Omega) = 1$ is called a **probability measure**.


## Probability Measure Spaces

A probability measure $\mathscr{P}$ on $(\Omega,\mathscr{F})$, a **measure** is a function $\mu: \mathscr{F} \to [0, 1]$ such that :

(i) $\mathscr{P}(\emptyset) = 0$

(ii) $\mathscr{P}(\Omega) = 1$

(iii) (countable additivity) : Given a countable collection $A_1, A_2, \dots$ of **disjoint** $\mathscr{F}$-measurable sets, then:

$$\mathscr{P}({\bigcup\limits_{i=1}^{\infty} A_i}) = \sum\limits_{i=1}^{\infty} \mathscr{P}(A_i)$$

**NOTE** The left side is a simple expression - it is always well defined. But the summation on the right will be a limit of some finite sums.

We call $(\Omega, \mathscr{F}, \mathscr{P})$ is a probability space.

And finally we have a definition of event! Any element of $\mathscr{F}$ is an event. i.e. the $\mathscr{F}$-measurable sets of the probability space $(\Omega, \mathscr{F}, \mathscr{P})$ are events.

# Lecture 6 - Properties of Probability measure

(i) **Finite Additivity**

Given a finite collection of disjoints sets in $\mathscr{F}$, the sum of their probabilities is the probability of their union.

(ii) $P(A^c) = 1 - P(A)$

This follows from finite additivity $(A \text{ and } A^c)$ are disjoint, their union is $\Omega$, and $P(\Omega) = 1$.

(iii) Monotonicity: If A,B are in $\mathscr{F}$, and $A \subset B$, then:

$P(A) \le P(B)$

Proof: $B = A \cup (B \setminus A)$. 

(iv) Probability of a union 

$P(A \cup B) = P(A) + P(B) - P(A \cap B)$

(v) Inclusion-Exclusion Principle for Probabilities

Prove by induction. Later proved by Indicator Random Variables.
