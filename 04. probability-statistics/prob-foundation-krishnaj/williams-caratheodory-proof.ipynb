Sigma-Algebra: Algebra, but "stable" under countable union


## Borel Sets

Let S be a topological space.

$\mathscr{B}(S) = \sigma$(open sets)

As a shorthand, we write : $\mathscr{B} = \mathscr{B}(\mathscr{R})$

## $\pi$-systems

Sigma-algebras are hard to work with, but $\pi$-systems are easy.

A collection of subsets is a **pi-system** if it is closed under finite intersections. This implies it is also closed under finite unions.


**$\pi(R)$**

$\pi(R) = \{(-\infty, x]: x \in R \}$ is a **pi-system**.

We can see that given any elements $A = (-\infty, a])$ and $B = (-\infty, b])$ of $\pi(R)$, $A \cap B$ = either A or B, so $\pi(R)$ is closed under finite intersection.

**$B = \sigma(\pi(R))$**

Proof:

Step 1: we see that $\pi(R) \in B$: For any x in R,

$(-\infty, x] = \bigcup\limits_{n \in N} (-\infty, x + 1/n)$. So by the fact that B is closed under countable union of open sets, $\pi(R)$ is in B. 

Now, we need to show that every open subset G of R is in $\sigma(\pi(R)$. Every such G is a countable union of open intervals. we need only show that given a and b in R, a < b: $(a,b) \in \sigma(\pi(R)$.

step 2: For any u > a, we have $(a,u] = (-\infty, u] \cap (-\infty, a]^c \in \sigma(\pi(R)$. This follows from the properties of $\sigma(\pi(R)$ - that it contains complements and finite intersections, and the fact that $(-\infty, u] \in \pi(R)$.

step 3: Using this with b, we find:

$(a,b) = \bigcup\limits_n (a, b-\frac{\epsilon}{n}] \in \sigma(\pi(R), \text{ given } \epsilon = 1/2(b-a)$

So we conclude that every open set is in $\sigma(\pi(R))$. Since B(R) is the smallest sigma algebra containing open sets, and every set in $\pi(R)$ is creatable as a countable union of open sets and hence in B anyway (i.e. there are no extra sets in $\pi(R)$ which would not be in B anyway), the two are equal.


# Definitions concerning Set Functions

Let S be a set, $\Sigma_0$ an sigma-algebra and $\mu_0: \Sigma_0 \to [0, \infty]$ be a non-negative set function.

$\mu_0$ is said to be

**additive** 

If $\mu_0(\emptyset) = 0$ and given disjoint F,G in $\Sigma_0$, $\mu_0(A \cup B) = \mu_0(A) + \mu_0(B)$

**countably additive**

If $\mu_0(\emptyset) = 0$ and given $\{F_n : n \in N \}$ of disjoint sets in $\Sigma_0$, and given $F = \bigcup\limits_{n \in N} F_n$ is in $\Sigma_0$

$\mu_0(\bigcup\limits_{n \in N} F_n) = \sum\limits_{n \in N} \mu_0(F_n)$

Note the subtlety: This needs to be true only if the union F is in $\Sigma_0$ - in general this may not be true, though it happens to be true in $\sigma$-algebras.

A countably additive set function is additive (because we can just comple a finite sequence with countably infinite empty sets).


## Measure Space

Let $(S,\Sigma)$ be a measurable space, so that $\Sigma$ is a $\sigma$-algebra on S. A function:

$\mu: \Sigma \to [0, \infty]$ is called a **measure** on $(S,\Sigma)$ if $\mu$ is countably additive, and $(S,\Sigma, \mu)$ is called a **measure space**.

$\mu$ is called **finite** if $\mu(S) \lt \infty$, it's called $\sigma$-finite if there is a sequence $\{ S_n: n \in N \}$ of elements in S such that :

$\mu(S_n) \lt \infty, \forall n \in N$ and $\bigcup\limits_{n \in N} S_n = S$

**Probability Measure, Probability Triple**

Our measure $\mu$ is called a probability measure if $\mu(S) = 1$


**$\mu$-null element of $\Sigma$**

An element F of $\Sigma$ is called $\mu$-null if $\mu(F) = 0$.


**almost everywhere (a.e)**

A statement S about points in S is said to hold *almost everywhere (a.e.)* if:

$F := \{ s: s \in S, \text{ S is false } \} \in \Sigma$, and $\mu(F) = 0$

Note that F has to be measurable.

## LEMMA 1.6: Uniqueness of Extension, $\pi$-systems

LEMMA : a) Let S be a set. Let I be a $\pi$-system on S i.e. a family of subsets stable(closed) under finite intersection. Let $\Sigma = \sigma(I)$. Suppose $\mu_1$ and $\mu_2$ are measures on $(S, \Sigma)$ such that $\mu_1(S) = \mu_2(S) \lt \infty$, and $\mu_1 = \mu_2$ on I. 

Then $\mu_1 = \mu_2$ on $\Sigma$.

b) COROLLARY: If two probability measures on a $\pi$-system then they agree on the $\sigma$-algebra generated by that $\pi$-system.

Obviously this applies to $B = \sigma(\pi(R))$

## Theorem: Caratheodory's Extension Theorem

Let S be a set. Let $\Sigma_0$ be an **algebra** on S, and let:

$$\Sigma = \sigma(\Sigma_0)$$

Let $\mu_0$ be a **countably additive** map $\mu_0: \Sigma_0 \to [0, \infty]$, then there exists a measure $\mu$ on $(S, \Sigma)$ such that : 

$$\mu = \mu_0 \text{ on } \Sigma_0$$

What's more, if $\mu_0(S) \lt \infty$, then this extension is unique, by the uniqueness lemma mentioned earlier (an algebra is a $\pi$ system)

## Lebesgue Measure Leb on ((0,1], B(0,1])

Let $S = (0,1]$

For $F \subset S$, say that $F \in \Sigma_0$ if F may be written as a finite union :

$$(a_1, b_1] \cup \dots \cup (a_r, b_r]$$

where: $r \in N$, $0 \le a_1 \le b_1 \le \dots \le a_r \le b_r \le 1$.

Then $\Sigma_0$ is an algebra on (0,1] and :

$\Sigma = \sigma(\Sigma_0) = B(0,1]$

Now, we define : $\mu_0(F) = \sum\limits_{k \le r} (b_k - a_k)$

Then, $\mu_0$ is well-defined and additive on $\Sigma_0$. Moreover, $\mu_0$ is **countably additive** on $\Sigma_0$ (remember that this is only required where the countable, disjoint union of sets being added is in $\Sigma_0$).

Hence, by Caratheodory's theorem, there is a unique measure $\mu$ on ((0,1], B(0,1]), extending $\mu_0$ on $\Sigma_0$.

This is called the **Lebesgue** measure on ((0,1], B(0,1]), or more loosely on (0,1].

We can easily extend this to $[0,1]$ by adding set 0, with measure 0.

## $\pi$-systems and d-systems

A collection of subsets is a **pi-system** if it is closed under finite intersections. This implies it is also closed under finite unions. e.g. 

A collection of systems is a **d-system** or a **$\lambda$-system** if :

(i) it contains the universal set  
(ii) closed under complements  
(iii) closed under union of countably monotonically increasing collection of subsets   

An alternative restatement of (ii) is :

(ii) If A,B are in D and $A \subset B$, then $B - A \in D$

Proof this is equivalent given (i) and (ii):

Given A is subset of B, B^c and A are disjoint. Then their union is $B^c U A$ is in D. The complement (by ii) is in A too i.e. $B \cap A^c = B - A$ is in D. 

IN the other direction: Suppose If A,B are in D and $A \subset B$, then $B - A \in D$ is trivial:

Since A and D belong to D and A is a subset of D, D - A = $A^c$ belongs to D.

## Theorem: A collection of subsets $\Sigma$ is a sigma algebra iff it is both a $\pi$-system and a d-system

Proof(only if):

Given $\Sigma$ is a sigma algebra: It is obviously closed un finite intersections - so it is a pi-system.

It contains the universal set. It is closed under complements. Closed under all countable unions. So done.

Proof (if):

(i) $\Omega$ is in $\Sigma$ - since it is a d-system

(ii) Closed under complements: Let A be in $\Sigma$. Because $\Omega$ is in $\Sigma$ and $A \subset \Omega$ and $\Sigma$ is a d-system, it follows that $A^c = \Omega - A$ is in $\Sigma$.

(iiia) $\Sigma$ is closed under finite union

Given A and B: Being a $\pi$ system, $A \cap B \in \Sigma$

Being a d-system, it is closed under complements, so $A^c$ and $B^c$ belong to $\Sigma$ Then $A^c \cap B^c$ is in $\Sigma$, because $\Sigma$ is a $\pi$-system.

It follows that $A \cup B = (A^c$ and $B^c)^c$ is in $\Sigma$, because as a d-system it is closed under complements.

(iii) Now let $A_1, A_2,...$ be a countably infinite collection of sets. We can define $B_n = \bigcup\limits_{i=1}^{n} A_n$. **$B_n$ is in $\Sigma$ because we have shown in iiia that $\Sigma$ is closed under finite unions.**

Then the sets $B_i$ are a monotonically increasing countably infinite collection of sets. Since this is a d-system, it follows that $\cup B_i = \cup A_i \in \Sigma$. So $\Sigma$ is closed under countably infinite unions, and is therefore a sigma-algebra.

## d(C)

The dynkinisation of C : Given a collection C of subsets, d(C) is the smallest d-system containing C.

Let d(C) = intersection of all d-systems containing C.

Is it empty? No : $\Omega$ is a d-system containing C.

Is the intersection of d-systems a d-system? Obviously $\Omega$ is in D. If A is in in the intersection, so is $A^c$ so it is closed under complements. 

Given a countably monotonically increasing sequence of sets in the intersection, the union is also in the intersection. So the intersection is a d-system.



# Dynkin's Lemma

If I is $\pi$-system, then $d(I) = \sigma(I)$

Thus, any d-system that contains a $\pi$-system, contains the sigma-algebra contained by the pi-system.

Proof (is d(I) a pi-system?)

We want to show that d(I) is a pi system, i.e. given any A and B d(I), $A \cap B \in d(I)$. If we do this, we know that d(I) is a sigma-algebra by the lemma proved earlier.


**Step 1: Define "good sets" for an element of d(I)**

For every set X in d(I), define :

$D_X = \{ B \in d(I) : X \cap B \in d(I) \}$

Obviously we have $D_X \subset D(P)$

**Step 2: $D_A$ is a Dynkin System for any $A \in d(I)$**

a) I is in $D_A$, because $X \cap I = A$ is obviously in d(I).

b) Suppose B is in $D_A$. Then $B^c$ is in d(I) since d(I) is closed under complements. Is $A \cap B^c$ in $D_A$? 

Note that $A \cap B^c = A - (A \cap B)$. Since $A \cap B \in A$, it follows that the set difference $A - (A \cap B)$ is also in d(I) by property of dynkin systems. Hence $B^c$ is in $D_A$, i.e. $D_A$ is closed under complements.

c) Disjoint unions : If any countable collection of disjoint $B_n$ are in $D_A$, they are also in d(I) and hence their union must be in d(I). Given each of these $B_n$ intersect with A, it follows that each of $B_n \cap A$ are in d(I) and are also disjoint. Hence there union must be in d(I). So if $B = \cup B_n$, it follows that $B \cap A = \cup (B_n \cap A$) exists in d(I) and hence in $D_A$.

**Step 3: $I \subset D_A \text{ if } A \in I$**

For every B in I, $A \cap B \in I$ and hence also in d(I), and hence in $D_A$. We have already shown that A is in $D_A$ as well.

Hence $I \subset D_A$

Since $D_A$ is a dynkin system containing I, and d(I) is the smallest dynkin system containing I, it follows that $d(I) \subset D_A$. But $D_A \subset d(I)$, by definition. Hence $D_A = d(I)$

**Step 3: Now build $D_C$, where $C \in d(I)$**

Now we build $D_C$, where C is any element in d(I). Is $P \subset d(I)$?

If course, because we have already shown earlier that the intersection of every element of d(I) with an element of I is in d(I). So it follows that $I \subset D_C$. And because we have shown $D_C$ is a dynkin system, it follows that $d(I) \subset D(C)$ ie. $d(I) = D_C$.

**Step 3: d(I) is a $\pi$-system**
This means C intersects every element of d(I). But we have chosen C as an arbitrary element of d(I), which implies that following this process for each element of d(I), d(I) is closed under finite intersection. Hence d(I) is a $\pi-system$, and by the earlier lemma, it is the smallest $\sigma$-algebra containing I i.e. $d(I) = \sigma(I)$
