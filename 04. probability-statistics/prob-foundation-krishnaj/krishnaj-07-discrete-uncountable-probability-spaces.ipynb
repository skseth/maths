# Lecture 8 - Generated $\sigma$-algebra, Borel sets

https://www.youtube.com/watch?v=P_ROD0l0hfk&list=PLbMVogVj5nJQqGHrpAloTec_lOKsG-foc&index=8

**Generated sigma-algebra**

We want to generate a sigma algebra that contains all intervals. But such a sigma-algebra has to necessarily contain all complements.

$\Omega$ is an arbitrary sigma-algebra, and C is some arbitrary collection of subsets. We want to create the smallest sigma-algebra that contains C.

Theorem: For any such C, there is a smallest sigma-algebra which contains it, and called $\sigma(C)$ - the sigma algebra generated by the collection C.

Smallest => given any sigma algebra contains C, this sigma algebra contains $\sigma(C)$ 

Let $\mathscr{F}_i$, $i \in I$, be a collection of all sigma-algebras that contain C. Can it be empty? No. $2^{\Omega}$ is an element.

Homework: Arbitrary intersections of sigma algebras is a sigma-algebra

Is the intersection a sigma algebra? The empty set is in all sigma algebras, so it is in the intersection, as is $\Omega$ itself. So the intersection is not empty. If fact the intersection contains all elements of C. Now we want closure under countable union. Given any collection of sets $A_1, A_2, ...$ in the intersection, each element of the collection exists in each sigma-algebra (which is why it is in the intersection). It follows that the union of all these collections also exists in each sigma-algebra and hence also in the intersection. So the intersection is closed under countable union i.e. it is a sigma-algebra.

We define $\sigma(C) = \bigcap\limits_{i=I} F_i$. So $\sigma(C)$ is a sigma-algebra. Does it contain C? Yes. Because every sigma-algebra contains all elements of C, by our construction. Is it the smallest such sigma-algebra? 

Let H be the smallest sigma-algebra such that C is in H. Then H = $F_i$ for some i in I. But this means $\sigma(C) \subset H$ - given H was the smallest sigma-algebra containing C, this is only possible if $\sigma(C) = H$.


Conclusion: The concept of a smallest sigma-algebra is a well-defined concept. It could, if you have chosen C unwisely, be $2^{\Omega}$ itself. 


**Borel sigma-algebra**

Take (0,1] as a sample space. Take open sub-intervals of this sample space as $C_0$.

$\sigma(C_0)$ is called the Borel sigma-algebra denoted by $\mathscr{B((0,1])}$

Elements of $\mathscr{B((0,1])}$ are called Borel-measurable sets, or simply Borel sets.
