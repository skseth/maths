**Corollary: Any non-empty subset of R which is bounded above has a least upper bound**

**Theorem: The set N of positive integers is not bounded above (Archimedean property)**



**Theorem: Q is dense in R**

Suppose $0 \le x \lt y$. Then $y - x > 0$, hence there is an n, such that 
$n > \dfrac{1}{y - x}$. This means $1/n < y - x$.

Define $M = \{m \in N: m/n > x \}$. By the Archimedean property, M is not empty, otherwise nx would be an upper bound for n. 

Since $M \subset N$, it has a least element $m_0$. So $m_0/n > x$ and $(m_0 - 1)/n \le x$, which means $m_0/n \le x + 1/n$.

Then $x \lt m_0/n \le x + 1/n \lt x + (y - x) = y$

Now, we can take cases where $x \lt 0, y \gt 0$ (0 is in the middle), or $x \lt 0, y \lt 0$ (take -y, -x, and apply the result above)



**Between any two real numbers is an irrational number**

Let x and y be two real numbers, x < y. Then there are 2 rational numbers p,q s.t. $x < p < q < y$, by applying the theorem above twice.

Let $z = p + \frac{(q-p)\sqrt{2}}{2}$. Clearly z is between p and q and is irrational.
