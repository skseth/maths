# Vector Derivatives

$$\dfrac{d\mathbf{r}}{dt} = \lim\limits_{\Delta{t} \to 0} \dfrac{\Delta{\mathbf{r}}}{\Delta{t}}$$, where $\Delta{\mathbf{r}} = \mathbf{r(t + \Delta{t}} - \mathbf{r(t)})$

If f(t) is a scalar function, then :

$$\dfrac{d}{dt}f\mathbf{r} = f\dfrac{d\mathbf{r}}{dt} + \dfrac{df}{dt}\mathbf{r}$$


Usually, given $\mathbf{r} = x\hat{\mathbf{x}} + y\hat{\mathbf{y}} + z\hat{\mathbf{z}}$, we expect to apply the product rule as above, but because the unit vectors usually don't vary with time, their derivatives are zero. However, this is **not always the case**.

Secondly, we are assuming time itself to be a constant universally i.e. that all observers agree on time. This too fails at higher speeds.

## Force (Newton - N)

Use Newton's second law to define force. 1N force accelerates a standard 1 kg mass by $1m/sec^2$

Holding up a medium sized phone would need about a 1N force. Roughly you are counteracting gravity at $9.8m/sec^2$, so holding up about 100 gm of anything would take 1N of force.

Some other magnitudes:

- Lifting a 1kg bag - 10 N
- Lifting a human being - 500-800 N
- Pushing a small car from a dead stop - 5000 N
- Driving a car at 50mph into a wall - about 290,000 N


## Newton's Laws for a point mass of mass m

Law of inertia - no change in velocity without force

Second law: $\mathbf{F} = m\mathbf{a} = m\mathbf{\ddot{r}}$ (F = net force on particle, r - position vector of particle)

If we define momentum $\mathbf{p} = m\mathbf{v}$, then $\mathbf{F} = \dot{\mathbf{p}}$

These two statements are equivalent in classical mechanics, but need reformulation when relativity comes in.
  

## The Third Law and Conservation of Momentum

If object 1 exerts a force $\mathbf{F_{21}}$ on the object 2, the first exerts an equal but opposite force $\mathbf{F_{12}}$ on the second, so that $\mathbf{F_{21}} = -\mathbf{F_{12}}$.

This law is the key to conservation of momentum. 

Basically if there are two objects 1 and 2, exerting forces on each other and also subject to external forces, we get:

net force on 1 : $\mathbf{F_1} = \mathbf{F_1^{ext}} + \mathbf{F_{12}} = \mathbf{\dot{p_1}}$  
net force on 2 : $\mathbf{F_2} = \mathbf{F_2^{ext}} + \mathbf{F_{21}} = \mathbf{\dot{p_2}}$

Adding, we get the change in total momentum:

$\mathbf{\dot{p}} = \mathbf{\dot{p_1}} + \mathbf{\dot{p_2}} = \mathbf{F_1^{ext}} + \mathbf{F_2^{ext}} = \mathbf{F^{ext} \text{ the total external force } }$

Because the internal forces cancel. So if the external force on this 2-body system is zero, momentum is conserved. And the extension to any number of particles is straightforward.

Suppose we have n-particles:

$\mathbf{\dot{p}} = \sum\limits_{\alpha} \mathbf{p_{\alpha}} = \sum\limits_{\alpha}\sum\limits_{\beta > \alpha} \mathbf{F_{\alpha\beta}} + \mathbf{F_{\beta\alpha}} + \sum\limits_{\alpha} \mathbf{F_{\alpha}^{ext}} = F^{ext}$

Again because the sum $\sum\limits_{\alpha}\sum\limits_{\beta > \alpha} \mathbf{F_{\alpha\beta}} + \mathbf{F_{\beta\alpha}}$ is zero by the third law. Note that the summation trick is basically like taking take a square of size $N^2$, and removing the diagonal and upper triangle, so each pair is counted only once, and no particle exerts a force on itself.

The law of conservation of momentum carries over even into relativity and quantum mechanics.

**Failure of Third Law in Relativity**

The third law fails to hold in relativity - because effectively Newton's third law is saying that the action and reaction happen at the same time. But simultaenity breaks down in relativity for different observers - time is not the same for every one.

Strangely it doesn't just break down for gravity over long distances. Even objects touching each other are mediated by electromagnetic fields, and the instantaneous assumption actually breaks down.

Even more dramatically, the magnetic force between two moving charges does not obey the third law - in practice, the magnetic force is of the order of $v^2/c^2$ of the electrostaic force and hence often neglibible, nevertheless, this is a clear violation of the third law.



## Polar Coordinates

$x = r\cos(\phi), y = r\sin(\phi)$

$r = \sqrt{x^2 + y^2}, \phi = \arctan(y/x)$

$\mathbf{\hat{r}} = \dfrac{\mathbf{r}}{|\mathbf{r}|}$

$\mathbf{\hat{\phi}}$ is the unit vector of length 1 in the direction we move when r is fixed. Unlike the cartesian coodinate system, these vectors change position based on position.

Force: $\mathbf{F} = F_r\mathbf{\hat{r}} + F_\phi\mathbf{\hat{\phi}}$

We get:

$\mathbf{\Delta{\hat{r}}} \approx \Delta{\mathbf{\phi}} \mathbf{\hat{\phi}} \approx \dot{\mathbf{\phi}}\Delta{t} \mathbf{\hat{\phi}}$

Dividing and taking limits, we get the derivative:

$$\boxed{\dfrac{d\mathbf{\hat{r}}}{dt} = \dot{\mathbf{\phi}} \mathbf{\hat{\phi}}}$$
